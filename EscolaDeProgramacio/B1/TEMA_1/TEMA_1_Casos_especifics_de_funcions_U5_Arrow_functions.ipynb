# TEMA 1: Casos específics de funcions

## Unitat 5: Arrow functions

Les funcions de fletxa, o _arrow functions_ en anglès, són un tipus de funcions amb una sintaxi molt compacta que en moltes ocasions ens permeten mantenir el nostre codi més curt i llegible. En el codi següent veiem la mateixa funció utilitzant la sintaxi convencional i la versió _arrow function_. En tots dos casos la funció rep un número i retorna un valor booleà indicant si aquest número és parell.

```javascript
function isEven1(number) {
    return number % 2 === 0;
}

const isEven2 = (number) => {
    return number % 2 === 0;
}
```

Aquestes dues funcions són pràcticament equivalents, encara que tenen alguna petita diferència que analitzarem en unitats posteriors. En una funció de fletxa, a més, si el cos de la funció inclou únicament una línia que retorna un resultat, es pot simplificar la sintaxi eliminant tant les claus como la paraula `return`. Així mateix, si la funció té un únic paràmetre d'entrada, és possible eliminar els parèntesis que l'envolten. Unint totes dues coses, la nostra _arrow function_ quedaria així:

```javascript
const isEven3 = number => number % 2 === 0;
```

En aquests exemples estem assignant les funcions de fletxa a unes constants (`isEven2` i `isEven3`) per mostrar que és possible assignar un identificador a aquestes funcions per al seu ús posterior, però en realitat aquesta assignació **no forma part de la funció de fletxa** en si mateixa, que consisteix estrictament en aquesta part:

```javascript
number => number % 2 === 0
```

Podem llegir aquesta funció com a "donat un número qualsevol, al qual anomenarem `number` i que rebrem com a paràmetre, retorna un valor booleà (cert o fals) que indiqui si aquest número és divisible de manera sencera entre 2".

En aquest altre exemple, a continuació indiquem "donada una cadena de text qualsevol, a la qual anomenarem `password` i que rebrem com a paràmetre, retorna un nombre enter que ens informi de la longitud d'aquesta cadena de text"_:

```javascript
password => password.length
```

Aquesta sintaxi especialment compacta per definir una funció resulta molt útil quan volem utilitzar aquesta funció dins d'una altra funció, com vam veure en el cas dels mètodes `map()` i `filter()` estudiats en la primera unitat d'aquest tema. Recuperem per un moment un codi que ja vam veure, utilitzant la sintaxi estàndard per definir funcions. Recordem que `lostNumbers` és un array definit com a `[4, 8, 15, 16, 23, 42]`.

```javascript
const lostNumbers = [4, 8, 15, 16, 23, 42];

function isEven4(number) {
    return number % 2 === 0;
}

console.log(lostNumbers.filter(isEven4));
```

Si optem per utilitzar la sintaxi compacta de funció de fletxa, reduirem el nostre codi de la manera següent:

```javascript
const isEven5 = number => number % 2 === 0;

console.log(lostNumbers.filter(isEven5));
```

Però podem anar més enllà i estalviar-nos el pas d'assignar la funció a una constant, ja que ja hem indicat que aquesta assignació no forma part de la funció de fletxa:

```javascript
console.log(lostNumbers.filter(number => number % 2 === 0));
```
