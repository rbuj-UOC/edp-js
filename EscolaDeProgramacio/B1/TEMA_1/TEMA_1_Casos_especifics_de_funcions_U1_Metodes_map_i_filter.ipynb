## Unitat 1: Mètodes `map` i `filter`

En els últims temes del nivell A2 vam tenir l'ocasió d'estudiar la creació d'arrays i d'objectes i vam incidir en el fet que, molt sovint, treballarem amb estructures que combinen aquests dos tipus de dades. Recuperem per un moment el codi d'un dels exemples que vam veure, on la constant `classroomStudents` (un array) emmagatzema un conjunt d'estudiants (objectes):

```javascript
// Declaring and initializing the array that represents the students in the classroom.
const classroomStudents = [
  {
      name: 'Carme',
      surname: 'Ibáñez Marchena',
      yearOfBirth: 1998,
      averageGrade: 8.3,
      repeater: false,
  },
  {
      name: 'Joan',
      surname: 'Espí Prats',
      yearOfBirth: 1999,
      averageGrade: 9.8,
      repeater: true,
  },
  {
      name: 'Maria',
      surname: 'Massats Perelló',
      yearOfBirth: 2003,
      averageGrade: 4.5,
      repeater: false,
  },
  {
      name: 'Ramon',
      surname: 'Cugat Llopis',
      yearOfBirth: 2000,
      averageGrade: 2.7,
      repeater: true,
  },
  {
      name: 'Anna',
      surname: 'Bernal Valls',
      yearOfBirth: 2002,
      averageGrade: 3.5,
      repeater: false,
  }
];
```

El bon ús d'arrays resulta una habilitat imprescindible per a qualsevol programador. Ja hem vist com funcionen operacions com el recorregut de tots els elements d'un array i també la cerca d'un element en concret. No obstant això, hi ha més operacions que volem dur a terme sovint amb aquestes col·leccions, entre les quals destaquen la conversió dels elements de l'array en nous elements diferents i el filtrat o selecció d'elements. En aquesta unitat veurem com podem dur a terme aquestes operacions amb les funcions `map()` i `filter()`.

### Conversió d'elements d'un array en nous elements

En la creació dels nostres algorismes podem trobar-nos davant una situació en què disposem d'un array d'elements, però necessitem un altre array diferent, amb el mateix nombre d'elements que el primer però on els elements del nou array consisteixin en variacions dels elements de l'array original. Per exemple, potser disposem d'un array amb els nombres sencers `[4, 8, 15, 16, 23, 42]` i necessitem un altre array on els elements siguin els quadrats dels nombres originals: `[16, 64, 225, 256, 529, 1764]`. O tenim aquest mateix array de nombres sencers `[4, 8, 15, 16, 23, 42]` però necessitem un array diferent format per valors booleans que indiquin si cadascun dels elements és un nombre parell: `[true, true, false, true, false, true]`. 

Els tres conceptes clau que cal tenir en compte en aquestes situacions són els següents:

- Partim d'un array i volem obtenir un altre array **amb exactament el mateix nombre d'elements**.
- Els elements del nou array poden ser de qualsevol tipus, coincideixi o no amb el tipus d'elements de l'array original.
- Cada element del nou array generalment guarda algun tipus de relació amb cada element de l'array de partida. 

En JavaScript podem aconseguir-ho mitjançant l'ús del mètode `map()`. Per utilitzar aquest mètode necessitem disposar de l'array original i també d'una nova funció, definida per nosaltres, on establim la conversió que s'ha d'aplicar a cada element de l'array original per aconseguir cada element del nou array. Vegem-ho a partir d'un parell d'exemples. En el primer volem obtenir un array amb els noms dels estudiants de la classe:  

```javascript
// Function that receives a student object and returns the name of the student.
function studentName(student) {
    return student.name;
}

const studentNames = classroomStudents.map(studentName);

console.log(studentNames);
```

En aquest cas, la funció que passem com a argument al mètode `map()` és una funció que rep un paràmetre `student` i retorna el valor de la propietat `name` d'aquest objecte. Així, en cada iteració, el mètode `map()` executa aquesta funció `studentName()` passant-li com a argument l'element de l'array original corresponent a la iteració actual i afegeix al nou array el valor que retorna la funció, que no és altra cosa que el valor de la propietat `name` de l'objecte que es passa com a argument. 

Vegem ara en un altre exemple de quina manera podem aconseguir un array amb les notes dels estudiants de la classe, convertides a lletres en l'escala habitual entre A i D:

```javascript
// Function that receives a student object and returns their grade as a letter in the A-D range.
function letterGrade(student) {
  if (student.averageGrade >= 9.0) {
      return 'A';
  } else if (student.averageGrade >= 8.0) {
      return 'B';
  } else if (student.averageGrade >= 3.0) {
      return 'C';
  } else {
      return 'D';
  }
}

// Example of `map()` to convert an array of students into an array of letter grades.
const grades = classroomStudents.map(letterGrade);

console.log(grades);
```

En aquest codi, primer creem una funció `letterGrade()` que rep per paràmetre un estudiant (és dir, un objecte) i retorna una cadena de text amb la qualificació (A, B, C o D) que correspon a aquest estudiant en funció del valor de la seva propietat `averageGrade`. 

Qualsevol funció que vulguem utilitzar amb el mètode `map()` ha de tenir aquesta mateixa estructura: rep com a paràmetre un element del mateix tipus que tenen els elements de l'array original i retorna un nou element, ja sigui del mateix tipus o de qualsevol altre. 

Tot seguit, utilitzem l'array de partida per cridar el mètode `map()` passant-li la nostra funció `letterGrade()`.

> <br>🚨&nbsp; **Atenció**
> 
> Observeu que en l'exemple anterior `letterGrade` no va seguit de parèntesis, com és habitual quan utilitzem una funció. Això és així perquè, quan afegim els parèntesis després del nom d'una funció, estem cridant (o invocant o executant) aquesta funció. Però en aquest cas no ho volem pas fer, això. Només volem informar al mètode `map()` que `letterGrade()` és la funció que ha d'utilitzar per convertir uns elements en uns altres. Posteriorment `map()` sí que executarà aquesta funció tantes vegades com calgui (una per cada element de l'array original). Però nosaltres no invoquem de manera directa aquesta funció en cap moment.

> <br>🧐&nbsp; **Què significa _mètode_?**
> 
> Fins ara sempre hem parlat de funcions i, de sobte, hem començat a parlar de mètodes en referir-nos a `map()`. No obstant això, `map()` sembla comportar-se com una funció. Estem davant del mateix concepte? Són _mètode_ i _funció_ simplement sinònims? No exactament, però gairebé. Els mètodes són funcions associades a objectes. Més enllà d'aquest matís, que quedarà més clar més endavant, quan parlem de classes, de moment podeu considerar que els mètodes senzillament són funcions.

### Filtrat d'elements d'un array

Una altra de les operacions que volem dur a terme sovint sobre un array d'elements és el filtrat, que consisteix a seleccionar quins elements d'un array original ens interessen i quins preferim descartar. En l'apartat anterior hem vist que podíem convertir l'array de nombres sencers `[4, 8, 15, 16, 23, 42]` en un altre array diferent de valors booleans que indiquen si cadascun dels elements és un nombre parell: `[true, true, false, true, false, true]`. Potser, tanmateix, el que volem és filtrar l'array original per obtenir un nou array que inclogui només els nombres parells: `[4, 8, 16, 42]`. 

Els tres conceptes clau que cal recordar en aquest cas són els següents:

- Partim d'un array i volem obtenir un altre array **amb un nombre de d'elements entre zero i el nombre d'elements de l'array original**.
- Cada element del nou array és un element de l'array de partida, sense haver estat sotmès a cap modificació.
- El nou array inclou només els elements de l'array original que han superat el nostre filtre d'inclusió.

En JavaScript ho podem aconseguir amb l'ús del mètode `filter()`. Per utilitzar aquest mètode, com passava amb el mètode `map()`, necessitem disposar de l'array de partida i també d'una nova funció, creada per nosaltres, on establim el filtre que indica quins elements volem prendre i quins preferim descartar. Vegem l'exemple següent, en què volem obtenir un nou array amb els nombres parells de l'array original:

```javascript
const lostNumbers = [4, 8, 15, 16, 23, 42];

// Function that receives a number and returns a boolean indicating if that number is even.
function isEven(number) {
    return number % 2 === 0;
}

// Example of `filter()` to filter an array of numbers to take only even numbers.
const evenLostNumbers = lostNumbers.filter(isEven);

console.log(evenLostNumbers);
```

Observeu que hem creat una funció `isEven()` que rep per paràmetre un nombre i retorna un valor booleà (vertader o fals) que ens indica si el nombre rebut és parell. Qualsevol funció que vulguem utilitzar amb el mètode `filter()` ha de tenir aquesta mateixa estructura: rep per paràmetre un element del mateix tipus que tenen els elements de l'array original i retorna un valor booleà.
