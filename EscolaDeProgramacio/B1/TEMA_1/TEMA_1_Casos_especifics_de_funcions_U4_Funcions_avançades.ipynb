# TEMA 1: Casos específics de funcions

## Unitat 4: Funcions avançades

En aquesta unitat veurem dues operacions que fan servir la mateixa sintaxi (en concret, tres punts `...`) per dur a terme dues operacions diferents, encara que relacionades. En el primer cas utilitzarem aquesta notació per obtenir un nombre indeterminat de paràmetres en una funció, mentre que en el segon cas la utilitzarem per expandir els elements d'una col·lecció, perquè puguin utilitzar-se de manera individual.

### Paràmetres rest

Coneixem la sintaxi bàsica per definir una funció que rep una sèrie de paràmetres. Per exemple, podem definir una funció que rep dos nombres i en retorna el més gran:

```javascript
function max(number1, number2) {
    return number1 > number2 ? number1 : number2;
}
```

Però què passaria si volguéssim passar-li a una funció no dos, ni tres, ni quatre... sinó un nombre arbitrari i a priori desconegut de paràmetres? Amb la sintaxi vista fins ara no ens resulta possible, ja que hem d'assignar un nombre específic a cadascun dels paràmetres. No obstant això, en JavaScript existeixen els anomenats **paràmetres rest**, que, amb l'ús del prefix format per tres punts `...`, ens permeten que l'últim paràmetre d'una funció esdevingui un array que rep qualsevol nombre de paràmetres. Per exemple, podem definir una funció que rep un nombre indeterminat de paràmetres i en retorna el més petit:

```javascript
function min(...numbers) {
    let result = Infinity;

    for (let i = 0; i < numbers.length; i++) {
        if (numbers[i] < result) {
            result = numbers[i];
        }
    }

    return result;
}

console.log(min(1, 3));
console.log(min(1, 3, 5, -6, -45, 99));
console.log(min(1, 3, 5, -6, -45, 99, -100, 200, -123_456_000, 512));
```

En l'exemple anterior utilitzem un bucle per recórrer els diferents elements de `numbers`, ja que un paràmetre rest pren la forma d'un array que inclou tots els elements que s'hagin passat en la crida a la funció (dues en la primera crida, sis en la segona i deu en la tercera). En cas que no s'hagin passat paràmetres, l'array quedarà buit.

> <br>🚨&nbsp; **Atenció**
> 
> Només l'últim paràmetre d'una funció pot utilitzar `...` per rebre un conjunt de paràmetres. Sovint trobareu funcions en què el paràmetre que utilitza aquesta notació és, de fet, l'únic paràmetre de la funció. Però, en altres casos, la funció pot tenir definits dos o més paràmetres. En qualsevol cas, només pot existir un paràmetre rest i sempre ha de ser l'últim definit en la funció. Per exemple, `Math.max()` rep un nombre indeterminat de paràmetres, però aquest paràmetre rest és l'únic que rep aquesta funció.

> <br>💫&nbsp; **Ampliació**
> 
> Què signifiquen aquests caràcters `_` enmig d'una xifra? No res, la veritat. Són simples ajudes visuals per llegir millor els números que escrivim, però JavaScript els ignorarà. Quan treballem amb el número `100000000` no ens resulta senzill destriar de manera ràpida si són 10 milions, 100 milions, 1.000 milions... En canvi, quan veiem el mateix número amb les ajudes visuals `100_000_000` llegim gairebé de manera immediata 100 milions. Aquesta notació és vàlida per a qualsevol número; per exemple `1_000_000_000_000` és un trilió.

### Operador spread

L'operador spread, que també consisteix en tres punts `...`, ens permet separar els elements d'una col·lecció perquè puguin utilitzar-se de manera individual en parts del codi on s'espera rebre paràmetres simples. Vegem-ho a partir d'una funció que ens permet calcular potències com 2^8 (dos elevat a vuit) o 13^3 (tretze elevat a tres, o tretze al cub). Per fer-ho podem utilitzar la funció `Math.pow()` que rep dos paràmetres: la base i l'exponent. Però, si aquesta funció no existís, podríem crear-la nosaltres mateixos de la manera següent:

```javascript
function pow(base, power) {
    let result = 1;

    for (let i = 0; i < power; i++) {
        result = result * base;
    }

    return result;
}

let base = 2;
let power = 8;
console.log(pow(base, power));

base = 13;
power = 3;
console.log(pow(base, power));
```

Què passaria si en el nostre codi, en lloc de tenir les variables `base` i `power` tinguéssim els números en un array com `[2, 8]` o `[13, 3]`? Hauríem de cridar la funció `pow()` accedint de manera individual als elements de l'array:

```javascript
const array1 = [2, 8];
console.log(pow(array1[0], array1[1]));

const array2 = [13, 3];
console.log(pow(array2[0], array2[1]));
```

Encara que aquest codi funciona correctament, resulta un pèl feixuc i es tornarà molt repetitiu si la funció, en lloc de rebre dos paràmetres, en rep alguns més. Tot i així, gràcies a l'operador spread `...` podem passar a la funció `pow()` els diferents elements que formen un array, de manera que els rebi de manera individual:

```javascript
console.log(pow(...array1));
console.log(pow(...array2));
```

A primer cop d'ull podria semblar que només estem enviant un únic valor en cridar la funció `pow()`, quan aquesta funció en realitat té definits dos paràmetres. No obstant això, l'ús de l'operador `...` transforma un array en els seus elements considerats per separat. Com que els arrays que estem utilitzant en el nostre codi tenen dos elements, en realitat estem cridant la funció `pow()` enviant-li aquests dos valors individuals. En cas que l'array tingués tres elements, estaríem enviant tres valors individuals, i així successivament.

> <br>💫&nbsp; **Ampliació**
> 
> L'operador spread és d'especial importància quan permet dues operacions bàsiques d'ús molt habitual: copiar un array o clonar un objecte. Per copiar un array, n'hi ha prou amb separar primer els elements de l'array original amb l'ús de l'operador `...` i, tot seguit, tancar aquests elements ja separats en un nou array mitjançant l'ús de claudàtors `[]`. Com ara, `const arr2 = [...arr1]`. Sense l'ús d'aquest operador, hauríem de dur a terme un procés més llarg consistent en crear primer un nou array buit, utilitzar després un bucle convencional per recórrer els elements de l'array original i, en cada iteració, afegir l'element en qüestió en el nou array creat. L'ús de l'operador spread per copiar un array ens permet obtenir un codi molt més breu i simple. Per clonar un objecte actuem de la mateixa manera: primer separem les propietats de l'objecte amb `...` i immediatament després tornem a agrupar totes aquestes propietats separades mitjançant l'ús de claus `{}`. Per exemple, `const obj2 = {...obj1}`. L'alternativa sense aquest operador ens tornaria a exigir iterar totes les propietats de l'objecte i suposaria utilitzar un codi més llarg i augmentar la possibilitat d'introduir-hi algun error.
