## Unitat 4: Gestió de la asincronicitat

Una vegada plantejat el problema que ens obre l'asincronicitat i l'existència de promeses, vegem com podem gestionar-les de manera convenient. Partim d'un primer codi que connecta mitjançant `fetch()` l'API pública PokéAPI, que ens ofereix informació sobre pokémons. 

```javascript
function getDittoWrong() {
  let response = fetch('https://pokeapi.co/api/v2/pokemon/ditto');
  let data = response.json();
  console.log(data);
}
```

Aquesta funció pretén dur a terme això:

- Realitzar una petició a l'API i obtenir una resposta amb la informació del pokémon Ditto (primera línia).
- Analitzar aquesta resposta assumint que està en format JSON per obtenir un objecte de Javascript que representi Ditto (segona línea).
- Mostrar per pantalla l'objecte Javascript obtingut corresponent al pokémon Ditto (tercera línia).

Tanmateix, el codi no funciona. En primer lloc, perquè `fetch()` no retorna una resposta en forma d'objecte `Response`, sinó només una promesa en un objecte `Promise`. És cert que aquesta promesa més endavant es convertirà en una resposta real amb la informació corresponent al pokémon Ditto, però de moment només és una promesa. I les promeses (els objectes `Promise`) no tenen disponible el mètode `json()` com sí que el tenen les respostes (els objectes `Response`); per tant, el codi donarà error en intentar fer aquesta crida al mètode `json()`.

I a continuació tornem a tenir el mateix problema: el mètode `json()` d'un objecte `Response` no retorna un objecte de JavaScript, sinó només una promesa que, més endavant, es materialitzarà en l'objecte JavaScript. Per tant, si duem a terme alguna acció amb la variable `data` assumint que allà hi tenim dades reals cometrem un error, perquè en realitat només tenim una promesa, un rebut.

Perquè el codi funcioni podem optar per dues estratègies diferents. 

### Solució mitjançant `async` I `await`

Per solucionar aquests dos problemes podem demanar al nostre codi que, cada vegada que es trobi amb un mètode que retorna una promesa (objecte `Promise`), hi renunciï i sol·liciti esperar fins que hi hagi una resposta (objecte `Response`) disponible, trigui el temps que trigui. El codi anterior quedaria així:

```javascript
async function getDittoUsingAwait() {
  let response = await fetch('https://pokeapi.co/api/v2/pokemon/ditto');
  let data = await response.json();
  console.log(data);
}
```

Observem tres canvis importants respecte al codi equivalent anterior:

- En realitzar la petició a l'API amb `fetch()` li indiquem amb `await` que no volem una promesa, sinó que esperarem el temps que calgui fins a obtenir una resposta real.
- En analitzar la resposta mitjançant `json()` li indiquem amb `await` que tampoc aquí volem cap promesa, sinó que preferim esperar el temps que calgui fins a aconseguir una resposta real.
- En la declaració de la funció hem afegit la paraula clau `async`. Sempre que en el codi d'una funció utilitzem `await` almenys una vegada per esperar, en la declaració de la funció hem d'indicar-ho afegint-hi `async`. D'aquesta manera, la funció es converteix en una funció asíncrona, és a dir, una funció que pot esperar que es compleixin certes condicions abans de continuar amb la seva execució.

> <br>
> Seguint amb l'analogia del vestit i la tintoreria, utilitzar `await` equival a parlar amb la persona que ens atén i ens vol donar un rebut per dir-li que no cal, que no volem cap rebut: preferim asseure'ns i esperar tant de temps com calgui fins que el nostre vestit estigui rentat i planxat, llest perquè ens el puguem emportar.
> <br><br>

<br>

Pareu sempre atenció als diferents missatges d'error que trobareu quan escriviu codi font. Sovint contenen informació molt valuosa que ens orienta en la direcció adequada per resoldre el problema.

### Solució mitjançant `then`

Encara que la gestió de l'asincronicitat mitjançant `async` i `await` és l'opció més utilitzada i recomanable actualment, resulta convenient conèixer també l'alternativa mitjançant l'ús de blocs `then()`. Amb aquesta variant, el codi anterior quedaria d'aquesta manera:

```javascript
function getDittoUsingThen() {
  fetch('https://pokeapi.co/api/v2/pokemon/ditto')
      .then(response => response.json())
      .then(data => console.log(data));
}
```

Amb el primer bloc `then()`, diem a `fetch()` que no ens quedarem esperant fins que arribi la resposta, però que tampoc no tenim especial interès a obtenir una promesa. En lloc d'això, continuarem executant la resta del codi i, quan la resposta arribi finalment, volem que es dugui a terme una determinada acció que indiquem mitjançant una funció de fletxa: `response => response.json()`. En aquest cas, el que volem que es dugui a terme quan la resposta arribi és processar-la assumint que el contingut està en format JSON, per obtenir un objecte de JavaScript; per això utilitzem el mètode `json()`.

No obstant això, aquest mètode `json()` també és un mètode asíncron. És a dir, el contingut de la primera funció de fletxa inclou una crida a un mètode que es duu a terme de manera asíncrona i que, per tant, també retorna una promesa. Mitjançant una altra instrucció `then()`, la segona, indiquem que en aquesta ocasió tampoc no volem esperar ni gestionar nosaltres la promesa, sinó que preferim que, quan es materialitzi en una resposta concreta, es dugui a terme una altra acció que també indiquem amb una _arrow function_: `data => console.log(data)`. Això és, quan `json()` hagi acabat d'analitzar la resposta i l'hagi convertit en un objecte de JavaScript corresponent al pokémon Ditto, volem mostrar per consola les dades d'aquest pokémon rebut.

> <br>
> En el nostre exemple de la tintoreria, l'ús de `then()` equival a no esperar-nos fins que el vestit estigui llest (ja que pot ser una espera bastant llarga), però tampoc acceptar un rebut a canvi del nostre vestit. En lloc d'això, deixem indicacions clares sobre què volem que facin amb el nostre vestit quan estigui net i planxat. Per exemple, enviar-nos-el a l'hotel on ens allotjarem la propera setmana.
> <br><br>

<br>

> <br>💫&nbsp; **Ampliació**
> 
> Però exactament per què `fetch()` I `json()` retornen promeses, sempre més difícils de controlar, en lloc de retornar la informació que realment volem obtenir? Al cap i a la fi la resta d'instruccions de JavaScript que hem vist fins ara no es comportaven d'aquesta manera. Per què ho fan aquests mètodes? La clau de la resposta rau en el temps. Totes les instruccions apreses fins ara tenen una execució pràcticament immediata, per la qual cosa es processen unes rere altres sense demora. Això és important, perquè mentre s'està executant el codi és possible que l'usuari no pugui interactuar amb normalitat amb l'aplicació. Les operacions `fetch()` i `json()`, en canvi i igual que passa amb molts altres mètodes, poden demorar un temps significatiu. En el cas de `fetch()`, si el servidor al qual s'està realitzant una petició està molt sol·licitat, és possible que la resposta trigui uns segons a rebre's. El mètode `json()` ha d'analitzar un contingut en format JSON i convertir-lo en objectes de JavaScript. Si el contingut en qüestió són només unes desenes o centenars de línies, el procés serà gairebé immediat. Si, per contra, el contingut que s'ha de processar són desenes o centenars de milers de línies, novament caldran uns quants segons per obtenir-ne el resultat. Com que no resulta desitjable deixar l'aplicació tant de temps en un estat en què l'usuari no pugui utilitzar-la, aquests mètodes decideixen retornar una promesa de manera immediata i seguir executant-se en un segon pla fins que acabi la seva missió (obtenir resposta per part del servidor o generar un objecte de JavaScript a partir d'un text en format JSON). Quan això passa, la promesa es materialitza en una resposta concreta real. En el cas de `fetch()`, aquesta resposta real és un objecte `Response` que conté la resposta del servidor. En el cas de `json()`, la resposta real és un objecte de JavaScript amb les dades del pokémon Ditto. En tots dos casos, la resposta real és el que volem obtenir, però no podem de manera immediata. Per això, `fetch()` i `json()` retornen una promesa que es materializarà en la resposta desitjada quan estigui disponible.
