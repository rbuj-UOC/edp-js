## Unitat 3: Introducció a l'asincronicitat

Hem esmentat que AJAX és l'acrònim d'_Asynchronous JavaScript And XML_. El fet que aquest mecanisme sigui asíncron significa que quan l'aplicació web envia una petició al servidor no es queda esperant fins que rep una resposta, sinó que segueix executant les línies de codi posteriors, si n'hi ha.

Això suposa un avantatge clar: l'aplicació web no quedarà bloquejada mentre espera rebre una resposta que pot trigar un temps indeterminat, potser llarg, sinó que l'usuari en tot moment podrà continuar interactuant amb la pàgina mentre, en segon pla, l'aplicació continua esperant que la resposta arribi. Aspectes com la velocitat de connexió o la latència poden fer que des que la aplicació sol·licita la informació fins que la rep transcorri un temps perceptible per a l'usuari. Si l'aplicació es quedés bloquejada fins que rebés la resposta, l'usuari podria pensar que l'aplicació no funciona correctament i podria tancar-la o abandonar-la. Si, en canvi, l'aplicació continua funcionant mentre espera la resposta, l'usuari podrà continuar interactuant-hi i, quan la resposta arribi, l'aplicació podrà mostrar a l'usuari la informació que li ha sol·licitat.

Però aquest comportament també té un inconvenient important: quan escrivim el nostre codi, hem de tenir present que les línies posteriors a l'enviament de la petició no poden utilitzar la resposta que s'espera rebre (si tot va bé, unes dades concretes), perquè quan s'executin aquestes línies posteriors encara no s'haurà rebut la resposta amb les dades per part del servidor. El que sí que es rep de manera immediata és una **promesa** que més endavant es rebrà una resposta real. Aquesta promesa consisteix en un objecte de tipus `Promise`. Hi ha diferents llenguatges de programació que ofereixen solucions a aquest problema amb l'ús de conceptes com futurs, promeses, retards o diferits, que sovint s'utilitzen de manera similar, malgrat que la naturalesa de cadascun presenta algunes diferències.

Per solucionar aquest inconvenient tenim diferents possibilitats que analitzarem en la unitat següent, com l'ús de funcions _callback_, que és una manera d'especificar quina operació volem que es dugui a terme amb la resposta un cop arribi, o l'opció d'indicar al codi que volem esperar el temps que calgui fins que arribin les dades en lloc de continuar executant el codi posterior.

> <br>
> Podem entendre el concepte de _promesa_ a partir d'una analogia sobre el rebut que ens donen quan portem un vestit a la tintoreria. Aquest rebut podrem canviar-lo pel nostre vestit rentat i planxat en algun moment del futur, però ara mateix només és un tros de paper i no podem utilitzar-lo per vestir-nos. El rebut és, doncs, una promesa que més endavant tindrem el nostre vestit llest, però evidentment el rebut no és un vestit. Quan el rebut, al cap d'unes hores o uns dies, es converteixi en el vestit, podrem per fi utilitzar aquesta peça de roba per vestir-nos.
>
> Amb un objecte `Promise` passa el mateix. Quan enviem una petició a un servidor, obtenim de manera immediata una promesa que ens indica que més endavant tindrem dades disponibles, però ara mateix no tenim res de real. Si intentem dur a terme cap operació amb aquesta promesa segurament tindrem problemes, perquè es tracta només d'una garantia que més endavant tindrem dades, però en aquest moment només és això... una promesa, un rebut. Al cap d'uns centenars de mil·lisegons, la promesa es traduirà en unes dades reals que, ara sí, podrem utilitzar per mostrar nova informació a l'usuari de la nostra aplicació web.
