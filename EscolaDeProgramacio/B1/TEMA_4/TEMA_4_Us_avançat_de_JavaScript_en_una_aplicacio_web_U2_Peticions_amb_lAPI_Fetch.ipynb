## Unitat 2: Peticions amb l'API Fetch

Encara que `XMLHttpRequest` ha estat l'eina més utilitzada durant anys per realitzar peticions HTTP a servidors, el seu ús pot arribar a resultar un pèl feixuc, per la qual cosa amb el pas del temps van anar sorgint llibreries externes que en simplificaven l'ús en els escenaris más habituals. Una de les més utilitzades va ser jQuery. El 2015, tanmateix, Javascript va incorporar de manera nativa un nou mecanisme, l'API Fetch, pensat per succeir a `XMLHttpRequest` incorporant mecanismes com les promeses, que revisarem en la propera unitat, i que faciliten la petició de dades sense haver d'utilitzar funcions de crida de retorn (_callback_). Actualment, l'API Fetch ha esdevingut l'estàndard _de facto_ per realitzar peticions de manera asíncrona. En les següents unitats veurem com utilitzar-la.

Fetch és una API pròpia de Javascript (per tant, una part del llenguatge que podem utilitzar directament) que ens permet dur a terme el mateix tipus d'operacions que el mecanisme AJAX descrit en la unitat anterior, però d'una manera més senzilla i amb una sintaxi més fàcil de comprendre, que està aconseguint que cada vegada més programadors la utilitzin en detriment de `XMLHttpRequest`. 

El mateix exemple mostrat per a `XMLHttpRequest` quedaria així utilitzant l'API Fetch:

```javascript
async function logStarWarsPeople() {
  let response = await fetch('https://swapi.dev/api/people');
  let data = await response.text();
  console.log(data);
}

logStarWarsPeople();
```

És important notar la presència de les paraules clau `await` i `async`: la funció `fetch()` és una funció asíncrona que no retorna directament una resposta, sinó que retorna només una garantia, una promesa (és a dir, un objecte de tipus `Promise`) del fet que més endavant es rebrà una resposta. Analitzarem el concepte d'asincronicitat en la unitat següent. Per ara, n'hi ha prou amb saber que `await` és una paraula clau que indica que l'execució del codi es deté fins que la promesa es materialitza en una resposta.

La resposta que, tot i que no de manera immediata, ens acabarà retornant la funció `fetch()` és d'un tipus específic `Response` que té algunes propietats d'ús molt freqüent:

- `status` ens indica el codi numèric amb el resultat de la petició, per exemple un codi 200 si tot ha anat bé i tenim dades disponibles, o un codi 404 si no ha estat possible trobar el recurs.
- `statusText` ens indica el resultat textual de la petició, per exemple _OK_ per a un codi 200 o _Not Found_ per a un codi 404.
- `ok` ens indica, mitjançant un valor booleà, si el resultat de la petició ha tingut èxit (equivalent a comprovar si el codi numèric està en el rang entre 200 i 299).
- `text()` inclou una promesa (més detalls sobre promeses en la pròxima unitat) que es materializarà en les dades sol·licitades al servidor en format text.
- `json()` inclou una promesa que es materialitzarà en les dades sol·licitades al servidor en format JSON.
