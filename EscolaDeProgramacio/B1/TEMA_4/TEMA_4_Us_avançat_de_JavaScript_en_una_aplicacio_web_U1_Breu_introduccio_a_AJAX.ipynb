## Unitat 1: Breu introducció a AJAX

Quan un navegador web sol·licita un document HTML a un servidor, aquesta petició inicial sol desencadenar altres peticions: alguns arxius amb estils CSS, altres arxius amb codi JavaScript, sovint un bon nombre d'arxius d'imatges, etc. Totes aquestes peticions formen part del cicle de peticions HTTP i es produeixen quan accedim a un lloc web per primera vegada i també quan refresquem la pàgina mostrada en el navegador.

A més, el navegador web també pot realitzar peticions en segon pla per rebre o enviar dades sobre qualsevol tema: llistats de clients, registres meteorològics, resultats de partits de futbol o, en general, qualsevol tipus d'informació que el servidor hagi emmagatzemat. Quan diem que aquestes peticions es duen a terme en segon pla, el que volem dir és que no es produeixen com a conseqüència que l'usuari hagi refrescat la pàgina en el navegador, sinó que es poden iniciar des del codi JavaScript de l'aplicació web quan el programador hagi decidit fer-ho. Per exemple, quan hagin passat trenta segons des de la càrrega inicial de la pàgina o quan l'usuari premi un botó o passi el ratolí per sobre d'una imatge. En aquest cas, el navegador web no refresca la pàgina, sinó que realitza una petició HTTP al servidor per obtenir o enviar dades.

Històricament, el format XML va ser un dels més emprats per dur terme aquest intercanvi de dades. Això va fer que l'objecte que s'utilitzava per realitzar aquestes peticions de dades rebés el nom d'XMLHttpRequest o, senzillament, XHR.

Així doncs, quan la nostra aplicació web vol intercanviar dades amb un servidor, ja sigui per rebre-les o per enviar-les, pot crear un objecte `XMLHttpRequest` des de JavaScript per enviar una petició HTTP, és a dir per enviar una _HTTP request_ (d'aquí el nom de l'objecte). El servidor rep la petició, decideix què en vol fer i acaba enviant una resposta _HTTP response_ al navegador web, dins del mateix objecte `XMLHttpRequest`. El navegador rep aquesta resposta i la processa amb JavaScript per fer el que correspongui, que habitualment consistirà a actualitzar alguns elements de la pàgina web, per exemple manipulant elements del DOM amb les eines que hem vist en el tema anterior.

I tot això succeeix sense necessitat de refrescar la pàgina en el navegador, de manera transparent per a l'usuari, a través d'un mecanisme anomenat AJAX (acrònim d'_Asynchronous JavaScript And XML_). AJAX no és una funció, ni un mètode, ni una classe, ni un objecte... AJAX és un concepte que utilitzem per referir-nos a aquest mecanisme asíncron d'intercanvi d'informació amb un servidor utilitzant el codi JavaScript i els objectes `XMLHttpRequest`. En concret, el terme AJAX es va introduir el febrer del 2005 per tractar aquesta manera, aleshores revolucionària, d'obtenir contingut des d'una pàgina web utilitzant JavaScript sense necessitat de tornar a carregar la pàgina web.

Encara que la _X_ d'XMLHttpRequest fa referència al format XML, avui dia el format més habitual per intercanviar dades entre diferents aplicacions i màquines és el format JSON, que ja vam veure en un tema anterior. A tall de recordatori, JSON és l'acrònim de _JavaScript Object Notation_ i és, ras i curt, un format molt lleuger per desar i intercanviar informació. La seva sintaxi és molt bàsica: claudàtors `[ ]` per marcar arrays, claus `{ }` per delimitar objectes i parelles de clau i valor `"key": "value"` per definir totes les propietats dels objectes. Encara que el nom ens pot portar a l'error de creure que és un format exclusiu de JavaScript, la realitat és que JSON és un format estàndard per a l'intercanvi d'informació entre màquines que utilitzen llenguatges de programació molt diferents. Per exemple, JSON és el format utilitzat per l'API de Twitter per enviar i rebre informació.

Únicament a tall d'exemple ràpid, es mostra tot seguit un possible ús d'`XMLHttpRequest` per obtenir dades d'un servidor. En la unitat següent, tanmateix, estudiarem un mètode més actual per realitzar peticions i rebre respostes. En aquest cas, l'objecte `XMLHttpRequest` es crea amb la funció `new XMLHttpRequest()`, es configura amb el mètode `open()` i s'envia amb el mètode `send()`.

```javascript
function logStarWarsPeople() {
  // Create XMLHttpRequest object that we will use to interact with server to
  // retrieve data from the Star Wars API without having to refresh the page.
  const xhttp = new XMLHttpRequest()

  // Set type of request (usually GET or POST) and requested resource. The
  // `open` method takes three arguments: the type of request, the URL of the
  // resource, and a boolean indicating whether the request is asynchronous or
  // not. The default value for the third argument is true, which means that
  // the request is asynchronous. If you set it to false, the request will be
  // synchronous, which means that the code will stop executing until the
  // request is complete. This is not recommended because it will block the
  // main thread and make your application unresponsive. You should only use
  // synchronous requests if you absolutely need to.
  xhttp.open('GET', 'https://swapi.dev/api/people')

  // The `onload` property of the XMLHttpRequest object is a function that is
  // called when the request is complete. The `this` keyword refers to the
  // XMLHttpRequest object itself, so we can access the response data using
  // `this.responseText`.
  xhttp.onload = function () {
    // `this` here is the XMLHttpRequest object itself.
    console.log(this.responseText)
  }

  // The `send` method sends the request to the server. If the request is
  // asynchronous, the `onload` callback will be called when the response is
  // received. If the request is synchronous, the `send` method will block
  // until the response is received.
  xhttp.send()
}

logStarWarsPeople();
```

> L'exemple anterior inclou diversos conceptes que encara no hem revisat, com l'asincronicitat i els _callbacks_. Per defecte, té un valor `true` que fa que, si no informem aquest paràmetre, es comporti de forma asíncrona. Si volem que ho faci de manera síncrona, hem d'indicar `false`.
