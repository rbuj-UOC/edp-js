# TEMA 2: Classes

## Unitat 1: Definició de classes

Una classe no és més que una plantilla per crear objectes. En uns altres llenguatges de programació, les classes tenen una consideració una miqueta diferent, però en JavaScript es poden entendre com a simples motlles per crear objectes de manera més còmoda i estructurada. Es van introduir en el llenguatge en la revisió ECMAScript 2015, sovint denominada ES6.

Les classes permeten emprar en JavaScript el paradigma de programació denominat *programació orientada* a *objectes* (POO, o en anglès OOP, d'*object oriented programming*). Qualsevol programador experimentat coneix i aplica diversos paradigmes de programació, sovint sense adonar-se'n. El cas més simple és, potser, el model de programació **imperativa**, en què es construeix un programa mitjançant una seqüència explícita d'instruccions que s'executen en ordre. És el que succeeix quan executem un script senzill de JavaScript en el qual, línia a línia, s'indiquen tots i cadascun dels passos que el programa ha de seguir per aconseguir l'objectiu desitjat. En contraposició, en un estil de programació **declarativa** es declaren les propietats del resultat final desitjat, però no de quina manera s'ha d'aconseguir. És el que fem quan en el llenguatge SQL creem una instrucció complexa que exposa els resultats que pretenem recuperar d'una base de dades, sense indicar quin procés cal seguir pas a pas per extreure'ls. Altres paradigmes interessants són la programació **lògica**, la programació **funcional** o la programació **reactiva**. Tot i així, quan comencem a treballar amb classes, el paradigma que guanya protagonisme és l'orientat a objectes.

No obstant això, i com s'ha comentat anteriorment, JavaScript no té classes equiparables a les classes de la majoria de llenguatges de programació orientats a objectes. En JavaScript les classes són un simple mecanisme per facilitar la definició d'objectes. Això és així perquè, en realitat, JavaScript no té classes en el sentit convencional, sinó prototips. Els prototips són objectes que serveixen de base per a la creació d'altres objectes. En aquest sentit, les classes són un mecanisme que ens permet definir prototips.

> <br>💫&nbsp; **Ampliació**
> 
> Els diferents paradigmes no s'apliquen de manera excloent, sinò que es poden combinar per formar un paradigma compost. Per exemple, l'ús dels mètodes `map()` i `filter()` estudiats en la unitat anterior són clars exemples de programació funcional.
> <br><br>

La programació orientada a objectes gira, doncs, al voltant del concepte d'objectes que tenen associats tant dades com funcionalitat o comportament. En el cas de les dades, es tracta de les propietats dels objectes. La funcionalitat o comportament es defineix mitjançant funcions associades a l'objecte, que en aquest cas reben el nom de _mètodes_.

En els últims temes del nivell A2 vam tenir l'ocasió d'estudiar la creació d'objectes i vam veure aquest exemple:

```javascript
const randomUser = {
    name: "Albert",
    surname: "Mas",
    gender: "M",
    age: 23,
    hasBook: true,
};
```

Si hem de crear molts objectes amb aquesta mateixa estructura (per exemple, per crear un array amb tots els usuaris d'una biblioteca), resultarà còmode disposar d'una classe `User` que ens serveixi com a plantilla per crear aquests objectes, de tal manera que no haguem d'escriure els noms de totes les propietats una vegada rere l'altra i eliminem alhora la possibilitat d'errors, com definir en uns objectes la propietat `hasBook` i en uns altres `hasbook`. Així, la nostra classe `User` tindrà les propietats `name`, `surname`, `gender`, `age` i `hasBook`. I podem afegir-hi mètodes com `borrowBook()` o `returnBook()` per prendre un llibre en préstec o retornar-lo, respectivament.

El mètode `borrowBook()` pot, per exemple, rebre el títol del llibre que es vol prendre en préstec i encarregar-se de comprovar si l'usuari té cap altre llibre en préstec i, en cas afirmatiu, retornar un missatge d'error. Si no té cap llibre en préstec, pot comprovar si el llibre està disponible i, en cas afirmatiu, actualitzar l'estat del llibre i de l'usuari per indicar que l'usuari ara té aquest llibre en préstec.
