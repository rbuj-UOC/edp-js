# TEMA 2: Classes

## Unitat 2: Constructors i instanciació

Creem una primera versió de la nostra classe `User` amb el codi següent:

```javascript
class User {
    constructor(name, surname, gender, age, hasBook) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
        this.hasBook = hasBook;
    }
}
```

Conceptes importants que cal recordar:

- Utilitzem la paraula clau `class` per definir una classe i li donem un nom que, per convenció, comença amb una lletra majúscula. En aquest cas, `User`.
- Definim una funció amb el nom exacte `constructor()` i sense afegir-hi davant la paraula clau `function`. Ja hem esmentat que les funcions definides dins de classes reben el nom de _mètodes_, així que `constructor()` és, en realitat, un mètode. Però un mètode especial que s'executa només quan es crea un objecte a partir d'aquesta classe.
- El mètode `constructor()` rep com a paràmetres els valors amb què iniciarà les propietats que tindrien els objectes creats amb aquesta plantilla. En el cas de `User`, cinc propietats.

Una vegada creada la classe, podem crear objectes utilitzant-la d'aquesta manera:

```javascript
const albert = new User("Albert", "Mas", "M", 23, false);
const paula = new User("Paula", "Llop", "F", 18, true);
const manel = new User("Manel", "Cau", "M", 46, false);
const berta = new User("Berta", "Font", "F", 38, true);
const jaume = new User("Jaume", "Camps", "M", 61, false);
const users = [albert, paula, manel, berta, jaume];
```

En utilitzar la sintaxi `new User()` es produeix una crida automàtica al mètode `constructor()` que hem definit en la classe. Aquest procés rep el nom d'_instanciació_ i consisteix en la creació de nous objectes prenent la classe com a plantilla per assegurar que tots ells tenen les mateixes propietats i mètodes disponibles. En aquest cas, cadascun dels cinc objectes creats tindria les propietats `name`, `surname`, `gender`, `age` i `hasBook`, amb els valors que hem passat com a paràmetres al mètode constructor.

A continuació podem utilitzar l'operador `typeof` per esbrinar el tipus de qualsevol d'aquests usuaris creats mitjançant `new User()` i comprovarem que tots són de tipus `object`. De la mateixa manera, podem comprovar amb `hasOwnProperty()` que tots els objectes disposen de les propietats que esperem:

```javascript
// We created `randomUser` as a simple object using {...} syntax.
console.log("> Information about randomUser (created with {...} syntax)");
console.log("Type:", typeof randomUser);
console.log("Has surname property:", randomUser.hasOwnProperty("surname"));

// We created `albert` using our newly created `User` class.
console.log("> Information about albert (created with class)");
console.log("Type:", typeof albert);
console.log("Has surname property:", albert.hasOwnProperty("surname"));
```

És important remarcar que els objectes creats a partir d'una classe no són objectes especials, sinó que es comporten com la resta d'objectes. Seria possible obtenir exactament el mateix comportament utilitzant objectes creats de manera manual, però l'ús de classes ens facilita el procés. A més, les classes ens permeten definir mètodes que s'executaran en tots els objectes que se'n creiïn, la qual cosa ens permet estalviar codi i evitar errors. Ho veurem en unitats posteriors.
