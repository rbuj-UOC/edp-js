## Unitat 6: Serialització d'objectes amb JSON

A la llibreria de JavaScript trobem una sèrie d'objectes globals que tenim a la nostra disposició en tot moment per dur a terme tasques d'ús habitual. Entre aquests objectes hi ha `Math`, que ens permet realitzar algunes operacions matemàtiques, com trobar el màxim o el mínim entre una sèrie de nombres, `Date`, que utilitzem per al treball amb dates, i també `JSON`, que resulta molt útil a l'hora de treballar amb un format de text molt popular emprat per a l'intercanvi d'informació, anomenat, precisament, JSON. En aquesta unitat veurem com podem utilitzar aquest objecte per convertir objectes de JavaScript en cadenes de text i al contrari.

El format [JSON](https://www.json.org/json-en.html) és una notació pensada per a l'intercanvi d'informació entre aplicacions o entre màquines, que també resulta fàcil de llegir per a les persones. Encara que significa _JavaScript Object Notation_, en realitat és un format utilitzat en molts llenguatges de programació, no només en JavaScript. Tot i així, rep aquest nom perquè el format, com veurem a continuació, és pràcticament el mateix que utilitzem per a la representació d'objectes en JavaScript. Per exemple, si recuperem el primer usuari que vam veure en aquest tema:

```javascript
const randomUser = {
    name: "Albert",
    surname: "Mas",
    gender: "M",
    age: 23,
    hasBook: true,
};
```

La conversió al format JSON quedaria així:

> <br>
> {"name":"Albert","surname":"Mas","gender":"M","age":23,"hasBook":true}
> <br><br>

I si prenem l'array de `users` que hem definit como a `[albert, paula, manel, berta, jaume]`, on cada element és un objecte creat a partir de la classe `User` mitjançant la sintaxi `new User()`, la conversió a JSON queda d'aquesta manera:

> <br>
> [{"name":"Albert","surname":"Mas","gender":"M","age":23,"hasBook":false},{"name":"Paula","surname":"Llop","gender":"F","age":18,"hasBook":true},{"name":"Manel","surname":"Cau","gender":"M","age":46,"hasBook":false},{"name":"Berta","surname":"Font","gender":"F","age":38,"hasBook":true},{"name":"Jaume","surname":"Camps","gender":"M","age":61,"hasBook":false}]
> <br><br>

Observem, en tots dos casos, que la conversió al format JSON no ens retorna elements de JavaScript, sinó simplement text. Per això no hem utilitzat en aquests dos casos la tipografia que emprem habitualment per marcar el codi font. JSON és un format de text que ens permet representar objectes amb cadenes de text. Això ens permet poder emmagatzemar aquests textos en arxius o enviar-los a través d'internet entre diferents màquines sense perdre informació. Podríem fins i tot imprimir un contingut JSON en paper i enviar-lo per correu postal. JSON és un tipus de text que, fent ús d'una notació molt específica, ens permet emmagatzemar i recuperar objectes de JavaScript i, en general, de molts altres llenguatges de programació que també utilitzen objectes.

L'objecte global `JSON` esmentat al principi d'aquesta unitat ens permet, precisament, convertir un objecte de JavaScript a la notació JSON (és a dir, a un text) i també realitzar el procés invers: interpretar un text en format JSON per construir un objecte de JavaScript. Per fer-ho ens ofereix dos mètodes estètics: `stringify()` y `parse()`. El primer d'ells ens permet convertir un objecte de JavaScript a un text en format JSON, mentre que el segon ens permet convertir un text en format JSON a un objecte de JavaScript.

Mitjançant l'ús de `JSON.stringify()`, doncs, convertim un objecte o un array de JavaScript en un text que inclou la mateixa informació:

```javascript
class User {
    constructor(name, surname, gender, age, hasBook) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
        this.hasBook = hasBook;
    }
}

const albert = new User("Albert", "Mas", "M", 23, false);
const paula = new User("Paula", "Llop", "F", 18, true);
const manel = new User("Manel", "Cau", "M", 46, false);
const berta = new User("Berta", "Font", "F", 38, true);
const jaume = new User("Jaume", "Camps", "M", 61, false);
const users = [albert, paula, manel, berta, jaume];

const usersAsJson = JSON.stringify(users);
console.log(usersAsJson);
```

I, de manera inversa amb `JSON.parse()`, tornem a generar un objecte o un array de JavaScript a partir d'un text codificat en format JSON:

```javascript
const usersFromJson = JSON.parse(usersAsJson);
console.log(usersFromJson);
```

> <br>💫&nbsp; **Ampliació**

Aquesta informació diversa estarà emmagatzemada en diferents servidors que utilitzen diferents llenguatges de programació i bases de dades. Tot i així, quan la sol·licitem des del nostre codi JavaScript, els servidors convertiran primer la informació a format JSON i serà aquest text el que ens enviaran. Posteriorment utilitzarem mètodes d'anàlisi per convertir aquest text a objectes JavaScript, amb els quals ens resultarà més còmode treballar. L'ús de `JSON.stringify()` i `JSON.parse()` és molt habitual en el desenvolupament d'aplicacions web. Per exemple, quan vulguem enviar informació a un servidor, primer la convertirem a format JSON i després l'enviarem. Quan rebem la resposta, la tornarem a convertir en objectes JavaScript per poder-hi treballar.
