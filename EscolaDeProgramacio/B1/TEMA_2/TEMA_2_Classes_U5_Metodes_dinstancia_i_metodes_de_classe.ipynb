# TEMA 2: Classes

## Unitat 5: Mètodes d'instància i mètodes de classe

Les classes poden tenir mètodes d'instància i mètodes de classe. Els mètodes d'instància són aquells que s'apliquen a una instància de la classe (és a dir, un objecte). Els mètodes de classe són aquells que s'apliquen a la classe en si mateixa. En aquesta unitat veurem com definir i fer servir tots dos tipus de mètodes.

### Mètodes d'instància

Habitualment, les classes tenen definits un conjunt de mètodes d'instància que determinen el comportament dels objectes que s'hi creen. Podem entendre el concepte de comportament com el conjunt de les accions que un objecte pot dur a terme. Per exemple, en el cas de la classe `User`, hem esmentat anteriorment els mètodes `borrowBook()` o `returnBook()` per prendre un llibre en préstec o retornar-lo. Així queda la classe `User` un cop afegits aquests mètodes:

```javascript
class User {
    constructor(name, surname, gender, age, hasBook) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
        this.hasBook = hasBook;
    }

    borrowBook() {
        if (!this.hasBook) {
            this.hasBook = true;
            console.log(`${this.name} ${this.surname} borrowed a book`);
        } else {
            // Let's assume users can only have one book at a time.
            console.log(`${this.name} ${this.surname} couldn't borrow a book`);
        }
    }

    returnBook() {
        if (this.hasBook) {
            this.hasBook = false;
            console.log(`${this.name} ${this.surname} returned book`);
        } else {
            console.log(`${this.name} ${this.surname} has no book to return`);
        }
    }
}
```

Noteu que els mètodes, com succeeix amb `constructor()`, tampoc empren la paraula clau `function`, encara que la resta de la seva sintaxi sigui idèntica. Per invocar aquests mètodes, hem de fer-ho sobre un objecte:

```javascript
const albert = new User("Albert", "Mas", "M", 23, false);

albert.returnBook();
albert.borrowBook();
albert.returnBook();
albert.borrowBook();
albert.borrowBook();
```

### Mètodes de classe

Una classe també pot tenir definits mètodes que no estiguin associats als objectes que se'n creen, sinó directament a la classe. En el cas de `User`, podem pensar en un mètode `genderInfo()` que ens doni informació sobre els possibles valors que pot prendre la propietat `gender` en cada objecte creat. Aquest mètode, però, no necessita estar associat a cada objecte, ja que per a tots ells retornarà exactament el mateix text. Té més sentit que es tracti d'un mètode associat directament a la classe. Per aconseguir-ho, utilitzem la mateixa sintaxi que per als mètodes d'instància, però afegint-hi la paraula clau `static`:

```javascript
class SpecialUser {
    constructor(name, surname, gender, age, hasBook) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
        this.hasBook = hasBook;
    }
    
    // Same code used in previous block code:
    // borrowBook() { ... }
    // returnBook() { ... }

    static genderInfo() {
        return "Users can have three values for `gender`: M for male, F for female, and NB for non binary";
    }
}

console.log(SpecialUser.genderInfo());
```

Per invocar els mètodes de classe, tanmateix, no utilitzem un objecte, sinó que fem la crida directament sobre la classe. En el codi anterior no accedim a `albert.genderInfo()`, sinó directament a `SpecialUser.genderInfo()` (hem canviat el nom de `User` per `SpecialUser` només per evitar problemes amb definicions duplicades). El mètode `genderInfo()` no té accés a les propietats dels objectes creats a partir de la classe, per la qual cosa no pot retornar informació sobre el gènere d'un usuari en concret. En canvi, pot retornar informació genèrica sobre els possibles valors que pot prendre la propietat `gender` en cada objecte creat.

Tot i que molt útils en alguns escenaris, els mètodes de classe tenen un paper secundari en la programació orientada a objectes. Els mètodes d'instància, en canvi, suposen una peça estructural d'aquest tipus de programació, així que ens els trobarem i els definirem amb molta més freqüència.

> <br>💫&nbsp; **Ampliació**
> 
> Com hem vist anteriorment, les propietats d'una classe es poden marcar com a privades afegint el prefix `#` al seu nom. Passa el mateix amb els mètodes. Si el mètode `borrowBook()` no ha de ser accessible des de fora de la classe, podem marcar-lo com a privat afegint el prefix `#` al vostre nom. Si ho fem, el codi `albert.#borrowBook()` generarà un error, perquè estarem intentant invocar un mètode privat des de fora del codi interior de la classe. L'únic mètode que no es pot marcar com a privat és `constructor()`. No obstant això, el nostre exemple `borrowBook()` no ha de ser privat, ja que és un mètode que volem que els usuaris de la classe `User` puguin invocar. Podeu pensar en algun mètode que es pogués afegir a la classe `User` i que sí que podria o hauria de ser privat?
