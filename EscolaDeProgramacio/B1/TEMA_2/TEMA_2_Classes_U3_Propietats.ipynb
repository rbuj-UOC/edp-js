# TEMA 2: Classes

## Unitat 3: Propietats

En la classe `User` definida anteriorment existeixen cinc propietats creades de manera directa a partir del mètode constructor: `name`, `surname`, `gender`, `age` i `hasBook`. Tanmateix, podem afegir-hi noves propietats calculades amb l'ús de la paraula clau `get`. Vegem-ho amb un exemple, que explicarem tot seguit:

```javascript
class User {
    constructor(name, surname, gender, age, hasBook) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
        this.hasBook = hasBook;
    }

    // Getter.
    get fullName() {
        return `${this.name} ${this.surname}`;
    }

    // Getter.
    get anonymousDescription() {
        return `${this.age}${this.gender}`;
    }
    
    // Getter. Users don't have a real favorite number, they say a different one each time!
    get favoriteNumber() {
        return Math.round(Math.random() * 100);
    }
    
    // Setter.
    set fullName(name) {
        const nameParts = name.split(' ');
        this.name = nameParts[0];
        this.surname = nameParts[1];
    }
}
```

L'anotació `get` ens permet definir una funció que externament es percebrà com una propietat. Aquesta funció, que no ha de rebre cap paràmetre, pot dur a terme qualsevol tipus de càlcul i retornar un resultat. Per cridar-la no utilitzarem la notació típica de les funcions o mètodes, `albert.fullName()`, sinó la que utilitzem per accedir a les propietats: `albert.fullName` (sense els parèntesis). Anomenem *getters* a aquesta mena de funcions, a les quals accedim com si fossin propietats per recuperar (_get_) un valor.

En la classe que acabem de definir, el primer `fullName` amb el prefix `get` ens permet obtenir el nom complet d'un usuari sense haver d'accedir cada vegada al nom de pila i al cognom. De la mateixa manera,`anonymousDescription` ens dona una descripció del tipus "28M" (per indicar 28 anys home) o "19F" (19 anys dona) que junta edat i gènere de manera abreujada. Mitjançant un tercer _getter_ podem generar una propietat que retorni un número a l'atzar entre 0 i 100 cada vegada que s'hi accedeixi.

Anàlogament, l'anotació `set` ens permet definir una funció de manera que externament es percebi com una propietat. Aquesta funció sí que rep un paràmetre, que s'utilitzarà per assignar el valor a una o més propietats de la classe. Denominem _setters_ a aquesta mena de funcions, a les quals accedim com si fossin propietats per establir (_set_) un valor.

En la classe `User`, accedint a la propietat `fullName` i assignant-li un nou valor, podem canviar internament les propietats `name` i `surname` del nostre objecte, assumint que el nom serà allò que aparegui abans del primer espai i el cognom allò que aparegui després.

Vegem-ho amb l'exemple següent, en el qual l'objecte `albert` té inicialment el nom Albert Mas i després passa a anomenar-se Miquel Martí (és evident que, si el nostre codi permetrà aquest tipus de canvi de noms, assignar el nom `albert` a l'objecte resulta una mala elecció):

```javascript
const albert = new User("Albert", "Mas", "M", 23, false);

console.log(albert.fullName);
albert.fullName = 'Miquel Martí';
console.log(albert.name);
console.log(albert.surname);
console.log(albert.fullName);
```

> <br>💫&nbsp; **Ampliació**
> 
> Tant _getters_ como _setters_ guanyen especial rellevància quan treballem amb propietats privades. Les propietats privades a JavaScript són un concepte crucial per a l'ocultació de dades dins de les classes, assegurant que certa informació sigui accessible i modificable únicament a través de mètodes específics definits a la classe. En marcar una propietat com a privada, es restringeix l'accés directe des de fora de la classe, permetent un control més estricte sobre com s'accedeix i es modifica l'estat intern d'un objecte.
>
> En concret, per declarar una propietat privada a una classe de JavaScript s'utilitza el prefix `#`. Aquest prefix fa que la propietat sigui accessible només des de mètodes definits dins de la mateixa classe.
>
> Quan és útil l'ús de propietats privades? En general, sempre que vulguem assegurar-nos que certes dades només puguin ser modificades a través de mètodes específics de la classe. Per exemple, si estem desenvolupant una classe `User` i volem assegurar-nos que el nom d'un usuari hagi de tenir almenys tres caràcters, podríem definir una propietat privada `#name` i un _setter_ `set name(name)` que comprovi la longitud del nom abans d'assignar-lo a la propietat privada. D'aquesta manera, garantim que el nom d'un usuari sempre compleixi certes condicions, evitant que se li assignin valors incorrectes. També podríem controlar que el nom d'un usuari no pugui contenir espais en blanc, de manera que si s'intenta assignar un nom amb espais, el _setter_ el rebutgi i no s'assigni el valor a la propietat privada.
>
> Les propietats privades han de ser declarades dins de la definició de la classe abans de ser utilitzades en mètodes, inclòs el mètode constructor. En general, es declaren al principi de la classe, just després de la declaració de les propietats públiques en cas que n'hi hagi. Cal fer-ho així perquè la declaració de la propietat privada no consisteix simplement en una assignació d'un valor, sinó que forma part de la definició de l'estructura de la classe.
