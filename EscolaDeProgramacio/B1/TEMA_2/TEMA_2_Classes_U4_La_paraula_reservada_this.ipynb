## Unitat 4: La paraula reservada `this`

La paraula reservada `this` té un significat especial en molts llenguatges de programació, especialment als que segueixen el paradigma de l'orientació a objectes. En Javascript, no obstant això, té un funcionament un pèl peculiar que depèn del context en què apareix, ja sigui a l'interior d'una classe, en el codi d'una funció o simplement en el que es coneix com a context global (fora de qualsevol funció o classe). Com a primera definició bàsica, podem considerar que `this` fa referència a l'objecte en què s'està executant el codi actual. Sovint això farà que `this` es refereixi a l'objecte global.

Quan ens referim a l'objecte global, estem parlant d'un espai en què estan disponibles algunes variables i funcions a les quals podem accedir des de qualsevol part del nostre codi. En el cas del codi Javascript d'una pàgina web, l'objecte global és per defecte la finestra (un objecte de tipus `Window`) a la qual podem accedir en tot moment mitjançant una referència a `window`. En el cas de Node.js, l'objecte global és l'objecte `global`. En qualsevol cas, l'objecte global és el que conté totes les variables i funcions que no estan definides dins de cap altre objecte.

Quan `this` apareix dins del cos d'una funció, podem imaginar que `this` és en realitat un paràmetre més d'aquesta funció (un paràmetre implícit o ocult), de manera que, si invoquem aquesta funció a través d'un objecte, `this` hi quedarà associat. Comprovem-ho amb un exemple:

```javascript
// This function just returns current value of `this`.
function whoIsThis() {
    return this;
}

// We create two basic objects here.
const object1 = { description: "I'm object 1" };
const object2 = { description: "I'm object 2" };

// And add a new property to each object.
object1.specialPropertyToCheckThis = whoIsThis;
object2.specialPropertyToCheckThis = whoIsThis;

console.log(object1.specialPropertyToCheckThis());
console.log(object2.specialPropertyToCheckThis());
```

> <br>🎯&nbsp; **Vist anteriorment**
> 
> En els últims temes del nivell A2 vam aprendre que, un cop creat un objecte, podem afegir-li propietats mitjançant, per exemple, la notació de punt. És el que estem fent aquí amb `object1.specialPropertyToCheckThis`. L'objecte no té aquesta propietat: la hi afegim en aquest moment assignant-li la funció definida prèviament.
> <br><br>

Observem que la primera crida a `specialPropertyToCheckThis()` la duem a terme sobre l'objecte `object1` i ens retorna un objecte amb la descripció _"I 'm object 1"_, mentre que la segona crida la fem sobre l'objecte `object2` i ens retorna un objecte amb la descripció _"I 'm object 2"_. Això és així perquè `this` es refereix a l'objecte que invoca la funció, en aquest cas `object1` i `object2`, respectivament.

En altres paraules, hem definit una funció `whoIsThis()`, que té com a única tasca `return this` per permetre'ns identificar amb què està associat `this` en aquest context. Després hem afegit aquesta funció com a propietat a dos objectes diferents. Finalment, hem cridat la funció a partir d'aquestes propietats d'objectes i observem que en cada cas ens informa que `this` és l'objecte concret sobre el qual s'ha cridat.

Això resulta especialment rellevant en el cas de les classes. Com hem dit, una classe no és més que una plantilla per crear objectes de manera més fàcil i estructurada. Aquesta plantilla pot incloure la definició d'algunes funcions, que denominarem mètodes (més informació sobre aquest punt en la unitat següent). Dins d'aquests mètodes, la paraula clau `this` identificarà en cada cas l'objecte concret sobre el qual s'invoca el mètode. Així, si tenim una classe `User` i n'hem creat deu objectes (instàncies), cada vegada que invoquem un mètode de la classe sobre cadascun d'aquests deu objectes, la paraula `this` guardarà una referència a l'objecte específic utilitzat per fer la crida. En l'**Anex 1: Exemple complet d'ús de classes** podrem veure un exemple que empra aquesta característica.

### Cas especial: arrow functions

En el tema anterior vam tenir l'ocasió d'estudiar les funcions de fletxa o _arrow functions_ i vam esmentar que les funcions definides de la manera habitual, mitjançant l'ús de la paraula clau `function`, i les funcions de fletxa són pràcticament equivalents, encara que amb alguna petita diferència que ara per ara decidim deixar pendent per a una explicació posterior.

Una de les diferències té a veure, precisament, amb el valor de `this`. Si dupliquem el codi de l'exemple anterior però utilitzem ara una _arrow function_ en lloc d'una funció definida de la manera convencional podrem comprovar-ho:

```javascript
// This function also just returns current value of `this`.
const arrowWhoIsThis = () => this;

// We create two basic objects here.
const object3 = { description: "I'm object 3" };
const object4 = { description: "I'm object 4" };

// And add a new property to each object.
object3.specialPropertyToCheckArrowThis = arrowWhoIsThis;
object4.specialPropertyToCheckArrowThis = arrowWhoIsThis;

console.log(object3.specialPropertyToCheckArrowThis());
console.log(object4.specialPropertyToCheckArrowThis());
```

La nova funció de fletxa, que per comoditat assignem a una constant de nom `arrowWhoIsThis`, realitza la mateixa acció que la funció anterior `whoIsThis()`: retornar el contingut de `this` en el seu context. Per definir-la necessitem utilitzar una parella de parèntesis `()` per indicar que la indicar que la funció no rep cap paràmetre, exactament com passa amb la funció `whoIsThis()` definida de la manera més convencional.

No obstant això, quan executem el codi observem que, mentre que en l'exemple anterior les dues instruccions `console.log()` mostraven els objectes `object1` i `object2`, en aquest últim codi les dues últimes instruccions `console.log()` no mostren aquests mateixos objectes, sinó uns objectes globals que inclouen un gran nombre de propietats. 

Això és així perquè, en una funció de fletxa, `this` manté el valor que tenia en el moment de definir-la. En aquesta mena de funcions, JavaScript no enllaça per defecte `this` amb l'objecte a partir del qual s'invoca la funció. Per tant, si definim una funció de fletxa en l'àmbit global, `this` es mantindrà associat a l'objecte global, que és l'objecte que conté totes les propietats i mètodes que estan disponibles en l'àmbit global. En el cas d'un navegador, aquest objecte global és `window`.
