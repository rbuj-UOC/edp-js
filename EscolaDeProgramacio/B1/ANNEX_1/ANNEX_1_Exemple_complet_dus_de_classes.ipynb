# ANNEX 1: Exemple complet d'ús de classes

En el tema 2 hem tingut l'ocasió d'estudiar la creació i ús de classes en JavaScript. En aquest annex mostrem un cas complet d'ús de diferents classes per a la gestió d'usuaris i llibres en una biblioteca.

```javascript
class User {
    // Private field - array of books this user has on loan at this moment.
    #books = [];

    constructor(name, surname, gender, age) {
        this.name = name;
        this.surname = surname;
        this.gender = gender;
        this.age = age;
    }

    // Add a book only if book is not already included.
    addBook(book) {
        if (!this.#books.includes(book)) {
            // Using spread operator as part of the strategy to copy an array.
            this.#books = [...this.#books, book];
        }
    }

    // Remove a book only if book has previously been included.
    removeBook(book) {
        if (this.#books.includes(book)) {
            // Instead or removing an element, we can create a new array using `filter`.
            this.#books = this.#books.filter((b) => b !== book);
        }
    }

    // Getter - calculated property.
    get numberOfBooksCurrentlyOnLoan() {
        return this.#books.length;
    }

    // Getter - calculated property.
    get fullName() {
        return `${this.name} ${this.surname}`;
    }
}

class Book {
    #currentUser = null;
    #totalTimesLent = 0;

    constructor(title, author, year) {
        this.title = title;
        this.author = author;
        this.year = year;
    }

    // Increase number of times lent and set current user.
    lendToUser(user) {
        this.#totalTimesLent++;
        this.#currentUser = user;
    }

    // Set current user to `null` so we know book is not currently on loan.
    markAsAvailable() {
        this.#currentUser = null;
    }

    // Getter - property.
    get currentUser() {
        return this.#currentUser;
    }

    // Getter - property.
    get totalTimesLent() {
        return this.#totalTimesLent;
    }
}

class Library {
    // Private field - arrays of users and books.
    #users = [];
    #books = [];

    constructor(name, address) {
        this.name = name;
        this.address = address;
    }

    // Add user to library only if user is not already included.
    addUser(user) {
        if (!this.#users.includes(user)) {
            this.#users = [...this.#users, user];
            console.log(`User ${user.fullName} can now borrow books from this library`);
        }
    }

    // Add book to library only if user is not already included.
    addBook(book) {
        if (!this.#books.includes(book)) {
            this.#books = [...this.#books, book];
            console.log(`Book '${book.title}' by ${book.author} added to library catalogue`);
        }
    }

    // Lend book to user only if both book and user are valid and book is not already on loan.
    lendBook(book, user) {
        // We could replace `book.currentUser === null` with `!book.currentUser` here. Try it yourself!
        if (this.#books.includes(book) && this.#users.includes(user) && book.currentUser === null) {
            user.addBook(book);
            book.lendToUser(user);
            console.log(`Book '${book.title}' lent to ${user.fullName}`);
        }
    }

    // Return book only if book is valid and currently on loan.
    returnBook(book) {
        if (this.#books.includes(book) && book.currentUser !== null) {
            const user = book.currentUser;
            user.removeBook(book);
            book.markAsAvailable();
            console.log(`Book '${book.title}' returned by ${user.fullName}`);
        }
    }

    // Generate a text report to show library history and state.
    report() {
        console.log(`---------------------- ${this.name} ---------------------\n`);
        this.#books.forEach((book) => console.log(`  📕 Book '${book.title}' has been lent ${book.totalTimesLent} times`));
        console.log('');

        this.#users.forEach((user) =>
            console.log(`  👤 User ${user.fullName} has ${user.numberOfBooksCurrentlyOnLoan} books on loan`)
        );
        console.log('');
        console.log('--------------------------------------------------------------');
    }
}

// Create some initial books.
const quijote = new Book('Don Quijote de la Mancha', 'Miguel de Cervantes', 1605);
const rings = new Book('The Lord of the Rings', 'J. R. R. Tolkien', 1954);
const kr = new Book('The C Programming Language', 'Brian Kernighan & Dennis Ritchie', 1978);
const books = [quijote, rings, kr];

// Create some initial users.
const albert = new User('Albert', 'Mas', 'M', 23);
const paula = new User('Paula', 'Llop', 'F', 18);
const manel = new User('Manel', 'Cau', 'M', 46);
const berta = new User('Berta', 'Font', 'F', 38);
const users = [albert, paula, manel, berta];

// Create a library.
const library = new Library('Biblioteca Walrus', 'Nowhere Street - Nowhere Land');

// Add books and users to library.
books.forEach((book) => library.addBook(book));
console.log();
users.forEach((user) => library.addUser(user));
console.log();

// Let's test everything!
library.lendBook(quijote, albert);
library.lendBook(rings, paula);
library.lendBook(kr, manel);
library.returnBook(rings);
library.lendBook(rings, albert);
library.returnBook(kr);
library.lendBook(kr, paula);
library.returnBook(kr);
library.lendBook(kr, albert);
console.log();

// Create a new book and lend it to user.
const hamlet = new Book('Hamlet', 'William Shakespeare', 1603);
library.addBook(hamlet);
library.lendBook(hamlet, berta);
console.log();

// Show final library report.
library.report();
```

Consideracions importants que cal recordar:

- Les propietats i mètodes marcats amb un prefix `#` són privats. Només el codi interior de la mateixa classe pot accedir a aquests elements. Ho veiem en la propietat `#books` de la classe `User`. No s'hi pot accedir des de fora de la classe.
- Amb la paraula clau `get` podem definir una propietat calculada en funció d'altres propietats de la classe. És el cas de `fullName` en la classe `User`. 
- La paraula reservada `this` dins del codi d'una classe es refereix a l'objecte concret (la instància) sobre el qual s'anomenarà la propietat o el mètode en qüestió.
