En els temes 3 i 4 hem tingut l'ocasió de visitar alguns aspectes clau del desenvolupament d'aplicacions web mitjançant la _technology stack_ habitual basada en HTML, CSS i JavaScript. En aquest annex mostrem un cas integrat d'alguns dels punts vistos. Per reproduir l'exemple en la vostra màquina heu de crear tres arxius amb els noms `index.html`, `styles.css` I `scripts.js` i copiar els continguts que es mostren tot seguit.

```javascript
<!-- index.html -->

<!DOCTYPE html>
<html lang="en">
    <head>
        <meta charset="UTF-8" />
        <meta http-equiv="X-UA-Compatible" content="IE=edge" />
        <meta name="viewport" content="width=device-width, initial-scale=1.0" />
        <link rel="stylesheet" href="styles.css" />
        <script src="scripts.js" defer></script>
        <title>Pokémons</title>
    </head>
    <body>
        <header>
            <h1>Pokémons</h1>
        </header>
        <div id="root"></div>
        <footer>
            Exemple complet d'ús de les API Fetch i DOM. Dades de pokémons obtingudes de
            <a href="https://pokeapi.co/docs/v2">PokéAPI</a>.
        </footer>
        <noscript>Aquesta pàgina necessita JavaScript per funcionar correctament.</noscript>
    </body>
</html>
```

```javascript
// scripts.js

const numberOfPokemons = 5;

// Using `async` here as there is at least an `await` inside the function.
async function loadPokemons() {
    // Get a reference to existing <div> element where we want to show many pokémons.
    const container = document.querySelector('#root');

    // Create a loop to show as many pokémons as `numberOfPokemons`.
    for (let i = 1; i <= numberOfPokemons; i++) {
        // Get pokémon data from PokéAPI.
        const response = await fetch('https://pokeapi.co/api/v2/pokemon/' + i);
        const pokemon = await response.json();

        // Create a new <div> element to show this pokémon.
        const div = document.createElement('div');
        div.classList.add('pokemon-card');
        container.appendChild(div);

        // Add an <h2> element for this pokémon name.
        const h2 = document.createElement('h2');
        h2.style.color = '#cc4466';
        h2.textContent = pokemon.name;
        div.appendChild(h2);

        // Add an <img> element for this pokémon image.
        const img = document.createElement('img');
        img.src = pokemon.sprites.other.home.front_default;
        img.alt = pokemon.name;
        img.classList.add('pokemon-image');
        div.appendChild(img);
    }
}

loadPokemons();
```

Consideracions importants que cal recordar:

- La línia on enllacem l'arxiu JavaScript en el nostre document HTML necessita l'anotació `defer`. Recordeu per què? Què passa si l'eliminem?
- Fins ara no havíem vist l'etiqueta `<noscript>`. Podeu imaginar per a què es fa servir?
- En l'arxiu CSS tenim exemples de selectors d'etiqueta (com `h1` o `footer`), de classe (com `.pokemon-card`) i d'identificador (com `#root`). Tingueu en compte que els selectors poden barrejar aquests tres formats per formar expressions amb una certa complexitat. Per aprofundir en aquest tema, investigueu el concepte d'**especificitat**.
- La funció `loadPokemons()` utilitza el mecanisme `async await` que hem explicat anteriorment. Enteneu per què cal fer-ho així? Què passa si eliminem les paraules `async` I `await`? Què passa si eliminem la paraula `async` però mantenim `await`?
- Per establir la font de les imatges fem servir l'expressió `pokemon.sprites.other.home.front_default`. Per obtenir-la, hem de dur a terme una inspecció visual prèvia del resultat que torna una petició a la URL https://pokeapi.co/api/v2/pokemon/4. Veurem que es tracta d'un objecte amb diferents propietats, entre les quals hi ha `sprites`. El valor d'aquesta propietat és, alhora, un altre objecte amb diverses propietats, entre les quals consta `other`. Si seguim el mateix procediment podem arribar a l'expressió que ens facilita la direcció de la imatge que desitgem mostrar.
