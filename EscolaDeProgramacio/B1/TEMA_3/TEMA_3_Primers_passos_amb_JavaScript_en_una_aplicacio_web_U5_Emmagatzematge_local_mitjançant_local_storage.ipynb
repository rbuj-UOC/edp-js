## Unitat 5: Emmagatzematge local mitjançant _local storage_

Els navegadors web moderns inclouen diversos mecanismes perquè una pàgina web pugui emmagatzemar dades en la màquina de l'usuari, sempre que la persona ho permeti, amb la finalitat de poder recuperar-los posteriorment. Aquests mecanismes obren un ampli ventall de possibilitats, entre les quals s'inclouen:

- Guardar informació en local perquè la nostra pàgina funcioni bé fins i tot quan l'usuari no tingui connexió a internet.
- Conservar els ajustaments (per exemple, colors o grandàries de font) que l'usuari ha seleccionat perquè cada vegada que entri a la nostra pàgina se li apliquin de manera automàtica.
- Mantenir el nombre i quantitat d'articles que l'usuari ha afegit a un carret de la compra, perquè no es perdin si l'usuari tanca la pestanya o finestra del navegador.
- Emmagatzemar credencials codificades perquè l'usuari no hagi d'introduir-les cada vegada que vol iniciar sessió en un lloc web.

Des dels inicis de la web s'han utilitzat les _cookies_ per a aquestes finalitats. Actualment, però, disposem de mecanismes més moderns i simples que ens permeten aconseguir els mateixos objectius. Un d'ells, el que aquí presentem, consisteix en l'ús d'un objecte `Storage`, al qual accedim a través de la propietat `localStorage` de la finestra del navegador. En aquest objecte `Storage` podem guardar qualsevol valor associant-lo a una clau. Les dades emmagatzemades estaran disponibles fins i tot després de tancar i tornar a obrir el navegador. Vegem-ho amb un exemple:

```javascript
// Store some values in local storage.
localStorage.setItem("userName", "Albert Mas");
localStorage.setItem("backgroundColor", "#9b3c68");

// Retrieve some values from local storage.
const userName = localStorage.getItem("userName");
const backgroundColor = localStorage.getItem("backgroundColor");

console.log(userName);
console.log(backgroundColor);
```

Observem que el mètode `setItem()` necessita dos paràmetres: primer, la clau que s'utilitzarà per identificar el valor i, després, el valor que s'ha d'emmagatzemar. En canvi, el mètode `getItem()` només necessita la clau del valor que s'ha de recuperar. El funcionament, com veiem, és senzill i efectiu. No obstant això, `Storage` presenta una limitació important: tant els valors com les claus han de ser cadenes de text.

Com hem de desar, aleshores, els valors que no siguin simples cadenes de text? En el nostre codi Javascript treballem sovint amb números, valors booleans, dates, objectes, arrays... podrem emmagatzemar tot tipus de dades a través de `localStorage`? La resposta és afirmativa i se serveix d'un mecanisme que hem estudiat en el tema anterior: l'objecte global `JSON` i els seus mètodes estàtics `stringify()` i `parse()`.

A través de l'ús de `JSON.stringify()`, convertim qualsevol tipus de dada en un text que inclou la mateixa informació en format JSON:

```javascript
const usersAsJson = JSON.stringify(users);
console.log(usersAsJson);
```

De manera inversa, amb `JSON.parse()` tornem a generar un objecte de JavaScript a partir d'un text codificat en format JSON:

```javascript
const usersFromJson = JSON.parse(usersAsJson);
console.log(usersFromJson);
```

Així doncs, si volem guardar un array d'usuaris `users` en l'emmagatzematge local només necessitem combinar els dos mètodes, com es mostra a continuació:

```javascript
localStorage.setItem("users", JSON.stringify(users));
```

Quan més endavant (potser l'endemà) accedim a aquesta pàgina web i necessitem recuperar les dades d'usuaris emmagatzemats en l'emmagatzematge local, hem de dur a terme el procés invers:

```javascript
const users = JSON.parse(localStorage.getItem("users"));
```

> <br>💫&nbsp; **Ampliació**
> 
> Semblant a `localStorage`, també existeix `sessionStorage`. Aquest últim va associat a la sessió actual: `localStorage` no té temps d'expiració, mentre que la informació emmagatzemada en `sessionStorage` s'elimina en finalitzar la sessió de la pàgina. Aquesta sessió es manté mentre el navegador es troba obert, fins i tot encara que es recarregui la pàgina, però si la pàgina s'obre en una nova pestanya o finestra s'iniciarà una nova sessió que no tindrà disponibles les dades emmagatzemades en `sessionStorage`. Cap dels mecanismes és superior a l'altre: optar per `localStorage` o `sessionStorage` dependrà de les necessitats de cada cas.
