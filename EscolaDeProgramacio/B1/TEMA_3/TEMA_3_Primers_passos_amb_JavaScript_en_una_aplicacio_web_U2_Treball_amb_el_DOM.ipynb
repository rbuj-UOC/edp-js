## Unitat 2: Treball amb el DOM

El DOM (model d'objectes del document, en anglès _document object model_) consisteix en una representació en forma d'arbre de nodes que ens permet interactuar amb qualsevol document HTML des del codi JavaScript. En aquesta representació, cada node representa una part del document, generalment un element. El DOM, doncs, permet la modificació dels elements que formen una pàgina web de manera programàtica.

Ho entendrem millor amb un exemple ràpid. Si en una part del `<body>` del nostre document HTML tenim aquest contingut:

```javascript
<div id="ditto">
    <h2 class="pokemon-name">Ditto</h2>
    <img 
        src="https://raw.githubusercontent.com/PokeAPI/sprites/master/sprites/pokemon/132.png"
        alt="Sprite">
    <p>Ditto és un pokémon introduït a la <strong>primera generació</strong> amb 
        una forma semblant a la <em>gelatina</em>.</p>
</div>
```

En el DOM disposarem d'un arbre de nodes que inclouran aquests elements:

<pre>body
|-- ...
|-- div
|   |-- h2
|   |-- img
|   `-- p
|       |-- strong
|       `-- em
|-- ...</pre>

Aquest DOM ens facilita accedir a tots aquests elements des del nostre codi JavaScript, amb la qual cosa ens permet dur a terme una gran varietat d'accions:

- Eliminar un element del DOM perquè desaparegui del document HTML.
- Afegir un nou element al DOM perquè es mostri en el document HTML.
- Afegir o treure una classe CSS d'un element perquè canviï la seva aparença en pantalla.
- Canviar el contingut (per exemple, el text) d'un element del DOM perquè s'actualitzi en el navegador.
- Canviar la imatge que està mostrant un element `<img>` perquè mostri una imatge diferent a l'usuari.

I moltes més. De fet, algunes estratègies modernes de desenvolupament web parteixen d'un document HTML, l'únic contingut del qual és el següent:

```javascript
<html>
    <head>
        <meta charset="UTF-8">
        <script src="index.js"></script>
        <title>Títol del document</title>
    </head>
    <body>
        <div id="root"></div>
    </body>
</html>
```

I a partir d'aquest document, on el `<body>` només inclou un únic element  `<div>` amb l'identificador `root`, creen de manera dinàmica tot el contingut per mostrar en el document HTML des del codi JavaScript i el van afegint gràcies als mètodes que proporciona el DOM. Per fer-ho calen alguns mètodes que podem classificar en tres grups:

- Accés a elements del DOM.
- Creació i eliminació d'elements del DOM.
- Modificació de propietats d'elements del DOM.

> <br>🚨&nbsp; **Atenció**
> 
> El nostre codi JavaScript generarà un error si intentem accedir a elements del DOM abans que s'hagin definit. Això pot produir-se si la línia `<script src="index.js"></script>` apareix en el bloc `<head>` del nostre document. Per evitar aquest problema, podem moure aquesta línia al final del bloc `<body>` o bé podem mantenir-la en el bloc `<head>` afegint-hi l'anotació `defer` per deixar-la com a `<script src="index.js" defer></script>`. En aquest cas, el navegador carregarà l'arxiu `index.js` en segon pla i l'executarà una vegada que hagi carregat tot el document HTML. D'aquesta manera, el codi JavaScript no intentarà accedir a elements del DOM abans que s'hagin definit.

### Accés a elements del DOM

Per accedir als elements del DOM des de JavaScript disposem de dos tipus de mètodes. Els primers resulten més explícits, ja que el seu propi nom indica si volem recuperar elements a partir del seu identificador, d'alguna classe CSS que tenen informada o del seu nom d'etiqueta:

```javascript
const elem1 = document.getElementById("ditto");
const elems2 = document.getElementsByClassName("pokemon-name");
const elems3 = document.getElementsByTagName("p");
```

Els segons són més genèrics, però també molt més versàtils, ja que permeten recuperar elements a partir de qualsevol tipus de selector (simple o complex), com ho utilitzaríem en un full d'estils CSS:

```javascript
const elem4 = document.querySelector("#ditto");
const elems5 = document.querySelectorAll(".pokemon-name");
```

> <br>🚨&nbsp; **Atenció**
> 
> Observeu que `getElementById()` i `querySelector()` retornen un únic element, mentre que la resta de mètodes en retornen un conjunt. Heu de tenir en compte aquesta diferència a l'hora de treballar amb el resultat. Si, per exemple, voleu afegir una classe CSS als elements rebuts, en el primer cas podreu fer-ho directament sobre l'element, mentre que en el segon cas primer haureu d'iterar els elements rebuts per afegir-ne la classe CSS a cadascun de manera individual. En el cas de `querySelectorAll()`, a més, el resultat és un objecte de tipus `NodeList`, que no és un array, per la qual cosa no podrem usar mètodes com `forEach()` per iterar-ne els elements, tret que prèviament el convertim en un array.
> <br><br>

### Modificació de propietats d'elements del DOM

Un cop hem accedit a un element concret del DOM, en podem canviar qualsevol de les propietats i atributs. Per exemple, si volem canviar el text d'un element `<p>` podem utilitzar la propietat `textContent`:

```javascript
document.querySelector("h2").textContent = "Pokémon Ditto";
```

De manera semblant, podem utilitzar la propietat `innerHTML`, encara que en aquest cas el contingut pot contenir etiquetes HTML que es convertiran en nous elements del DOM:

```javascript
document.querySelector("h2").innerHTML = "Pokémon <em>Ditto</em>";
```

> <br>🚨&nbsp; **Atenció**
> 
> En general resulta més recomanable emprar `textContent` i prescindir d'`innerHTML`, ja que té un rendiment millor, preveu possibles atacs de tipus XSS i, sobretot, evita possibles errors tipogràfics. Quan utilitzem `innerHTML` podem escriure etiquetes d'obertura i de tancament, però no es duu a terme cap validació sobre la sintaxi, de manera que resulta senzill cometre algun error i que passi desapercebut. Per exemple, si escrivim `<p>Text de prova</p` en lloc de `<p>Text de prova</p>`, el navegador no mostrarà cap error, però el resultat serà que l'element `<p>` no es tancarà i la resta de contingut del document HTML s'hi mostrarà a dins. Això pot provocar que la resta d'elements del document es mostrin de manera incorrecta o que es produeixi un error de sintaxi en el document HTML.
> <br><br>

Mitjançant la propietat `classList` podem afegir o treure classes CSS d'un element:

```javascript
const elem6 = document.querySelector(".pokemon-name");
elem6.classList.remove("pokemon-name")
elem6.classList.add("pokemon-title")
```

Encara que, si només volem canviar una classe CSS per una altra, podem fer-ho més directament d'aquesta manera:

```javascript
document.querySelector(".pokemon-name").classList.replace("pokemon-name", "pokemon-title");
```

### Creació i eliminació d'elements del DOM

Podem crear elements en el DOM directament des de JavaScript mitjançant la instrucció següent, on indiquem l'etiqueta corresponent a l'element que volem crear:

```javascript
const myParagraph1 = document.createElement("p");
```

No obstant això, crear un element amb `createElement()` ha d'anar acompanyat d'altres instruccions que el configurin i (molt important no oblidar aquest últim pas) l'afegeixin al DOM amb la instrucció `appendChild()`. En cas contrari, simplement haurem creat un objecte de JavaScript que en cap moment es mostrarà en el navegador web de l'usuari. A més, per afegir-ho al DOM en un lloc concret de l'arbre, primer necessitarem recuperar aquest element amb algun dels mètodes vistos anteriorment. Així doncs, el bloc d'instruccions mínim que necessitarem queda així:

```javascript
const myParagraph2 = document.createElement("p");
myParagraph2.textContent = "Contingut del paràgraf creat des de JavaScript.";

const myDiv = document.querySelector("#ditto");
myDiv.appendChild(myParagraph2);
```

Eliminar elements del DOM resulta molt més senzill: n'hi ha prou amb utilitzar el mètode `remove()` després d'haver recuperat un element amb algun dels mecanismes coneguts. Qualsevol d'aquestes línies farà que el títol **Ditto** de l'exemple mostrat al principi d'aquesta unitat deixi de mostrar-se en pantalla:

```javascript
document.querySelector(".pokemon-name").remove();

document.querySelector("h2").remove();
```

> <br>💫&nbsp; **Ampliació**
> 
> Crear una estructura complexa d'elements niats amb crides repetides a `createElement()` i `appendChild()` pot resultar un treball bastant feixuc. Investigueu, buscant en la documentació que s'ofereix a la [MDN](https://developer.mozilla.org/en-US/docs/Web/HTML/Element/template), el funcionament de l'etiqueta `<template>` d'HTML i el seu ús des de JavaScript. Amb aquest mecanisme resulta possible crear contingut de manera dinàmica des de JavaScript amb un codi concís i net.
