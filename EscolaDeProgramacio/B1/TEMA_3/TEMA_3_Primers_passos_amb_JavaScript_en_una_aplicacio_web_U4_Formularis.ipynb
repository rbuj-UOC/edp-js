## Unitat 4: Formularis

Un escenari concret en què es barregen el treball amb el DOM, gestió d'esdeveniments i la programació amb Javascript és la validació de formularis. Els formularis són un mecanisme molt potent per interactuar amb els usuaris del nostre lloc web, ja que permet que ens enviïn informació o que interactuïn amb els diferents apartats de la pàgina. En aquest apartat veurem com podem crear formularis en HTML i com podem validar-los amb Javascript.

Disposem de diversos elements diferents per a la creació dels nostres formularis: quadres de text, caselles de selecció (_checkboxes_), botons normals, botons d'opció (_radio buttons_), quadres desplegables, barres de progrés... Haurem de consultar la documentació per estudiar tots els controls a la nostra disposició i les propietats i usos de cadascun quan requerim crear un formulari per al nostre lloc web.

### Creació del formulari

A continuació mostrem l'exemple d'un formulari per sol·licitar a l'usuari algunes dades personals per crear el seu compte en un lloc web de comerç electrònic:

```javascript
<!-- index.html -->

<!DOCTYPE html>
<html lang="en">
    <head>
        <meta charset="UTF-8" />
        <meta http-equiv="X-UA-Compatible" content="IE=edge" />
        <meta name="viewport" content="width=device-width, initial-scale=1.0" />
        <link rel="stylesheet" href="styles.css" />
        <script src="index.js" defer></script>
        <title>Form example</title>
    </head>
    <body>
        <form action="" method="get">
            <label for="name">Name</label>
            <input type="text" id="name" name="name" />
            <label for="email">Email</label>
            <input type="email" id="email" name="email" />
            <label for="age">Age</label>
            <input type="number" id="age" name="age" />
            <label for="country">
                <span>Country</span>
            </label>
            <select id="country" name="country">
                <option disabled selected value="">(Select your country)</option>
                <option value="argentina">Argentina</option>
                <option value="chile">Chile</option>
                <option value="colombia">Colombia</option>
                <option value="espana">España</option>
                <option value="mexico">México</option>
                <option value="peru">Perú</option>
                <option value="uruguay">Uruguay</option>
                <option value="venezuela">Venezuela</option>
            </select>
            <fieldset>
                <legend>Interests</legend>
                <input type="checkbox" name="politics" id="politics" />
                <label for="politics">Politics</label>
                <input type="checkbox" name="sports" id="sports" />
                <label for="sports">Sports</label>
                <input type="checkbox" name="programming" id="programming" />
                <label for="programming">Programming</label>
                <input type="checkbox" name="cooking" id="cooking" />
                <label for="cooking">Cooking</label>
            </fieldset>
            <fieldset>
                <legend>Gender</legend>
                <input type="radio" name="gender" id="male" value="male" />
                <label for="male">Male</label>
                <input type="radio" name="gender" id="female" value="female" />
                <label for="female">Female</label>
                <input type="radio" name="gender" id="other" value="other" />
                <label for="other">Other</label>
            </fieldset>
            <label for="comments">Comments</label>
            <textarea id="comments" name="comments"></textarea>
            <button type="submit">Sign up</button>
        </form>
    </body>
</html>
```

Consideracions importants:

- La creació d'un formulari utilitza l'etiqueta `<form>` amb dos atributs rellevants: `action` i `method`. 
- L'atribut `action` defineix l'adreça URL a la qual s'enviarà la informació recollida en el formulari quan l'usuari premi el botó d'enviar.
- L'atribut `method` defineix quin mètode HTTP s'utilitzarà per enviar aquesta informació, que en general serà GET (valor per defecte) o POST.
- És una bona idea incloure etiquetes `<label>` per millorar la usabilitat i accessibilitat dels nostres formularis. Vinculem una etiqueta amb un element del formulari a través de l'atribut `for` de l'etiqueta, que ha de tenir el mateix valor que l'atribut `id` de l'element amb el qual vinculem l'etiqueta. D'aquesta forma, en prémer sobre l'etiqueta, el cursor se situarà automàticament en l'element al qual està vinculada.
- L'atribut més important dels elements `<input>` és `type`, ja que canvia tant la seva aparença com el seu comportament. Els valors més comuns són `text`, `password`, `email`, `number`, `date`, `casella de selecció`, `ràdio`, `file`, `submit`, `reset` i `button`.
- La propietat `name` de cada element del formulari s'utilitzarà per enviar la informació i perquè en rebre-la sigui senzill saber a quin camp correspon cada valor introduït. És important que els noms dels camps siguin descriptius i que no es repeteixin en el mateix formulari.
- L'element `<button>` té un atribut `type` que accepta tres valors: `submit`, `reset` o `button`. Utilitzem `submit` perquè en prémer el botó s'enviï el formulari. Perquè no dugui a terme cap acció per defecte utilitzem el valor `button` (posteriorment, des del nostre codi Javascript, definirem quina acció ha de realitzar el botó). Amb `reset` aconseguim que en prémer el botó es restableixi el formulari als seus valors per defecte, però en general es considera una mala pràctica que cal evitar.

### Validació de les dades

Volem acceptar edats inferiors a 0 o superiors a 150? El nom de l'usuari pot tenir un únic caràcter? Els comentaris poden estar buits? El correu electrònic ha de tenir un format determinat? El país ha de ser un dels que apareixen en el desplegable? L'usuari ha de seleccionar almenys un interès? L'usuari ha de seleccionar almenys un interès i no més de dos, però si en selecciona dos, un ha de ser esports? 

HTML5 ens permet afegir atributs per marcar quins camps són requerits i quins valors màxims o mínims acceptarem, entre molts altres aspectes. Per fer-ho, disposem dels atributs `required`, `minlength` i `maxlength`, `min` i `max` o `pattern`. Però en altres ocasions ens interessarà dur a terme una validació molt més detallada i específica aprofitant el potencial que ens ofereix Javascript i l'accés al DOM vist anteriorment. 

Aplicant uns criteris un pèl curiosos, podríem decidir que una persona amb gènere femení no pugui mostrar interès per la cuina o una de gènere masculí no pugui mostrar interès pels esports. Per fer-ho, necessitem crear un arxiu **index.js** amb el codi següent i enllaçar-lo des del nostre document **index.html**:

```javascript
// index.js

function unselectInterest(event) {
    if (event.target.id === 'male') {
        // If radio button was `male`, uncheck `sports` checkbox.
        document.querySelector('#sports').checked = false;
    } else if (event.target.id === 'female') {
        // If radio button was `female`, uncheck `cooking` checkbox.
        document.querySelector('#cooking').checked = false;
    }
}

// Using `forEach` here to add same event listener to many elements.
document.querySelectorAll("input[name='gender']").forEach((input) => {
    input.addEventListener('change', unselectInterest);
});
```

Però hem de tenir en compte que l'usuari podrà tornar a marcar qualsevol interès després d'haver seleccionat el seu gènere. La validació de camps des de Javascript exigeix preveure sempre totes les possibilitats.

> <br>💫&nbsp; **Ampliació**
> 
> Recordeu l'ús dels mètodes `map()` i `filter()` vistos anteriorment? Són mètodes que tenim disponibles quan treballem amb arrays i que ens permeten, respectivament, crear un nou array aplicant alguna transformació als elements de l'array original, o bé crear-ne un de nou amb una selecció dels elements de l'array original. En el codi anterior trobem un altre mètode similar: `forEach()`. Aquest mètode, també disponible per a arrays i estructures de dades similars, com una `NodeList`, que és el que ens retorna el mètode `querySelectorAll()`, ens permet executar una acció per cada element de la col·lecció. En l'exemple anterior, `querySelectorAll()` retorna tots els elements `<input>` que tenen l'atribut `name` informat al valor `gender`. És a dir, tres elements: els corresponents a les opcions Male, Female i Other. L'ús del `forEach()` posterior ens serveix per afegir un gestor d'esdeveniments a cadascun d'aquests tres elements. Podríem aconseguir el mateix resultat amb l'ús d'un bucle convencional, però `forEach()` ens proporciona una manera molt concisa i simple de dur a terme aquesta acció.

Provem una cosa una mica més interessant: quan l'usuari introdueixi el seu nom, emplenarem automàticament el seu correu electrònic amb una proposta plausible.

```javascript
function fillEmail() {
    // Get name introduced by user...
    const name = document.querySelector('#name').value;
    // ...and convert it to lower case and create array with name substrings.
    const nameParts = name.toLowerCase().split(' ');
    // This only works OK for names including only one space - can you improve it to work in all cases?
    document.querySelector('#email').value = `${nameParts[0]}.${nameParts[1]}@somerandomdomain.com`;
}

document.querySelector('#name').addEventListener('change', fillEmail);
```

Amb aquest codi aconseguim que, quan l'usuari introdueix el nom Albert Mas, automàticament s'informa el correu electrònic albert.mas@somerandomdomain.com.

Finalment, ens podem assegurar que l'usuari introdueixi una edat en el rang exacte entre 18 i 94 anys i que aparegui una alerta en cas contrari.

```javascript
// Instead of creating a separate function with a name, we use an arrow function here.
document.querySelector('#age').addEventListener('change', (event) => {
    const age = event.target.valueAsNumber;
    if (isNaN(age) || age < 18 || age > 94) {
        // Is showing an `alert` good user experience? What do you think? 🧐
        alert('Your age must be a number between 18 and 94');
    }
});
```
