Javascript és un llenguatge de programació molt versàtil que actualment s'utilitza en un gran nombre d'escenaris diferents: creació d'aplicacions d'escriptori, programació d'apps per a mòbils, desenvolupament d'aplicacions de _backend_ per a servidors... i, per descomptat, també per aportar interactivitat a una pàgina web, per convertir-la en una autèntica aplicació web. Recordem el paper que té cadascuna d'aquestes tres tecnologies:

- HTML serveix per definir el contingut i l'estructura d'una pàgina web.
- CSS serveix per especificar el disseny i la maquetació d'una pàgina web.
- Javascript permet programar el comportament de tots els elements que formen una pàgina web.

## Unitat 1: Nocions bàsiques d'HTML i CSS

En aquesta primera unitat fem un repàs ràpid a les nocions bàsiques que hem de conèixer d'HTML i CSS, perquè això després ens permeti treballar de manera eficient amb Javascript, interactuant amb aquests dos llenguatges de marcatge i estils. Malgrat això, tant HTML com CSS, especialment aquest darrer, tenen una especificació molt més extensa que hem d'intentar conèixer al màxim si volem convertir-nos en programadors d'aplicacions web. Així, hem d'entendre aquesta primera unitat només com una exposició ràpida i no exhaustiva dels elements més bàsics. Programar aplicacions web útils i robustes exigeix un estudi molt més profund d'aquestes dues tecnologies i una bona dosi de pràctica.

### Llenguatge de marcatge HTML

HTML (llenguatge de marques d'hipertext, en anglès _HyperText Markup Language_) és el llenguatge de marcatge que farem servir per definir el contingut i l'estructura de les pàgines web. Consisteix en una sèrie d'elements que utilitzem per marcar diferents parts del contingut d'una pàgina web, amb la qual cosa s'aconsegueix que apareguin d'una certa manera o es comportin d'una determinada forma. Per exemple, aquesta és l'estructura bàsica d'un document HTML:

```javascript
<html>
    <head>
        <meta charset="UTF-8">
        <link rel="stylesheet" href="styles.css">
        <script src="scripts.js"></script>
        <title>Títol del document</title>
    </head>
    <body>
        <header>
            <h1>Encapçalament de nivell 1 (máximo)</h1>
        </header>
        <p>Contingut del primer paràgraf</p>
        <table>
            <tbody>
                <tr>
                    <td>Primera cel·la de la primera fila</td>
                    <td>Segona cel·la de la primera fila</td>
                </tr>
                <tr>
                    <td>Primera cel·la de la segona fila</td>
                    <td>Segona cel·la de la segona fila</td>
                </tr>
            <tbody>
        </table>
    </body>
</html>
```

Consideracions introductòries importants:

- HTML està basat en un conjunt d'elements amb una determinada semàntica. Per exemple, `<h1>` serveix per definir un encapçalat important i `<p>` serveix per definir un paràgraf. Hi ha més d'un centenar d'elements, però habitualment n'utilitzem unes desenes com a molt, per la qual cosa no resulta difícil aprendre'ls. En l'enllaç següent podem consultar la llista completa d'elements d'HTML: [https://developer.mozilla.org/en-US/docs/Web/HTML/Element](https://developer.mozilla.org/en-US/docs/Web/HTML/Element).
- Els elements, en general, tenen etiqueta d'obertura i etiqueta de tancament (la que incorpora un caràcter  `/` abans del nom de l'etiqueta). Tanmateix, hi ha elements que no requereixen etiqueta de tancament, com per exemple `<img>`. En aquests casos, l'etiqueta d'obertura és suficient per indicar que l'element ha acabat.
- Els elements segueixen una estructura jeràrquica o d'arbre. En l'exemple anterior, l'element `<html>` inclou dos elements fills: `<head>` I `<body>`. Alhora, `<head>` té quatre elements fills, mentre que `<body>` en té tres: un element `<header>`, un element `<p>` I un element `<table>`.
- Els elements s'han de tancar sempre seguint l'ordre adequat: si `<tr>` inclou un element `<td>`, hem de tancar `</td>` abans de tancar `</tr>`. Si no ho fem, el navegador web no podrà interpretar correctament l'estructura de la pàgina. 

El `<head>` és la part d'un document HTML que no es mostra en el navegador web quan es carrega la pàgina. Conté informació com el títol de la pàgina (amb l'element `<title>`), els enllaços a arxius CSS per aplicar-hi estils (amb elements `<link>`), els enllaços a arxius JavaScript (amb elements `<script>`) i altres metadades, com l'autor de la pàgina o algunes paraules clau (amb elements `<meta>`). Encara que tota aquesta informació no serà, en general, directament visible per a l'usuari, sí que resulta interpretable per part de navegadors i cercadors. Per exemple, el títol de la pàgina apareix en la pestanya del navegador i en els resultats de cerca.

El `<body>`, en canvi, és la part on afegim tot el contingut HTML que sí que s'ha de mostrar en pantalla perquè l'usuari pugui veure'l i interactuar-hi. Dins del `<body>` afegirem una gran varietat d'elements:

- Etiquetes per donar estructura al document, `<nav>`, `<div>`, `<span>`, `<main>`, `<section>`, `<aside>`, `<header>` o `<footer>`.
- Etiquetes de text com `<h1>`, `<h2>`... fins a `<h6>`, `<p>`, `<strong>` o `<em>`.
- Etiquetes `<a>` per crear enllaços, que són un element fonamental i identitari d'una pàgina web.
- Etiquetes per crear taules, com `<table>`, `<thead>`, `<tbody>`, `<th>`, `<tr>` o `<td>`.
- Etiquetes per mostrar llistes com `<ol>`, `<ul>`, `<li>`, `<dl>`, `<dd>` o `<dt>`.
- Etiquetes per mostrar text amb formats especials, com `<code>`, `<pre>`, `<blockquote>`, `<cite>` o `<address>`.

### Llenguatge d'estils CSS

CSS (fulles d'estil en cascada, en anglès _Cascading Style Sheets_) és un conjunt de regles que ens permetrà descriure com es mostrarà la pàgina web en la pantalla de l'usuari. Els navegadors tenen nivells de suport de CSS lleugerament diferents, així que convé tenir present quins navegadors i quines versions tenen suport per a les diferents característiques de CSS. En l'enllaç següent podem consultar la compatibilitat de CSS: [https://caniuse.com](https://caniuse.com).

El llenguatge d'estils CSS gira entorn del que anomenem regles. Aquestes regles poden prendre diferents formes, com per exemple «si afegim una determinada classe a un element, volem que hi apliquin aquests estils», o bé «si un element A és fill d'un element B, llavors volem que hi apliquin aquests altres estils». 

Les regles de CSS són blocs amb aquesta sintaxi:

<pre>div {
    color: #e5e5e5;
    background-color: #930345;
}</pre>

En concret, una regla està formada per un selector, que en l'exemple anterior és `div`, i un grup de declaracions, que és tot el bloc entre les claus `{...}`. El grup de declaracions inclou una o més declaracions, en aquest cas dos. Cada declaració està formada per una propietat i un valor. En l'exemple anterior, la primera declaració és `color: #e5e5e5`, que indica que el color del text de tots els elements `<div>` serà el color #e5e5e5. La segona declaració és `background-color: #930345`, que indica que el color de fons de tots els elements `<div>` serà el color #930345.

Tot seguit mostrem un exemple d'un arxiu CSS bàsic que inclou tres regles:

<pre>h1 {
    font-size: 16px;
    color: #990033;
}

.container {
    margin: 1rem 2rem;
    padding: .5rem 1rem;
}

#first {
    font-weight: 700;
    background-color: salmon;
}</pre>

Consideracions introductòries importants:

- El selector `h1` assigna estils a tots els elements `<h1>` de la nostra pàgina. En aquest cas, el selector és el nom de l'etiqueta d'HTML.
- El selector `.container` (amb punt inicial) assigna estils a tots els elements que tenen assignada la classe `container` amb l'ús de `class='container'` en el document HTML. En aquest cas, el selector és el nom de la classe.
- El selector `#first` (amb coixinet inicial) assigna estils a tots els elements que tenen assignat l'identificador `first` amb l'ús `id='first'` en el document HTML. Tanmateix, aquesta regla s'hauria d'aplicar només a un element, ja que en un document HTML no haurien d'haver-hi elements amb identificadors repetits. En aquest cas, el selector és el nom de l'identificador.
