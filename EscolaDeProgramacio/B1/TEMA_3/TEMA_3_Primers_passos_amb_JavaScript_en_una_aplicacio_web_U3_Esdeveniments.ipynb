## Unitat 3: Esdeveniments

Podem entendre el concepte d'esdeveniment com una mena de succés que s'ha produït en un document HTML i al qual pot interessar-nos reaccionar des del nostre codi JavaScript. Un exemple d'esdeveniment és l'usuari prement un botó. També es disparen esdeveniments quan el document acaba de carregar-se i quan l'usuari passa el ratolí per sobre d'un text o d'una imatge. O, quan prem una tecla, canvia el valor d'un quadre de text o fa scroll amb la rodeta del ratolí. En qualsevol cas, l'esdeveniment que volem prendre en consideració de manera més habitual és, amb molta diferència, l'acte de fer clic sobre qualsevol element del document HTML. Aquest esdeveniment és el que més utilitzarem en els nostres desenvolupaments web.

Des del nostre codi JavaScript podem expressar interès per saber quan es produeix un determinat esdeveniment i establir què voldrem dur a terme (quines instruccions desitjarem executar) quan succeeixi. Per fer-ho emprem els mètodes d'accés a elements del DOM vistos en la unitat anterior i el mètode `addEventListener()`. Amb aquest mètode indiquem quin esdeveniment en concret volem monitorar (usuari fent clic, usuari passant el ratolí per sobre, usuari fent scroll, etc.) i també quin codi volem executar cada vegada que l'esdeveniment es produeixi. El codi que volem executar s'anomena **controlador d'esdeveniments** i se sol expressar com una funció anònima.

> <br>🚨&nbsp; **Atenció**
> 
> Com tots els exemples que treballen amb el DOM, convé copiar el codi que mostrem aquí en un arxiu JavaScript convenientment enllaçat des d'un document HTML. En general, una de les millors maneres de provar el nostre codi JavaScript sol ser a través d'una estructura bàsica formada per un document HTML (p. ex. **index.html**), un arxiu CSS (p. ex. **styles.css**) i un fitxer JavaScript (p. ex. **index.js**).
> <br><br>

```javascript
<!-- index.html -->
<div id="ditto">
    <h2 class="pokemon-name">Ditto</h2>
    <img 
        src="https://raw.githubusercontent.com/PokeAPI/sprites/master/sprites/pokemon/132.png"
        alt="Sprite">
    <p>Ditto és un pokémon introduït en la <strong>primera generació</strong> amb 
        una forma semblant a la <em>gelatina</em>.</p>
</div>
```

```javascript
// index.js
function doSomething() {
    console.log("User has clicked on some element");
}

document.querySelector("img").addEventListener("click", doSomething);
```

> <br>🎯&nbsp; **Vist anteriorment**
> 
> Observeu que, en el codi anterior, dins d'`addEventListener()` el nom de la funció `doSomething` no va acompanyat de parèntesis, com és habitual quan utilitzem una funció. Això és així perquè quan afegim els parèntesis després del nom de la funció estem cridant (o invocant o executant) aquesta funció. En aquest cas, però, no volem fer-ho. Només volem informar al mètode `addEventListener()` que, quan en algun moment l'usuari faci clic sobre l'element indicat (si és que això arriba a passar), `doSomething()` és la funció que s'ha d'executar. És perfectament possible que la funció `doSomething()` no arribi mai a executar-se si l'usuari en cap moment fa clic sobre la imatge. Si ho recordeu, aquest comportament i aquest ús del nom de la funció sense els parèntesis el vam veure anteriorment, quan vam estudiar els mètodes `map()` i `filter()`.
> <br><br>

Sovint volem poder accedir, des de dins de la funció que s'executa, a les metadades de l'esdeveniment que s'ha produït. Serà el cas si hem enllaçat la mateixa funció amb diferents esdeveniments o bé amb el mateix esdeveniment sobre diferents elements. Per exemple, podem tenir deu imatges i voler que quan l'usuari cliqui sobre qualsevol d'elles s'executi la funció `showInfo()`, però en el codi d'aquesta funció caldrà que sabem quina imatge en concret és la que l'usuari ha clicat. Ho aconseguim definint un paràmetre `e`, `evt` o `event`(el nom és arbitrari, però se sol emprar algun d'aquests per fer obvi que es tracta de l'esdeveniment) en la funció:

```javascript
function showInfo(event) {
    console.log(event.target);
}

document.querySelector("img").addEventListener("click", showInfo);
```

L'esdeveniment que es rep com a paràmetre és un objecte amb moltes propietats que aporten informació diversa i interessant sobre l'esdeveniment: en quines coordenades de la pantalla s'ha produït l'esdeveniment, quin element l'ha desencadenat, quin tipus d'esdeveniment era, etc. En general, `event.target` serà una de les propietats que utilitzarem amb més freqüència, ja que recull l'element HTML sobre el qual s'ha produït l'esdeveniment. En l'exemple anterior, `event.target` serà la imatge sobre la qual l'usuari ha fet clic. Podem utilitzar aquesta propietat per accedir als atributs de la imatge, com per exemple la seva `src` o la seva `alt`.

Encara que el codi anterior és perfectament correcte, recordem que durant l'estudi de les funcions de fletxa vam veure que aquestes dues funcions són pràcticament equivalents:

```javascript
function isEven(number) {
    return number % 2 === 0;
}

number => number % 2 === 0;
```

Utilitzant el mateix concepte, la manera més habitual de definir els detectors d'esdeveniment (_event listeners_) és mitjançant l'ús de funcions de fletxa (_arrow functions_):

```javascript
document.querySelector("img").addEventListener("click", event => console.log(event.target));
```

> <br>🚨&nbsp; **Atenció**
> 
> En alguns tutorials podeu trobar una manera alternativa de definir esdeveniments directament en el document HTML mitjançant la sintaxi `<img ... onclick="someFunction()">`. Encara que sembli més senzill i directe, aquest mètode resulta menys flexible i el seu ús és cada cop menys habitual. Convé, doncs, comprendre i utilitzar el mecanisme explicat en aquesta unitat, basat en l'ús del mètode `addEventListener()`.
