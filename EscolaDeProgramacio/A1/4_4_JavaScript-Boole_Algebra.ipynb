# NOT operator: !

```javascript
// Define the variable
let subscribed;

// Initialize the variable: here we should set the value stored in the system
subscribed = false;

// Change the value stored in the system
subscribed = !subscribed;

// Output the result
console.log(subscribed);
```

 # OR operator: ||

```javascript
// Define variables 
let administrator; 
let student; 
let access;

// Initialize variables 
administrator = true; 
student = false; 
access = false; // We don't know this value yet, but it is a good idea to initialize the variable in any case.

// Calculate access 
access = administrator || student; 

// Output the result 
console.log("Can the user access the document? " + access); 
```

# AND operator: &&

```javascript
// Define the variables
let citizenship;
let ofLegalAge;
let canVote;

// Initialize the variables: here we describe if the person has citizenship and if they are of legal age
citizenship = true; // In this case, the person has citizenship
ofLegalAge = false; // The person is not of legal age
canVote = false; // We don't know this value yet, but it is a good idea to initialize the variable in any case.

// Calculate if they can vote
canVote = citizenship && ofLegalAge;

// Output the result
console.log("Can the person vote? " + canVote);
```
