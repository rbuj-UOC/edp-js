# IF/ELSE

```javascript
// Define the variables 
let legalAge = 18; // Majority of age 
let age = 21;  // Define the age of the person and assign a value

// Check if the person is of legal age 
if (age >= legalAge) { 
    // Print the result 
    console.log("The person is of legal age"); 
} else { 
    // Print the result 
    console.log("The person is not of legal age"); 
}
```

# FOR/TO

```javascript
// Define variables
let numberOfEggs; 
let numberOfPotatoes; 

// Initialize variables 
numberOfEggs = 5; 
numberOfPotatoes = 3; 

// Crack the eggs 
for (let counter = 1; counter <= numberOfEggs; counter++) { 
    console.log("Crack egg number " + counter); 
} 

// Beat the eggs 
console.log("Beat the eggs..."); 

// Peel and chop potatoes 
for (let counter = 1; counter <= numberOfPotatoes; counter++) { 
    console.log("Peel potato number " + counter); 
    console.log("Chop potato number " + counter); 
} 

// Deep fry the potatoes 
console.log("Deep fry the potatoes..."); 

// Blend potatoes and eggs 
console.log("Mix potatoes and eggs..."); 

// Cook the omelette 
console.log("Cooking the omelette...");
```

# WHILE

```javascript
// Define variables
let phonePrice; // Price in euros
let savedAmount; // Amount of money already saved
let spareChange; // Spare change you save every day

// Initialize variables
phonePrice = 350.00;
savedAmount = 0.00;
spareChange = 0.00;

// Add money every day until you have saved enough to buy the phone
while (savedAmount < phonePrice) {
    spareChange = Math.random() * 10.0; // We simulate spare change as a random number from 0 to 10
    savedAmount += spareChange; // Add spare change to amount saved
    console.log("Spare change: " + spareChange + " Total saved: " + savedAmount); // Print the result
}
```
