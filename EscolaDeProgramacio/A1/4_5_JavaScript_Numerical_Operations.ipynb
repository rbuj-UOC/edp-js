# Numerical operators examples

```javascript
// Define variables
let x;
let y;
let z;
let b;

// Initialize variables
x = 5;
y = 3;
z = 7;
b = false;

// Print separator
console.log("Internal operators");
console.log("------------------");

// Sum operator
x = 5 + 3;
console.log("Sum operator (first example):");
console.log("x = " + x); // Print the value of x

x = y + z;
console.log("Sum operator (second example):");
console.log("x = " + x); // Print the value of x

console.log(" "); // Blank line

// Subtraction operator
x = 8 - 9;
console.log("Subtraction operator (first example):");
console.log("x = " + x); // Print the value of x

x = y - z;
console.log("Subtraction operator (second example):");
console.log("x = " + x); // Print the value of x

console.log(" "); // Blank line

// Product operator
x = 6*y;
console.log("Product operator (first example):");
console.log("x = " + x); // Print the value of x

x = y*z;
console.log("Product operator (second example):");
console.log("x = " + x); // Print the value of x

console.log(" "); // Blank line

// Quotient operator
x = 6/3;
console.log("Quotient operator (first example):");
console.log("x = " + x); // Print the value of x

x = y/z;
console.log("Quotient operator (second example):");
console.log("x = " + x); // Print the value of x

console.log(" "); // Blank line

// Print separator
console.log("External operators");
console.log("------------------");

// Equal to operator
b = (x==y);
console.log("Equal to operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line

// Different from operator
b = (x!=y);
console.log("Different from operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line

// Greater than operator
b = (x>y);
console.log("Greater than operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line

// Greater or equal than operator
b = (x>=y);
console.log("Greater or equal than operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line

// Smaller than operator
b = (x<y);
console.log("Smaller than operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line

// Smaller or equal than operator
b = (x<=y);
console.log("Smaller or equal than operator:");
console.log("b = " + b); // Print the value of b

console.log(" "); // Blank line
```
