```javascript
// Define and initialize variables
let numPeople = 4;  // Number of people that are having omelette. We will calculate the recipe for 4 people
let numEggs = 0; // Number of eggs needed. We don't know the value yet, but initialize to 0
let numPotatoes = 0;  // Number of potatoes needed. We don't know the value yet, but initialize to 0
let numOnions = 0;  // Number of onions needed. We don't know the value yet, but initialize to 0

let withSalt = true;  // If we want to add salt, this variable should be true.
let withOnion = true;  // If we want to add onion, this variable should be true.

// Calculate the number of potatoes and eggs needed.
numPotatoes = numPeople + 1;  // One potato per person + 1 extra.
numEggs = numPeople + 2;  // One egg per person + 2 extra.

// Tell the cook how many potatoes and eggs are needed.
console.log("Ingredients:");
console.log("To cook a 'Tortilla' for " + numPeople + " people, you need:");
console.log("   - " + numPotatoes + " potatoes");
console.log("   - " + numEggs + " eggs");

// If onion is needed, inform the cook.
if (withOnion) {
    numOnions = 1;
    console.log("   - " + numOnions + " onion(s)");
}

// If salt is needed, inform the cook.
if (withSalt) {
    console.log("   - Salt");
}

// Start cooking
console.log("Let's start cooking.");

// Step 1: Peel and chop the potatoes.
console.log("Peel and chop the potatoes:");

for (let i = 1; i <= numPotatoes; i++) {
    console.log("   - Peel and chop potato number " + i);
}

// Step 2: Heat the oil and fry the potatoes.
console.log("In a pan, pour enough oil to cover the potatoes and put it to heat.");
console.log("When the oit is hot, add the potatoes.");

// Step 3: Crack and beat the eggs.
console.log("While we cook the potatoes, crack and beat the eggs:");

for (let i = 1; i <= numEggs; i++) {
    console.log("   - Crack and pour into a bowl egg number " + i);
}

// If salt is needed, add it to the mixture.
if (withSalt) {
    console.log("Add salt to the eggs.");
}

// Beat the eggs.
console.log("Beat the eggs.");


// Step 4: When the potatoes are ready, mix the with the eggs.
console.log("When ready, take the potatoes off the pan, rinse them, and mix with the eggs.");

// Step 5: If onion is needed, fry and add it to the mixture.
if (withOnion) {
    console.log("Chop the onion. Fry it till golden and add it to the mixture.");
}

// Step 5: Cook the mixture to obtain the tortilla.
console.log("Pour the potato and egg mixture into the pan.");
console.log("   - Cook for 5 minutes.");
console.log("   - Carefully flip the tortilla.");
console.log("   - Heat for 5 more minutes.");

console.log("Your tortilla is ready to eat!");
```
