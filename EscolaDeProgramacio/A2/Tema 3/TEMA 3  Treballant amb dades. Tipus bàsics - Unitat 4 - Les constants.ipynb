## <font color='#000078'><center>Unitat 4: Les constants</center></font>

Ja hem vist que les variables ens permeten guardar valors que poden variar al llarg de l’execució d’un programa. De vegades, però, l’algorisme necessita treballar amb valors constants, que sabem del cert que no canviaran amb les operacions que faci el programa, llavors farem servir **constants** per emmagatzemar aquest valor. Una constant és igualment una caixa on podem guardar una dada d’un tipus i que s’identifica amb un nom, però aquesta caixa “guarda” un **valor predeterminat** i fix durant tot el programa, d’aquí el seu nom.

Pel que fa al nom de les constants, s’apliquen les mateixes regles que per a les variables, és a dir, ha de ser un nom simple, clar i significatiu, i ha de complir amb les mateixes regles i restriccions.

### Declaració de les constants:

Pel que fa a la declaració de constants, emprarem const per declarar-les. 

Penseu que const té la particularitat que no es pot tornar a redefinir tot i que, per a la resta de funcionalitats, tindrà una manera d’actuar semblant a la de let.

La manera de declarar una constant serà la següent:

#### 🧐 Exemple 


```javascript
//Variables
const constantName = 103;
```

en què després de la paraula clau **const** hi indicarem el nom que vulguem que tingui la constant.
