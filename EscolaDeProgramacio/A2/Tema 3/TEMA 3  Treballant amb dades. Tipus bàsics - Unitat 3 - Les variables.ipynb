## <font color='#000078'><center>Unitat 3: Les variables</center></font>

Una variable és un espai de la memòria de l’ordinador que emmagatzema valors. Ara, però, no ens preocupa com s’implementa físicament una variable a la memòria de l’ordinador, sinó com es pot declarar i fer servir en un algorisme per guardar les dades que s’hagin de tractar.
 
Les variables són un element clau en l’algorísmica, perquè ens permeten tractar amb les dades: podem guardar-hi valors, fer-hi operacions, intervenen en la presa de decisions quan hem de decidir entre una alternativa o altra, i ens ajuden a determinar quan ha de parar una iteració, com veurem més endavant.

Un algorisme necessita recordar la informació d’entrada, com pot ser un número, per poder-la tractar i operar-hi, i també necessita la informació que va obtenint per solucionar el problema, les dades intermèdies, com són el resultat de les operacions que es van fent. Les variables es fan servir per emmagatzemar aquesta informació. Una variable es pot veure com una caixa on es deixa cada una de les dades que va utilitzant l’algorisme.

### Nom de les variables:

Perquè un algorisme pugui treballar amb una variable necessita tenir un nom per poder-s’hi referir, així com el valor de la dada. Una variable sempre està definida per aquests dos elements. El valor guardat en una variable pot canviar (“variar”) perquè l’algorisme el modifiqui amb els tractaments que faci.

#### Regles per triar el nom d’una variable

El nom que li donem a una variable ha de ser simple, clar i significatiu, ha de tenir sentit i fer referència al valor que hi haurem de guardar. Això facilita la llegibilitat de l’algorisme a qualsevol persona que l’hagi de llegir, i a nosaltres mateixos si passat un temps l’hem de modificar.

Aquesta convenció de nomenclatura és adoptada generalment per tots els llenguatges de programació com un costum que els bons programadors segueixen.

Serien noms correctes: <code>segonsInicials</code> o <code>segons_inicials</code>.

I incorrectes o no recomanables: <code>s</code> (no s’entén el significat), <code>s_ini</code> (poc concret),  <code>segons inicials</code> (incorrecte, conté espais en blanc).

A l'hora de decidir el nom d’una variable hem de tenir en compte algunes regles i restriccions:

* El nom de les variables pot contenir lletres (a-z), números (0-9), el símbol dòlar ($) i guió baix (_).  
* Tot i que pot contenir números, el nom no pot començar per un número.  
* El nom no pot contenir espais en blanc
* El nom no pot coincidir amb cap paraula reservada de Javascript (consulteu les paraules reservades)
* Com ja hem dit abans, els noms de les variables son case sensitive, és a dir, es diferencien les majúscules i les minúscules (la variable nom i la variable Nom seran variables diferents).

### Declaració de les variables:

La manera habitual de declarar una variable, tant en aquesta assignatura com generalment quan programem amb JavaScript, és emprant la paraula clau **let**. Tot i que tenim una altra paraula clau que és capaç de declarar també una variable: **_var_**, malgrat que aquesta s’empra per fer l’àmbit d’aplicació de la variable en un rang més global del programa i, per això, nosaltres no la farem servir. Dit això, un exemple de com podem fer una declaració d’una variable seria el següent.

<code>let variableName;<code>

En què, després de la paraula <code>let</code> hi indicarem el nom que vulguem que tingui la variable. 

Cal tenir en compte que JS és un llenguatge “feblement tipat”, això vol dir que no cal declarar el tipus d’una variable quan es defineix. El tipus s’assumeix dinàmicament en el moment que se li assigna un valor, és a dir, que quan donem un valor a la variable el programa sap de quin tipus és la variable.
De fet, en JS ni tan sols és obligatori declarar les variables abans de fer-les servir, tot i que és una pràctica gens recomanable, i per això, sempre les declararem.

#### 🧐 Exemple 


```javascript
//Variables
let start_seconds = 103;
let minutes;
let remaining_seconds;
```
