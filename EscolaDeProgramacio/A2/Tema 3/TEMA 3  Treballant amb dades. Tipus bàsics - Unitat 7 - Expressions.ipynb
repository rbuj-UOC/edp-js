## <font color='#000078'><center>Unitat 7: Expressions</center></font>

Una **expressió** és un conjunt d’**operadors** i **operands** que formen una unitat de codi: que té sentit per si mateixa i que ens retorna un valor.

Podríem fins i tot assegurar que tot el que hem vist fins ara són expressions, tot i que no ens havíem posat a definir-les. Tot i això sí que hauríem de definir casos particulars d’expressions

Normalment es tracta d’operacions amb números (aritmètiques). I les podem indicar amb els operadors associats a cada tipus: **+**, **-**, **/**, etc. Però també podem fer expressions amb **booleans** i **strings**.

Podem construir expressions amb valors (per exemple, <code> 3 + 5 </code>) o amb variables (per exemple, <code> let plugesPrimerTrimestre = plugesGener + plugesFebrer + plugesMarç; </code>; en què les quatre variables fossin de tipus enter).
Això sí, és important vigilar que els tipus de valors o de variables que fem servir a l’expressió siguin correctes.

### Precedència dels operadors (ordre de prioritats)

És molt important tenir en compte l’ordre en què JavaScript interpreta els operadors d’una expressió d'acord amb la **precedència** que defineix per cadascun que intervé en la mateixa estableix entre els diferents tipus d’operadors.

La precedència d'operadors determina l'ordre en què els operadors són avaluats un respecte de l'altre en el moment d’avaluar una expressió. Els operadors amb més precedència es converteixen en els operands dels operadors amb menys precedència.

> IMPORTANT: Sempre podrem redefinir la precedència dels operadors mitjançant l'ús de parèntesis.

Fonamentalment, tindrem les regles següents:
* Per a una mateixa prioritat, els operadors s’avaluaran d’esquerra a dreta.
    És a dir, a l’expressió <code>1 – 8 + 5 + 2</code> s'avaluarà primer <code>1-8</code>(-7), després <code>-7 + 5</code> (-2), i finalment <code>-2 + 2</code> (0) i donarà com a resultat <code>0</code>. 

* No passarà de la mateixa manera amb prioritats diferents, en què s’avaluaran primer per prioritats i, quan s'igualin les prioritats sí que ho farà d’esquerra a dreta. 
És a dir: a l’expressió <code>1 + 4 * 2</code>, avaluarà primer <code>4 * 2</code>(8) i ens quedarà <code>1 + 8</code> (9) i donarà com a resultat <code>9</code>.

#### 🧐 Exemple 

```javascript
console.log(10 - 6 / 2);   // will throw 7 (the result of 10 - 3)
console.log(10 - (6 / 2)); // as above it will throw 7 because the parentheses are not involved
console.log((10 - 6) / 2); // will throw 2 (the result of 4 / 2)

console.log(false && true || true); // will throw true (the result of false || true)
console.log((false && true) || true); // as above it will throw true because the parentheses are not involved
console.log(false && (true || true)); // will throw false (the result of false && true)
```
