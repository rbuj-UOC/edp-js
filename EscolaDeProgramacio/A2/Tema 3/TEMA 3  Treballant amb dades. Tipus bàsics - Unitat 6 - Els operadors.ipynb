## <font color='#000078'><center>Unitat 6: Els operadors</center></font>

Fins ara hem estudiat les variables i les constants, però per si soles són de poca utilitat. Per implementar un programa és necessari tenir més eines.

Els operadors ens permeten manipular el valor de les variables, fer operacions matemàtiques i comparar els valors de diferents variables. D’aquesta manera, els operadors permetran fer càlculs complexos i prendre decisions lògiques en funció de comparacions i altres tipus de condicions.

#### Classes d’operadors:

Bàsicament, podem agrupar els operadors en dos grups:
* **Operadors Unaris**: són els que requereixen un sol operand sigui davant o darrere de l’operador (per exemple x++ en què x seria l’**operand** i ++ l’**operador**)
* **Operadors Binaris**: són els que requereixen dos operands (un bon exemple és l’assignació x = y en què x i y són els **operands** i = l’**operador**).

Com a cas particular, tenim l’**Operador Ternari** que empra tres operands.

#### Tipus d’operadors:

Hi ha molts tipus d’operadors, i tots tenen la seva importància. De totes maneres, caldrà entendre bé el funcionament dels següents:


### Operadors d’assignació

En un algorisme, ja sabem com s'han d'indicar quines variables farem servir i de quin tipus són. Per poder utilitzar una variable en un algorisme, ens cal poder donar-li (assignar-li) un valor, és a dir, saber com guardar un valor a la caixa per poder-lo emprar en l’algorisme quan ens calgui recordar-lo.

Mireu si és important poder fer això que l’acció d’assignar un valor a una variable és una operació bàsica i fonamental sense la qual no es pot pensar cap algorisme.

Per assignar valors a variables o constants farem servir el símbol **=**

<code>nomVariable = valor;</code>

Llavors, no podrem utilitzar el símbol **=** per comparar valors perquè seria confús. Per comparar si dos valors són iguals s’utilitza el símbol **==**


### Operadors relacionals (de comparació)

Els operadors relacionals ens permeten comparar expressions de tipus compatibles i retornar un resultat de tipus lògic (booleà): el resultat de la comparació és veritable o fals (**_true_** o **_false_**). 

En el cas del **tipus Number**, els operadors de comparació són: <font style="color:blue">**==**</font> (igual), <font style="color:blue">**!=**</font> (diferent), <font style="color:blue">**>**</font> (més gran), <font style="color:blue">**<**</font> (més petit), <font style="color:blue">**<=**</font> (més petit o igual) i <font style="color:blue">**>=**</font> (més gran o igual). 

Cal tenir en compte els casos de **més petit o igual** i **més gran o igual**, tenint en compte que l’operador s’escriu gramaticalment tal com s’expressa i exclusivament funcionarà d’aquesta manera (és a dir <font style="color:blue">**>=**</font> i <font style="color:blue">**<=**</font> i mai no <font style="color:red">**=>**</font> ni <font style="color:red">**=<**</font>).

Les operacions d’igualtat (**==**) i diferència (**!=**) tenen dues versions, aquesta i l’estricte (**===** i **!==**). La diferència principal entre aquests dos tipus d’igualtats i desigualtats són que en la no estricta no es té en compte el tipus. Així, <ode>1 == '1'</code> és true (veritat) i, en canvi, <code>1 === '1'</code> és false perquè <code>1</code> i <code>'1'</code> són de tipus diferents. Habitualment, serà més adequat fer servir les operacions estrictes.

Tingueu en compte les diferències entre igualtat i desigualtat estrictes o no estrictes (bàsicament quan tots dos operands tenen el mateix valor i tipus seran estrictament iguals, i quan no tenen el mateix valor ni el mateix tipus seran estrictament desiguals)


#### 🧐 Exemple 

```javascript
let a=1;
let b="1";
let c=1;
let d=2;
console.log(a==b); //true (same value)
console.log(a===b);//false(same value but different type)
console.log(a===c);//true(same value and same type)
```


### Operadors aritmètics

En el cas del tipus Number, els operadors aritmètics són els que “manipulen” dades numèriques. Els símbols dels operadors aritmètics són: 
* **<code style="color:blue">+</code>**	(suma)
* **<code style="color:blue">-</code>**	(resta)
* **<code style="color:blue">*</code>**	(multiplicació)
* **<code style="color:blue">/</code>**	(divisió)
* **<code style="color:blue">%</code>** (mòdul (residu de la divisió)) 

La coma decimal es representa amb un “**.**”. 

#### 🧐 Exemple 

```javascript
//We can assing the value on declaring the variable
let start_seconds = 103;
let minutes;
let remaining_seconds;

//gets the integer part of the result
minutes = parseInt(start_seconds/60); 

console.log(minutes);

//gets the rest of the division
remaining_seconds = start_seconds % 60; 
console.log(remaining_seconds);
```

### Operadors lògics

Serveixen per fer operacions lògiques a l’hora d’avaluar una operació.
Els símbols dels operadors són: 
* **<code style="color:blue">&&</code>** La **i** lògica (AND):  per obligar que es compleixi almenys una sentència.
* **<code style="color:blue">||</code>** La **o** lògica (OR): per exigir que es compleixin diverses sentències.
* **<code style="color:blue">!</code>**	(NOT): per negar una sentència lògica.

I els valors true i false s’escriuen en minúscula. 

Amb els **valors booleans** es poden fer operacions, això és, combinar diferents condicions certes o falses entre si i veure si el resultat és cert o fals. Els valors booleans es combinen entre si d’acord amb la lògica binària i amb tres operadors lògics: **no** <code>!</code> (negació), **i** <code>&&</code> (es llegeix com a _i lògic_) i **o** <code>||</code> (_o lògic_).

Podem veure totes les combinacions dels valors **cert**(<code>true</code>) i **fals**(<code>false</code>) amb aquestes operacions en una única taula que es coneix amb el nom de **taula de veritat**:


#### 🧐 Exemple 

```javascript
console.log (!true); // -> false
console.log (!false); // -> true
console.log (true && true);// -> true
console.log (true || true);// -> true
console.log (true && false);// -> false
console.log (true || false);// -> true
```

### Operadors de cadena de caràcters

Com ja hem vist a la unitat anterior, el tipus **String** es representa entre cometes dobles o simples, per exemple: <code>"Joan"</code> o <code>'Joan'</code>. 
Podem concatenar diverses dades de tipus string amb l’operador aritmètic <code>+</code>.

> Per tenir en compte: com ja s’ha comentat abans i, com a conseqüència de la filosofia de llibertat que ens proporciona JavaScript, amb la pràctica, us adonareu que és un **llenguatge feblement tipat**, és a dir, que un valor prèviament definit d'un tipus pot ser tractat com d'un altre tipus; per exemple, una cadena pot ser operada com un nombre.  

Per això mateix, si concatenem un Number amb un String amb l’operador de concatenació <code>+</code>, JS fa la conversió automàtica de l’enter a String, per això no cal tenir funcions predefinides de número a caràcter.

#### 🧐 Exemple 

```javascript
let StringNum= "1";
let NumberNum = 1;
let Concatenate = StringNum + NumberNum;
console.log(Concatenate); //Result will be a string with value "11" 
typeof(Concatenate); //String
```

### Operador condicional (ternari)

Aquest cas d’operador és fonamental que l’arribeu a comprendre correctament, ja que ens pot estalviar moltes definicions de condicionals al nostre codi.

La manera de funcionar d’aquest operador és la següent: avalua una **condició** (al primer operand) i, si és veritable, assigna el **valor1** (el segon operand); si no ho és, assigna el **valor2** (el tercer operand).

#### 🧐 Exemple 

```javascript
let country = "Mexico";
//shows "Language: Spanish" because we are in Mexico
console.log(country == "Mexico" ? "Language: Spanish" : "Language: English"); 
country = "USA";
//shows "Language: English" because we are in USA
console.log(country == "Mexico" ? "Language: Spanish" : "Language: English");
```

### Operadors unaris

Per definició, són els **operadors que tan sols necessiten un operand**; aquest curs veurem operadors unaris d’increment i de decrement

- <code>++</code> **Increment**: afegeix una unitat al seu operand. Si s’empra com a operador prefix (<code>++x</code>), retorna el valor del seu operand després d'afegir-ne una unitat; si es fa servir com a operador sufix (<code>x++</code>), retorna el valor del seu operand abans d'afegir-ne una unitat.

- <code>--</code> **Decrement**: resta una unitat al seu operand. Si s’empra com a operador prefix (<code>--x</code>), retorna el valor del seu operand després de restar-ne una unitat; si es fa servir com a operador sufix (<code>x--</code>), retorna el valor del seu operand abans d'afegir-ne una unitat.

Com podeu comprovar, l'anteposició o la posposició a la variable d'aquests operadors unaris és molt important, sobretot quan apareixen combinats en altres operacions, ja que, de manera general, quan posposem l'operador serà efectiu després d'executar l'operació, mentre que, si l’anteposem, primer aplicarà l'operador unari i després es calcularà l'expressió general amb el valor nou de la variable. Ho veurem més clarament amb un exemple:

#### 🧐 Exemple

```javascript
let a = 1;
let b = 2;
console.log(a++ + b);//returns the calculated value, then increases [a]  
console.log(a);
a = 5;
console.log(++a + b);//increases the value of [a] and then calculates the operation
console.log(a);
```
