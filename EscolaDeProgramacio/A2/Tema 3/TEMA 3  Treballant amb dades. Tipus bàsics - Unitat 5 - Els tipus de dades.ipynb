## <font color='#000078'><center>Unitat 5: Els tipus de dades</center></font>

Els **tipus de dades** són atributs que indiquen al programa (i al programador) informació sobre la classe de valors que es manejaran; això comporta imposar restriccions a les dades, com ara quins valors poden prendre i quines operacions es poden fer.

Podem parlar d’uns tipus de dades principals a JavaScript que són: **String**, **Boolean** i **Number**. i que ara passem a explicar, tot i que també en tenim d’altres que veurem més endavant en unitats més avançades.

### String

Representa qualsevol caràcter alfanumèric: una lletra, un dígit (xifra), un espai o un símbol, per exemple: la lletra 'a' o el dígit '3' o el signe d’interrogació '?'. Per distingir-los dels números, s’escriuen normalment entre cometes simples o dobles (sempre considerant que hem d’emprar el mateix tipus de cometes a l’obertura i al tancament) : <code>'a'</code>, <code>'3'</code>, o <code>"@"</code>.

També és de tipus string un conjunt de caràcters, és a dir: un text, i també s’escriuen entre cometes (simples o dobles igualment que els caràcters individuals): <code>"Això és un text"</code>, </code>'una cadena de caràcters'</code>.

Amb els caràcters no podem fer operacions aritmètiques, òbviament; no els podem sumar ni restar, etc., però sí que els podem concatenar. 
No és el mateix el número enter 3 que el caràcter ‘3’, encara que per nosaltres puguin tenir el mateix significat.

Particularment aquí, amb els **String** tenim la propietat **length** per obtenir la llargària del nombre de caràcters d’un String. És a dir, si volem saber quants caràcters té un literal definit a una variable, només cal que invoquem la propietat **length** d’aquesta manera:

#### 🧐 Exemple 

```javascript
let village = "Cardona";
console.log(village.length); // -> 7
```


### Number
El tipus Number representa un nombre positiu o negatiu, enter o real.

Així doncs, per exemple podem declarar nombres enters o reals de la següent manera respectivament:

#### 🧐 Exemple 

```javascript
let integerNumber = 100; //integer number
let realNumber = 100.33; //real number
```

Com a cas particular d’aquest tipus de dades, tenim la paraula clau <code>NaN</code>, que ens indica que el valor que estem mirant de treballar com un tipus **Number** no ho és realment:

#### 🧐 Exemple 

```javascript
let a = 4 - 'hello';
console.log(a); // NaN
```

### Boolean
Afegirem un darrer tipus bàsic importantíssim: el tipus **BOOLEÀ**. 
El tipus booleà té el seu origen en l’_àlgebra de Boole_, ideada pel matemàtic _Georges Boole_ al segle XIX. Treballa amb dos valors lògics binaris, **cert** i **fals**, que amb JavaScript estan representats com a <code>true</code> i <code>false</code> , respectivament (i que, com a curiositat, es corresponen amb els valors numèrics **1** i **0**), i que són la base de la computació.
Aparentment, és un tipus que pot semblar molt poca cosa perquè només pot tenir dos valors: **1** (<code>true</code>) i **0** (<code>false</code>). I ja està. Però malgrat que és un tipus peculiar, és imprescindible, per exemple, per poder prendre decisions en un algorisme. 
En lògica, aquest tipus es fa servir per indicar la certesa o falsedat d’un fet (d’un enunciat, diríem en lògica).


### Altres tipus de dades

Particularment, tenim el cas de **null** i d'**undefined**. 

Pel que fa a **undefined**, en l’àmbit de les variables, podríem dir que una variable que és **undefined** és perquè és una variable a què **no s'ha assignat un valor**, i això, en JavaScript, ho hem d’entendre com que és una variable que no ha estat definida.

#### 🧐 Exemple 

```javascript
//this variable, since we have not yet assigned it a value will be undefined
let noValueVar;

console.log(noValueVar); // ->undefined

noValueVar = null;
console.log(noValueVar); // ->null
```


Dit això, recordeu que JavaScript és case-sensitive i haurem d’emprar aquestes paraules clau sempre en minúscules.

Finalment, tenim el tipus **Object**, que és un cas particular, ja que aquest tipus podrà contenir qualsevol cosa que puguem definir a JavaScript i que la funcionalitat del qual podrem veure en unitats més avançades

### Conversions de tipus de dades

El llenguatge ens dona la possibilitat de saltar-nos les restriccions que ens imposen els tipus de dades i ens facilita eines de conversió entre uns tipus i els altres. 

Així doncs, tenim la possibilitat de convertir qualsevol valor definit en una variable a un tipus de dades String emprant mètodes propis del llenguatge com ara toString(), que sempre ens tornarà un nou valor convertit a tipus String del contingut de la variable sobre la qual l’apliquem.

#### 🧐 Exemple 

```javascript
let firstNumber = 100;
console.log(firstNumber.toString()); //will throw "100"
let secondNumber = 100.33;
console.log(secondNumber.toString()); //will throw "100.33"
let myBoolean = true;
console.log(myBoolean.toString()); //will throw "true"
```

Així mateix, és important entendre com funcionen els mètodes **_parseInt()_** i **_parseFloat()_**, atès que sovint se solen fer servir per convertir **tipus de dades** per obtenir un **Number** amb les característiques d’un enter o d’un real respectivament.

Aquí teniu dos exemples per acabar d’entendre'n el funcionament:

#### 🧐 Exemple de parseInt

```javascript
console.log(parseInt("100")); //will throw 100
console.log(parseInt("100.00")); //will throw 100
console.log(parseInt("100.33")); //will throw 100
console.log(parseInt(100.33)); //will throw 100
console.log(parseInt("100 55 6.6")); //will throw 100
console.log(parseInt(" 100 ")); //will throw 100
console.log(parseInt("100 years")); //will throw 100
console.log(parseInt("He was 100")); //will throw NaN
```

#### 🧐 Exemple de parseFloat

```javascript
console.log(parseFloat(100));//will throw 100
console.log(parseFloat("100"));//will throw 100
console.log(parseFloat("100.33"));//will throw 100.33
console.log(parseFloat("100 550 600"));//will throw 100
console.log(parseFloat("He was 100"));//will throw NaN
```

### Literals

Pel que fa als **Literals**, aquí els heu d’entendre com la “_**part de la dreta de la igualtat quan definim una variable**_”. És a dir, si definim, per exemple, un **String** de la manera següent:
#### 🧐 Exemple 

```javascript
let myVillage = "Cardona";
```

Tindrem que village és una variable de tipus **String** i **"Cardona"**; és el **literal** que en defineix el valor. Aquesta definició és aplicable a tots els tipus de dades del llenguatge JavaScript i que s’especifiquen a la Guia de Mozilla.
