### Esquema de recorregut d’una seqüència

Pensem com dissenyar un patró que ens permeti recórrer tots els elements d’una seqüència i anar-los tractant un a un de manera ordenada, des del primer fins a l’últim…

Per recórrer tota la seqüència ens haurem de situar en el primer element, és a dir, a la primera posició de la seqüència. Després, haurem d’accedir a l’element i així successivament anirem passant per tots els elements de la seqüència fins que arribem a l’últim element. Ens caldrà, doncs, una **estructura iterativa**.

> L’esquema segueix sempre una mateix estructura:  
> - **1** Obtenir el primer element.  
> - **2** Inicialitzar les variables.  
> - **3** Iterar fins a arribar al final i, dins la iteració: 
>    - **3.1** Fer el tractament que calgui a cada element per resoldre el problema.
>    - **3.2** Passar a l'element següent.
> - **4** Si cal, calcular i mostrar el resultat final un cop tractades totes les dades de la seqüència.

##### Sintaxi:




#### 🧐 Exemple 

> Seguint amb l’exemple dels alumnes que hem definit a l’aula, pensem ara que ens demanen obtenir tots els alumnes que són repetidors; doncs ens haurem de posar a la primera posició i recórrer l’Array que hem definit abans (classroomStudents) per anar comprovant cadascun dels alumnes i, si són repetidors, desar-los a un nou Array anomenat, per exemple, *classroomRepeaters*, per això emprarem l’estratègia de recórrer l’Array amb una estructura iterativa (while) en què ens haurem situat a la primera posició (<code>let i=0;</code>):


```javascript
//Declaring the Array that represents the students in the classroom:
let classroomStudents = new Array();
//Pushing students on the classroom: 
//First
classroomStudents.push ({ 
    name: "Carme", 
    surname: "Ibáñez Marchena", 
    yearOfBirth: 1998, 
    averageGrade: 8.3,
    repeater: false
});
//Second
classroomStudents.push ({ 
    name: "Joan", 
    surname: "Espí Prats", 
    yearOfBirth: 1999, 
    averageGrade: 7.8,
    repeater: true
});
//Third
classroomStudents.push ({ 
    name: "Maria", 
    surname: "Massats Perelló", 
    yearOfBirth: 2003, 
    averageGrade: 4.5,
    repeater: false
});
//Fourth
classroomStudents.push ({ 
    name: "Ramon", 
    surname: "Cugat Llopis", 
    yearOfBirth: 2000, 
    averageGrade: 5.7,
    repeater: true
});
//Fifth
classroomStudents.push ({ 
    name: "Anna", 
    surname: "Bernal Valls", 
    yearOfBirth: 2002, 
    averageGrade: 3.5,
    repeater: false
});
//Declaring the new Array that will represents the repeaters in the classroom:
let classroomRepeaters = new Array();
//go to the beginning of the array
let i = 0; 
//CAUTION: Array classroomStudents has to be defined as in the previous exercise
//and we will iterate until we reach the end of it
while(i<classroomStudents.length){
    //we check if the student in position i is a repeater or not
    if(classroomStudents[i].repeater) {
        //if so we store it in the new array
        classroomRepeaters.push(classroomStudents[i]);  
    }
    i++;
}
console.log(classroomRepeaters);
```

> Com veieu, amb aquest programa hem pogut fer un recorregut per tots els elements de la taula. El mateix codi podria servir per llegir qualsevol seqüència, encara que tingués milers d’elements. Per exemple, si s'amplia el nombre d’alumnes de la classe, de 5 a 50, aquest mateix codi ens valdria perfectament per recórrer la nova seqüència, però l’esquema seria exactament el mateix. O, si el que volem és obtenir els alumnes que han nascut abans de l’any 2000, l’esquema continua sent vàlid perquè en tots els casos fem el mateix: un recorregut per tots els elements de la taula.

#### 🧐 Exemple 

> D’altra banda, hem de comentar aquí que JavaScript té una eina molt potent per fer recorreguts als Arrays: **forEach**; tal com s’explica a la unitat **6.4 Iterant Arrays**, **forEach** ens permet recórrer cadascun dels elements d’un Array (és a dir, iterar un Array), tot i que, les principals diferències d’aquesta estratègia són que 
> 1. **No necessitarem posicionar-nos al primer element de l’Array**, ja que forEach directament fa el recorregut per tots els elements que componen l’Array (independentment de si estan a la posició 0 o no, els recorrerà tots). 
> 2. Per això mateix tampoc **no ens és necessari iterar fins a arribar a la mida de l’Array** (atès que, com diem, forEach ja s’ocuparà d’iterar tots els elements que són a l’Array, és a dir, podem tenir elements a les posicions 1, 2, 4, 6 i amb forEach iterarem directament aquests 4 elements sense haver de gestionar posicions de l’Array que estan amb valor undefined, que en aquest cas serien les posicions 0, 3 i 5.  
> 
> Per completar aquest exemple us deixem la manera d’obtenir els alumnes nascuts abans de l’any 2000 emprant el forEach:


```javascript
//Declaring the new Array that will represents the students born before 2000:
let classroomBefore2000 = new Array();
//iterating using forEach
//CAUTION: Array classroomStudents has to be defined as in the previous exercise
//and we will iterate directly all elements in the array
classroomStudents.forEach( function (myStudent) {
    //we check if the student is born before 2000
    if(myStudent.yearOfBirth < 2000) {
        //if so we store it in the new array
        classroomBefore2000.push(myStudent);    
    }
});
console.log(classroomBefore2000);
```
