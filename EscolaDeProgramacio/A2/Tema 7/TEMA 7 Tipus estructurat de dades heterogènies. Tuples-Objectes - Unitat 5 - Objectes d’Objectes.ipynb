### Objectes d’Objectes

Com ja s’ha comentat, els objectes ens permetran **agrupar diferents tipus de dades**. 

Això vol dir que **un objecte pot contenir qualsevol altra entitat que puguem definir al llenguatge**, fins i tot **un altre objecte** o un **Array**, així, com ja hem vist als exemples anteriors, si volem accedir a una propietat de l’objecte per definir, per exemple, els cognoms com un Array de dos strings ho podem fer sense cap problema.

Pensem ara quin avantatge ens pot donar definir un objecte dintre d’un altre, ja que, a l’exemple amb què estem treballant podem emmagatzemar, per exemple, un Array amb tots els llibres que té llogat un usuari i accedir a les seves propietats perfectament per obtenir o desar informació emprant la **Notació de Punt** que ja hem vist:

#### 🧐 Exemple 

```javascript
//Declaring the the Array that represents the books of the library:
let librayBooks = new Array();
//Pushing books on the library: 
//First
librayBooks.push ({ 
    title: "La plaça del Diamant", 
    year: 1989, 
    author: "Rodoreda, Mercè", 
    editorial: "Pagès Editors",
    hardCover: true,
    rented: false
});
//Second
librayBooks.push ({ 
    title: "El elegido", 
    year: 1999, 
    author: "Thomas Mann", 
    editorial: "Ediciones Altaya S.A.",
    hardCover: true,
    rented: false
});
//Third
librayBooks.push ({ 
    title: "Aleph", 
    year: 2011, 
    author: "Coelho, Paulo", 
    editorial: "La Butxaca",
    hardCover: false,
    rented: false
});
 
//Declaring the the Array that represents the archiver:
let librayUsers = new Array();
//Assignment of new user: 
librayUsers.push ({ 
    name: "Pere", 
    surnames: { 
        firstSurname : "Llopis", 
        secondSurname : "Deltell" 
    }, 
    genre: "M", 
    age: 35, 
    books: new Array()
});
//Assignment of new user in the second position (1): 
librayUsers.push ({ 
    name: "Marta", 
    surnames: { 
        firstSurname : "Vidal", 
        secondSurname : "Castell" 
    }, 
    genre: "F", 
    age: 22, 
    books: new Array()
});
 
 
//Now we'll assingn the third book as rented by the first user:
librayUsers[0].books.push(librayBooks[2]);
//And define this book as rented:
librayBooks[2].rented = true;
 
//And now we'll assingn the first and second book as rented by the second user:
librayUsers[1].books.push(librayBooks[0]);
librayUsers[1].books.push(librayBooks[1]);
//And define this books as rented:
librayBooks[0].rented = true;
librayBooks[1].rented = true;
 
 
//Showing all library users and the books they has rented, using nested two WHILE
let stringAllUsers = "";
let i = 0;
while(i<librayUsers.length){
    let j = 0;
    let stringUsersBooks = "";
    while(j<librayUsers[i].books.length){
        stringUsersBooks += "             " + librayUsers[i].books[j].title + "(" + librayUsers[i].books[j].author + ")" + "\n";
        j++;
    }
    stringAllUsers += librayUsers[i].name + " " + librayUsers[i].surnames.firstSurname + " " + librayUsers[i].surnames.secondSurname + " \n    RENTED BOOKS: \n" + stringUsersBooks + "\n";
    i++;
}

console.log(stringAllUsers);
console.log(librayBooks);
```
