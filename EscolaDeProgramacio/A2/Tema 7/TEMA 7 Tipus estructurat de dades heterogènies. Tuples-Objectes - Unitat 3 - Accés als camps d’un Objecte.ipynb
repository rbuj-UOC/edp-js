### Accés als camps d’un Objecte

Fonamentalment, tenim dues maneres d’accedir a les propietats (o camps) d’un Objecte: la **Notació de Punt** i la **Notació de Claudàtors**.

Les **propietats** d’un **Objecte** seran accessibles tant per consultar el seu valor com per establir-lo.

#### Notació de punt: 
Emprant la **Notació de punt** podrem accedir a una **propietat** emprant l'estratègia següent: <code>nom_objecte.nom_propietat</code>

També podem navegar amb aquesta jerarquia de **Notació de punt** per tots els objectes continguts dintre d’altres objectes i accedir a totes les seves propietats.

Com hem dit, per fer referència als valors de cada camp (**propietat**) d’un Objecte cal indicar el nom de la variable de tipus Objecte seguit d’un punt i del nom del camp. Un exemple de la seva sintaxi en JavaScript seria el següent:

#### 🧐 Exemple 

```javascript
//Declaring the user object with properties:
let user = {
    name: "Albert",
    surname: "Mas",
    gender: "M",
    age: 23,
    hasBook: true,
}
console.log(user.name); //Albert
```

Seguint aquesta jerarquia de Notació de punt, podem dir que les expressions següents seran <font color="green">correctes</font>:

```javascript
console.log(user.name); //Accessing the user's name
console.log(user.age); //Accessing the user's age
console.log(user.hasBook); //Checking if the user has books on loan
```

I que, pel contrari, aquestes expressions seran <font color="red">incorrectes</font>:

```javascript
console.log(name); //ERROR: we cannot refer to the property without specifying the Object's name
console.log(user01.age); //ERROR: we cannot refer to a non-existent object
console.log(hasBook.user);//ERROR: the command must be the name of the object and then property and not the other way around
```

Acabem de veure com **obtenir els valors de les propietats** d’un Objecte;  per **definir els valors d’una propietat** d’un objecte també farem servir aquesta jerarquia de **Notació de punt**. És a dir, una vegada definit un Objecte, podem establir els valors de les seves propietats (o modificar-los sempre que ens sigui necessari):

```javascript
user.name = "Marta"; //assigning the new name "Marta" to the name property of the user
user.gender = "F";//changing gender's user to female
user.surname = "Vidal";//assigning the value of the new surname
console.log(user.name + " " + user.surname + " GENDER : ["+ user.gender +"]");//showing this property through the console
```

#### Notació de claudàtors: 

Anàlogament a la notació de punt, podem navegar per aquesta Jerarquia amb el nom de l’element tancat entre claudàtors <code>[]</code>. Ho veurem amb un exemple en què definim un objecte usuari amb un cognom que és un altre objecte:

#### 🧐 Exemple 

```javascript
//Declaring the user object with properties and methods:
let userClaudators = { 
    //PROPERTIES
    name: "Pere", 
    surnames: { 
        firstSurname : "Llopis", 
        secondSurname : "Deltell" 
    }, 
    genre: "M", 
    age: 35, 
    hasBook: false  
};
console.log(userClaudators);
```

Amb els claudàtors podem accedir a les propietats i els mètodes de la manera següent:

```javascript
//ACCESSING PROPERTIES
console.log(userClaudators['surnames']['firstSurname']); //"Llopis”
```

> Recordeu que **podem emprar aquestes estratègies d’accés a la jerarquia tant per consultar els valors ja definits com per establir-los**.
