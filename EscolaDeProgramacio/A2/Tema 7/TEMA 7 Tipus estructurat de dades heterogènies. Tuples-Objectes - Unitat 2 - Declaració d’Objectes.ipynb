### Declaració d’Objectes

De la mateixa manera que vam veure amb els **Array**, un **Objecte** és una construcció que agrupa un conjunt de dades diverses, tot i que, com ara veurem, les emmagatzemarem de maneres diferents en els Arrays i els Objectes. 

De cada **Objecte** podem definir-ne **propietats** que donen molta potencialitat de codificació; indicant les propietats (els camps) ja podem fer servir els objectes per representar objectes de la vida quotidiana.

Per declarar un objecte n'hi haurà prou d'emprar una **declaració de variable** amb una assignació que contingui una **apertura i tancament de claus** <code>{}</code> (normalment entre aquestes claus hi aniran les propietats de l’objecte):

#### 🧐 Exemple 


```javascript
//Declaring the user object:
let user1 = {};
console.log(user1);
```

Tot i que, el més habitual és fer una declaració de l’objecte especificant-ne les seves propietats separades per comes (,) de la manera següent: <code>nom_de_la_propietat : valor_de_la_propietat</code>:

```javascript
//Declaring the user object with properties:
let user2 = {
    name: "Albert",
    surname: "Mas",
    gender: "M",
    age: 23,
    hasBook: true,
}
console.log(user2); //object 
```
