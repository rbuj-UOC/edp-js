### Arrays d’Objectes

Fins ara hem vist com treballar amb una fitxa d’un usuari de la biblioteca, però, habitualment, a la biblioteca hi ha un arxivador ple de fitxes. 

De fet, hi ha tantes fitxes com usuaris registrats té la biblioteca. 

Aprofitant el concepte de *taula* (Array) que hem vist a la unitat anterior, podem combinar aquests dos tipus i donar solució a aquesta situació.

#### 🧐 Exemple 

> Segons el que estem dient, ara aquest seria l’arxivador de la biblioteca:
> 
> Imaginem que a la biblioteca, en principi, hi ha un nombre indeterminat d’usuaris registrats i, per tant, a l’arxivador hi anirem afegint fitxes amb les seves dades. Tot seguit veurem com ho podem representar en JavaScript, per exemple, per als primers dos usuaris. En què els definirem i després modificarem el camp hasBook (té llibre) del primer usuari i, finalment els llistarem per consola:

```javascript
//Declaring the the Array that represents the archiver:
let librayUsers = new Array();
//Assignment of new user in the first position (0): 
librayUsers[0] = { 
    name: "Pere", 
    surnames: { 
        firstSurname : "Llopis", 
        secondSurname : "Deltell" 
    }, 
    genre: "M", 
    age: 35, 
    hasBook: false
}
//Assignment of new user in the second position (1): 
librayUsers[1] = { 
    name: "Marta", 
    surnames: { 
        firstSurname : "Vidal", 
        secondSurname : "Castell" 
    }, 
    genre: "F", 
    age: 22, 
    hasBook: true
}

//Modifying the hasBook property of the first user
librayUsers[0].hasBook = true;

//Showing all library users defined so far, using WHILE
let stringAllUsers = "";
let i=0;
while(i<librayUsers.length){
    stringAllUsers += librayUsers[i].name + " " + librayUsers[i].surnames.firstSurname + " " + librayUsers[i].surnames.secondSurname + " - HAS BOOK?: " + (librayUsers[i].hasBook ? " YES " : " NO ") + "\n";
    i++;
}

console.log(stringAllUsers);
```

Per anar afegint objectes a un Array que ja tenim inicialitzat tenim la instrucció <code>push</code> de JavaScript que ens permet afegir un element a la darrera posició d’un Array. Així, doncs, si al codi anterior li volem afegir un nou usuari, ho podríem fer de la manera següent sense necessitat de saber quina és la posició immediatament lliure de l’Array (d’això ja se n’ocupa la instrucció <code>push</code>):

```javascript
//Creating the new user:
let newUser = { 
    name: "Carles", 
    surnames: { 
        firstSurname : "Martí", 
        secondSurname : "Sarrià" 
    }, 
    genre: "M", 
    age: 15, 
    hasBook: false
}
//Assignment of new user in the next free position of the Array: 
librayUsers.push(newUser);
 
//Showing all library users defined so far, using WHILE
let stringNewAllUsers = "";
let j=0;
while(j<librayUsers.length){
    stringNewAllUsers += librayUsers[j].name + " " + librayUsers[j].surnames.firstSurname + " " + librayUsers[j].surnames.secondSurname + " - HAS BOOK?: " + (librayUsers[j].hasBook ? " YES " : " NO ") + "\n";
    j++;
}
console.log(stringNewAllUsers);
```
