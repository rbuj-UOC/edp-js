### Introducció als Objectes

Fins ara, hem estudiat els tipus bàsics (enters, reals, caràcters, booleans, etc.) i hem vist com podem agrupar dades en taules (Arrays).
 
De vegades, però, per representar la realitat ens caldria poder agrupar elements de tipus diferents però que es refereixen a un mateix concepte, d’una manera més estructurada. Així podrem definir tipus de dades que s’adaptin millor a les dades del món real que el problema ha de tractar.

Un objecte és un **conjunt de dades**, anomenades **propietats**, que ens permeten agrupar informació diversa. Cadascuna d’aquestes propietats té un nom únic que l’identifica dins l’Objecte. 

Així, amb els Objectes podem emmagatzemar informació relacionada entre si que ens servirà per representar la realitat.

Els Objectes es componen de diverses **propietats**, que poden ser de tipus diferents i que, en conjunt, representen un **concepte estructurat proper a la realitat que volem representar**. D’aquesta manera, el programa és més llegible i més entenedor, ja que aconseguim una abstracció de la realitat més ajustada al que es necessita per resoldre el problema. 

### Ens quins casos ens són útils els Objectes? 

Quan el programa ha de tractar dades compostes (formades per altres dades) i que no es poden definir amb els tipus de dades bàsics de Javascript. Així podrem crear variables a la mida de les necessitats del programa i més adequades al concepte o objecte real que representen.
