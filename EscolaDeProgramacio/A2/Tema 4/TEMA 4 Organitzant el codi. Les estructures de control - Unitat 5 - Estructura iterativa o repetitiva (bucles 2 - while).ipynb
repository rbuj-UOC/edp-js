### Introducció 

Cada repetició que fem en un bucle l’anomenem **iteració**. Les estructures iteratives permeten repetir l’execució del bucle mentre es compleix una determinada condició. A la unitat anterior hem vist la instrucció **for**, que té un funcionament molt particular.

L'estructura iterativa següent que estudiarem és **while**. La seva sintaxi és aquesta:


##### Sintaxi:
> **while (** <code>condició</code> **)**  
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>accions;</code>  
**}**  

Les accions només es repetiran si la condició es compleix. I com que la condició és el primer que s’avalua, podria passar que les accions no s’executessin mai.


Vegem-ne un exemple. Farem un programa que calculi el factorial d’un nombre donat. 

#### 🧐 Exemple 


```javascript
let factorialOf = 5;
let i = 1;
let factorial = 1;
while (i <= factorialOf){
    console.log("i value: "+i);
    factorial = factorial * i;
    i++; //We increase the value of current integer by 1
}     
console.log("Factorial of "+ factorialOf + ": " + factorial);
```

Fes proves amb aquest codi, canvia el valor de *factorialOf* i executa de nou el programa per veure què passa. Prova a posar 0 a *factorialOf*. Quants cops s’executen les instruccions que hi ha dins de l’estructura while?


#### 🧐 Exemple 

En la mateixa línia de raonament pensem que volem fer una determinada operació un nombre de vegades determinat, per exemple mostrar per la consola separadament totes les lletres d’una cadena de text emmagatzemada a una variable, això amb un **while** ho tenim fet:

```javascript
let myText = "JAVASCRIPT!";
let i = 0;
while (i <= myText.length-1){
    console.log(myText[i]);
    i++; //We increase the value of current integer by 1
}     
```
