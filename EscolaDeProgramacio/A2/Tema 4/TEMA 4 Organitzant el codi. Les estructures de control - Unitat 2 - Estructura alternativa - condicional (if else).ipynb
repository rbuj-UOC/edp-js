### Introducció

En aquest apartat començarem amb el tractament de les estructures de control de flux. Començarem per les estructures de control condicionals, en què veurem els dos tipus de sentències condicionals que emprem a JavaScript: if...else i switch

L’estructura seqüencial que hem vist a la unitat anterior ens permet definir (seguir) un camí (un flux d’execució) per resoldre el problema, en què una instrucció s’executa darrere l’altra, un camí únic. A vegades, però, necessitem que un algorisme faci determinades accions només si es compleix una certa condició, o que faci unes accions si es compleix una condició i, si no es compleix, que en faci unes altres. Per a això es fa servir l’**estructura alternativa o condicional**.

Aquest concepte d’accions condicionades s’il·lustra molt bé amb l’algorisme que seguim per travessar un pas amb semàfor: si és en verd, travessem; si és en vermell, esperem. I, així, podríem pensar en molts altres exemples. 

Farem servir l'estructura alternativa quan vulguem executar una acció o una altra depenent de si es compleix una determinada condició (per això es diu també *composició condicional*), és a dir, quan vulguem variar el flux d’execució del nostre algorisme per definir diversos camins alternatius.

### Estructures condicionals senzilles (if)

Heu de tenir clar que una sentència condicional és un bloc que s’executarà si una condició és vertadera. 

##### Sintaxi:
> **if (** <code>condició</code> **)**  
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>acció;</code>   
**}**  

L’expressió més senzilla d’una sentència condicional seria la següent, en què executem un bloc de codi si la condició expressada és certa:

#### 🧐 Exemple 


```javascript
controllet country = "EEUU";
let currency;
if(country === "EEUU") 
{
    currency = "Dollar";
} 
console.log(currency);
            
```

Simplement, el que fem és comprovar si es compleix certa condició i, si es compleix, executem conseqüentment unes accions definides en un bloc.

> L’**estructura condicional** representa una part del codi en què necessitem comprovar si alguna condició és certa o falsa abans d’executar les accions d’aquesta part. I ens permet indicar quina acció cal fer.

Les decisions en els algorismes es prenen fent comparacions entre variables, números, caràcters o fent servir expressions booleanes, en definitiva, expressions que donen com a resultat cert o fals.

Les **expressions booleanes** es fan servir per determinar quin flux d’execució (quina “ruta”) ha de seguir el programa depenent de si el resultat de l’expressió és cert o fals. D’aquí, que tant les expressions algebraiques com el tipus booleà siguin tan importants en programació.

### Estructures condicionals dobles (if else)

##### Sintaxi:
> **if (** <code>condició</code> **)**   
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>acció1;</code>   
**}**  
**else**   
**{**   
&nbsp;&nbsp;&nbsp;&nbsp;<code>acció2;</code>  
**}**  

El següent avenç lògic és pensar què hauríem de fer si volem executar unes altres accions si aquesta condició no es compleix. Per a aquest cas sorgeix la paraula clau **else**, amb què controlem els casos que no entrin dintre d’aquesta condició:

#### 🧐 Exemple 

```javascript
let country= "UK";
let currency;
if(country == "EEUU")
{
    currency = "Dollar";
} 
else 
{
    currency = undefined;
} 
console.log(currency);
```

#### 🧐 Exemple 

> Han començat les rebaixes a la nostra botiga de roba i cal ajustar el sistema
de càlcul del preu final de cada peça. Una feinada! Sort que podem fer un
algorisme senzill que ens simplificarà la feina. La nostra promoció és la
següent: a tots els articles els apliquem un 25 % de descompte, però si
es compren més de tres peces, llavors s’aplica un 50 % a totes. Una
ganga!
> 
> 1. **Entenguem el problema**
Tenim dues dades d’entrada: el preu de la compra i el nombre d’articles
totals de la compra. I, com a sortida, ens demanen que calculem l’import
després d’aplicar les rebaixes.
> 1. **Pensem com resoldre’l**
Decidim els càlculs a fer per obtenir el que demana l’enunciat:</br>
• El tipus de descompte és 25 % si el nombre de productes és d’1 a 3 i del 50 % si és de més de 3.</br>
• Import descomptat = import x tipus descompte.</br>
• Import final = import - import descomptat.</br>
> 1. **Fem l'algorisme:**

```javascript
const discontType1 = 25;
const discontType2 = 50;

let initialAmount, totalItems;// -> input data
let discontedAmount, finalAmount;//  -> output data

initialAmount = 120; //change this value for testing
totalItems = 4; //change this value for testing

if (totalItems <= 3){
    discontedAmount = initialAmount * (discontType1 / 100); 
} else {
    discontedAmount = initialAmount * (discontType2 / 100);     
}

finalAmount = initialAmount - discontedAmount;
console.log(finalAmount);
```

L'avenç lògic ens porta a pensar que no tot és blanc o negre, no sempre estem davant un problema amb una solució única, més aviat sempre tenim diferents possibilitats per resoldre, per això tenim la possibilitat d’encadenar-li un **if** a l’**else**, això es coneix com **estructures imbricades o niuades**, què ens permet continuar avaluant una nova condició si l’anterior no s’ha complert.

##### Sintaxi:
> **if (** <code>condició1</code> **)**  
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>acció1;</code>   
**}**  
**else if(**<code>condició2</code>**)**   
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>acció2;</code>  
**}**  

#### 🧐 Exemple 


```javascript
let country= "UK";
let currency;
if(country === "EEUU")
{
    currency = "Dollar";
} 
else if (country == "UK")
{
    currency = "Pound";
} 
console.log(currency);
```

Aquest encadenament de **else if** el podem fer servir per comprovar si la variable té un valor determinat, fent servir un **if** per a cada valor. Podem, a més, afegir un **else** final en cas que el valor de la variable no es correspongui amb cap dels que hem comprovat als **if**:


```javascript
let country = "Vietnam";
let currency;
if(country == "EEUU")
{
    currency  = "Dollar";
} 
else if (country == "UK")
{
    currency = "Pound";
} 
else if (country=="Japan")
{
    currency = "Yen";
} 
else 
{
    currency = undefined;
} 
console.log(currency);
```

No oblideu mai que per fer una comparació hem d’emprar com a mínim <code>==</code> , preferiblement <code>===</code> (recordeu els operadors de comparació de la unitat anterior).

D’altra banda, cal comentar que quan fem una sentència condicional tenim la possibilitat de no emprar les claus <code>{}</code> quan definim el bloc si aquest es resumeix tan sols a una sola línia de codi, tot i que aquesta pràctica no és gens recomanable, ja que pot conduir-nos a errades i a un codi que no sigui ben intel·ligible, dit això, el següent codi seria vàlid:


```javascript
let country = "UK";
let currency;
if(country == "EEUU")
    currency = "Dollar";
else 
    currency = undefined;
console.log(currency);
```
