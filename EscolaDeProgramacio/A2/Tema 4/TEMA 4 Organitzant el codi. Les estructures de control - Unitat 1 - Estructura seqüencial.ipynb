### Introducció

L'estructura seqüencial és la més senzilla de totes, simplement indica a l’ordinador que ha d'executar de manera consecutiva una llista d'accions (que poden ser altres estructures de control); per construir una seqüència d'accions només cal escriure cada acció en una línia diferent. 

Així doncs, cada instrucció s’executa una darrere l’altra de manera ordenada. Els passos per resoldre el problema es van fent un darrere l’altre en l’ordre en què estan escrits en l’algorisme. Així doncs, tot i que pot semblar una estructura molt simple, és primordial en programació.

Tots els exemples de codi que heu vist en aquests materials són estructures de control seqüencial. Vegem, però, un altre exemple: definir dues variables per fer una suma (operand1 i operand2), fer aquesta suma i mostrar-la per la consola. Això es tradueix en una sèrie d’ordres que executarem seqüencialment (independentment de com es produeixin els resultats durant la seva execució) per obtenir un resultat final.

Aquest codi quedaria així (posant valors a les variables):

#### 🧐 Exemple 


```javascript
controllet op1 = 5, op2 = 8; // We define the two variables that we will add
let sum = op1 + op2; // We do the sum and save the result
console.log(sum); // We write the result on the console
```

### Els Blocs

Com ja anirem veient, a JavaScript sovint ens interessarà agrupar conjunts d’instruccions. Per fer aquestes agrupacions farem servir les claus <code>{ }</code>.

Així, si tenim dos **blocs** d’instruccions que fan coses diferents, els podrem agrupar amb claus.

Un exemple:

#### 🧐 Exemple 

```javascript
{
  let op1 = 5, op2 = 8;
  let sum = op1 + op2;
  console.log(sum);
}
{
  let op3 = 7;
  let op4 = 9;
  let mult = op3 * op4;
  console.log(mult);
}
```

Un parell d’apunts sobre aquest exemple: 

1. Hem creat dos blocs per fer dues operacions diferents.
1. Perquè visualment es puguin diferenciar bé els blocs, *indentem* sempre el codi que hi ha dins d’un bloc.

>  *Indentar*: Acció d’endinsar els marges d’un paràgraf, de tal manera que la seva amplada disminueixi.
https://www.enciclopedia.cat/gran-enciclopedia-catalana/indentacio

A JS tenim una estratègia per acotar qualsevol estructura que puguem identificar amb certa entitat pròpia (seqüencial o no) i què ens servirà com a fonament per començar amb aquests tipus de sentències, això és el concepte de **bloc**. A JavaScript es defineix una **sentència de bloc** bàsicament amb l’emprament de les claus <code>{}</code> per emmarcar el seu començament i la finalització. 

Amb el que hem vist fins ara podríem definir en la pràctica un bloc ràpidament: tot i que haurem d’anar amb compte amb l’àmbit de les variables, atès que una variable pot actuar de manera diferent definida fora d'un bloc i dintre d'aquest encara que tingui el mateix nom.

Podem provar el codi següent per veure com funcionen els blocs i com podem definir variables locals dintre de distints blocs amb el mateix nom:

#### 🧐 Exemple 

```javascript
let globalVariable = "first definition of globalVariable";
//BLOCK 1
{
    console.log ("BLOCK 1:" + globalVariable );
    let localVariable = "definition of localVariable at BLOCK1";
    console.log ("BLOCK 1:" + localVariable );
    globalVariable = "redefinition of globalVariable at BLOCK1";
}
//BLOCK 2
{
    console.log ("BLOCK 2:" + globalVariable );
    let localVariable = "definition of localVariable at BLOCK2";
    console.log ("BLOCK 2:" + localVariable );
    globalVariable = "redefinition of globalVariable at BLOCK2";
}
console.log("GLOBAL:"+globalVariable );
```
