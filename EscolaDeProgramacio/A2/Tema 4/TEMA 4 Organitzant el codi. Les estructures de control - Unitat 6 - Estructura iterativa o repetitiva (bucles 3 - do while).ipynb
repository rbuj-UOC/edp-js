### Estructures iteratives (do while)

És fonamental que entengueu com funciona la sentència **while** ja que, amb aquest mateix raonament ens adonarem que **do..while** fa pràcticament el mateix tot i que neix d’un cas particular que **while** no preveu.

La diferència principal i més important que hauríem de tenir en compte quan fem servir una composició amb **do..while** en comptes d'una amb **while** és en el moment en què fem la comprovació de la condició.

En un bucle **do..while**, l'avaluació de la condició és al final del bucle. Això vol dir que el codi que tenim dins del bucle s'executarà almenys una vegada abans que la condició sigui avaluada.


##### Sintaxi:
> **do**  
**{**   
&nbsp;&nbsp;&nbsp;&nbsp;<code>accions;</code>  
**} while (** <code>condició</code> **)**  

#### 🧐 Exemple 

> L'exemple següent és aclaridor: penseu que volem llançar dos daus un nombre indeterminat de vegades fins que coincideixi el mateix resultat en tots dos daus, doncs si, almenys, no llancem una vegada el daus abans d’inicialitzar el bucle mai no podrem fer aquesta primera comprovació i saber si hem de continuar llançant.
>  
> El que ens plantegem és que sempre (passi el que passi) haurem de fer almenys un primer llançament, ja que, en funció del resultat, continuarem llançant o no:

```javascript
let turn= 0;
let givenScore1;
let givenScore2;
do
{
    turn++; //We increase the number of throws by 1
    console.log("Turn nº:" + turn);
    
    givenScore1 = parseInt(Math.random() * (7 - 1) + 1);
    givenScore2 = parseInt(Math.random() * (7 - 1) + 1);
    console.log("-> DICE_1 scored [" + givenScore1 + "] points on the pitch " + turn);
    console.log("-> DICE_2 scored [" + givenScore2 + "] points on the pitch " + turn);
        
    //Let's see if we have exceeded the maximum score, if so we warn that we continue to throw.
    if(givenScore1 != givenScore2){
        console.log("-> PERFECT !! we keep throwing.");
    
    } else{
        console.log("-> Ohhh !! We got the same value on both dice ...");
    }
} while (givenScore1 != givenScore2);
```

### Conclusió

Una vegada vistes les tres maneres que tenim per fer iteracions (**for**, **while** i **do while**) podem concloure que **while** és la més estàndard de les tres i que és capaç de preveure totes les situacions en què empraríem les altres dues. 
Fonamentalment, es pot dir que quan emprem **while** només es passa a executar el contingut de l’acció si es comprova que la condició és vertadera, fet que pot passar 0, 1 o més vegades. 
D’altra banda, **do while** funciona de manera similar, però ens assegurem que el contingut sigui executat almenys una vegada, és a dir, que encara que la seva condició no es compleixi, el seu contingut s'executa.
I finalment, **for**, actua de manera semblant a while, malgrat que podem dir que tenim més centralitzades la variable amb què iterem i el lloc on definim les condicions d’iteració (que és sempre a la capçalera d’aquesta instrucció).
