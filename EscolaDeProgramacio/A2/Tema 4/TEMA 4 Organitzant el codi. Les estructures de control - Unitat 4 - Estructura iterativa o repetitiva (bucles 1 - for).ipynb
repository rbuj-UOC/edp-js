## <font color='#000078'><center>Unitat 4: Estructura iterativa o repetitiva (bucles 1: for)</center></font>

Un dels principals objectius i aplicacions de la informàtica en general (també de la programació en particular) és el d’automatitzar processos, és a dir, definir un algorisme per solucionar un problema i tenir la possibilitat d’executar-lo tantes vegades com calgui: per aquest motiu, als llenguatges de programació surten estructures de control de flux destinades a gestionar aquestes repeticions: les **estructures iteratives** o bucles. 

Amb l’estructura condicional ja es pot resoldre un nombre més gran de problemes, però encara ens cal anar més enllà. Sovint, cal que un algorisme repeteixi una acció (o un conjunt d’accions) unes quantes vegades fins que es compleixi una determinada condició. Per a això, es fa servir l’**estructura iterativa o repetitiva**. 

Aquest concepte d’accions repetides es representa amb la idea de bucle. I s’il·lustra molt bé amb una pila d’algorismes que seguim de manera quotidiana: beure aigua mentre tinguem sed, defensar fins que recuperem la pilota, pujar la muntanya fins que arribem al cim, repetir la sèrie d’abdominals tres cops cada dia, córrer fins a arribar a la meta, baixar esglaons fins que arribem a peu d’escala; i així, molts d’altres. 

Farem servir la composició iterativa quan vulguem repetir una acció que ja s’ha fet (és a dir, recular a una acció anterior) i tornar-la a fer, per això es diu també *composició repetitiva*. O, dit d’una altra manera, quan vulguem variar el flux d’execució del nostre algorisme per poder tirar enrere i executar diversos cops (iteracions) una part del codi.

Nosaltres farem servir les tres sentències de bucle següents: 
* Sentència <code>for</code>.
* Sentència <code>do</code>...<code>while</code>.
* Sentència <code>while</code>.

Penseu que totes tenen un objectiu comú: fer de manera senzilla la repetició d’una mateixa acció un nombre de vegades determinat, tot i que, com comprovareu, saber triar per a cada casuística determinada un tipus de bucle o un altre ens pot estalviar feina.


### Estructures iteratives (for)

Respecte d’aquestes estructures, us heu de quedar que sempre, a la definició d’una sentència **for**, tindrem tres expressions en aquest ordre:

##### Sintaxi:
> **for (** <code>expressió inicial</code> **;** <code>condició</code> **;** <code>increment</code> **)**   
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;<code>accions;</code>  
**}**    

En què habitualment:
* L’**expressió inicial** inclourà la declaració d’una variable índex i l’assignació d’un valor inicial.
* La **condició** farà referència a la variable índex.
* L’**increment** (o decrement) es farà sobre la variable índex.

El primer cop que s’executa el for, es crea la variable índex i se li assigna el valor inicial. A continuació, es comprova si es compleix la condició i, si és així, s’executen les accions que hi ha dins.
Quan s’han executat totes les accions, s’incrementa el valor de la variable índex, segons l’indicat a la instrucció.
Aleshores, es torna a comprovar si es compleix la condició i, si és així, es tornen a executar les accions, es torna a incrementar la variable índex i es torna a comprovar la condició. I així fins que la condició no es compleix.

Vegem-ne un exemple: volem mostrar 5 vegades un text per la consola. La manera, emprant una estructura iterativa for és bastant intuïtiva:

#### 🧐 Exemple 


```javascript
for(let i = 0; i < 5; i++){
     console.log("my first for");
}
```


Es fa servir la construcció **for** quan es coneix el nombre de vegades que s’ha de repetir el codi. És una estructura molt útil per treballar amb seqüències de dades, com es veurà més endavant.

Una construcció **for**, sempre es pot fer fent servir una composició **while**, però a l’inrevés no passa. Per això diem que la construcció **while** és més general que la construcció **for**.

L’exemple que hem vist de la composició iterativa **for** (que a cada volta incrementa el comptador en una unitat) és la més habitual. Però no respon a totes les situacions amb què ens podem trobar. Per exemple, si volem simular un compte enrere des d’un valor determinat fins a 0 o fer la suma de tots els nombres parells entre dos nombres inicials. 

La composició **for** permet també indicar com volem que variï el comptador a cada iteració.

#### 🧐 Exemple 

```javascript
//Countdown from 10 to 0
for(let i = 10; i >= 0; i--){
     console.log(i);
}
```

En darrer lloc, les estructures iteratives (tal com passava amb les condicionals) es poden combinar amb d’altres estructures iteratives i condicionals, fent servir una (o diverses) estructures una dins l’altra. És el que es coneix com a estructures **imbricades** o **niuades**, que ens permeten definir moltíssimes combinacions de fluxos d’execució.
