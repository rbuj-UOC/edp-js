### Estructures condicionals (switch)

Ja hem vist que l’estructura alternativa dona joc a moltes combinacions de l’execució del flux d’un programa quan fer una acció depèn d’una determinada condició. De vegades caldrà, fins i tot, fer servir una (o diverses) estructures alternatives una dins l’altra. És el que es coneix com a estructures *imbricades* o *niuades*, que ens permeten definir diversos camins alternatius. 

Hi ha, però, una altra variant que serveix només en casos molt concrets en què es pot estalviar haver d’imbricar diverses estructures alternatives. Es tracta d’escollir entre diferents alternatives possibles depenent del valor d’una variable de tipus enter o caràcter.

Per a aquestes circumstàncies tenim la sentència SWITCH, que ens permetrà executar una part de codi en funció del resultat d’una expressió avaluada. 

##### Sintaxi:
> **switch (**<code>condició</code>**)**  
**{**  
&nbsp;&nbsp;&nbsp;&nbsp;**case** <code>valor1</code>**:**  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<code>acció1;</code>  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**break;**  
&nbsp;&nbsp;&nbsp;&nbsp;**case** <code>valor2</code>**:**  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<code>acció2;</code>  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**break;**  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;...  
&nbsp;&nbsp;&nbsp;&nbsp;**default:**  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<code>accióDefault;</code>  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**break;**  
**}**  

Seguint amb l’exemple de les monedes, podríem fer el mateix emprant aquesta sentència de la manera següent:

#### 🧐 Exemple 


```javascript
let country = "EEUU";
let currency;
switch (country) {
    case "EEUU":
    currency = "Dollar";
    break; //This keyword defines the end of the block
    case "UK":
      currency = "Pound";
      break; //This keyword defines the end of the block
    case "Japan":
      currency = "Yen";
      break; //This keyword defines the end of the block
    default:
      currency = undefined;
      break; //This keyword defines the end of the block
}
console.log(currency);
```

#### 🧐 Exemple 

Volem fer un algorisme que simuli una calculadora que fa les operacions
aritmètiques elementals (suma, resta, multiplicació i divisió) a partir de dos
números donats **a** i **b** enters positius i una operació que triarem definint un d'aquests quatre valors: A(*Addition*), S(*Substracion*), M(*Multiplication*), D(*Division*) a la variable **operation**. Podeu canviar aquestes tres variables per veure com es comporta el programa amb els nous valors:

```javascript
let a = 10;//first summand
let b = 5;//second summand
let result;//result of operation
//A(Addition)
//S(Substracion)
//M(Multiplication)
//D(Division)
let operation = "S"; 
switch (operation) {
    case "A":
      result = a + b;
      break; //This keyword defines the end of the block
    case "S":
      result = a - b;
      break; //This keyword defines the end of the block
    case "M":
      result = a * b;
      break; //This keyword defines the end of the block
    case "D":
      result = a / b;
      break; //This keyword defines the end of the block
    default:
      result = undefined;
      break; //This keyword defines the end of the block
}
console.log(result);
```

Aquesta estructura “busca” quina acció cal fer entre les diferents opcions (case) que es descriuen. L’opció ***default*** indica què cal fer si el valor de la variable no coincideix amb cap dels casos detallats.

Hi ha un punt important a tenir en compte amb aquesta instrucció. A diferència del que hem vist fins ara, dins d’un ***case***, no necessitem posar les claus per agrupar les instruccions. A canvi, hem d’acabar el bloc d’instruccions del ***case*** amb la instrucció ***break***. Aquesta instrucció fa que saltem fora de la instrucció ***switch*** sense comprovar cap altra opció.

El fet de necessitar la instrucció ***break*** per acabar el ***case*** té un avantatge: podem agrupar valors. Un exemple.
A la nostra empresa l’horari varia depenent del dia de la setmana, així, dilluns i dimecres l’horari és de 8 h a 12 h, i de 14 h a 18 h; dimarts i dijous l’horari és de 9 h a 13 h, i de 15 h a 19.30 h. Finalment, divendres l’horari és de 8 h a 14 h. En aquest cas podríem fer això:

#### 🧐 Exemple 

```javascript
let weekDay = "Thursday";
let morning, afternoon;
switch (weekDay) {
    case "Monday":
    case "Wednesday":
        morning = "8am to 12pm";
        afternoon = "2pm to 6pm";
        break;
    case "Tuesday":
    case "Thursday":
        morning = "9am to 1pm";
        afternoon = "3pm to 7:30pm";
        break;
    case "Friday":
        morning = "8am to 2pm";
        afternoon = "closed";
        break;
    default:
        morning = "closed";
        afternoon = "closed";
}
console.log("Working time. Morning: "+morning + " Afternoon: " + afternoon);
```
