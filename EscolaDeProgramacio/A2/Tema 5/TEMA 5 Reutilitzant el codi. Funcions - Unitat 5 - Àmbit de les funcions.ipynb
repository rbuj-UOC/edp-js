### Àmbit de les funcions

Hem d’entendre que, anàlogament al que hem vist quan definíem blocs d’estructures seqüencials amb les claus ({}), en què explicàvem que haurem d’anar amb compte amb les definicions de les variables i els valors que poden prendre, atès que una variable pot actuar de manera diferent definida fora d'un bloc i dintre d'aquest encara que tingui el mateix nom, doncs, de la mateixa manera, una funció defineix el **seu propi àmbit**: això vol dir que el que es defineix dintre d’una funció es queda a la funció, però això no vol dir que dintre d’aquest àmbit no puguem treballar amb variables definides fora d’aquest àmbit amb els seus valors, i, fins i tot, modificar-les.

La millor manera d’entendre aquest concepte és amb un exemple:

#### 🧐 Exemple 

```javascript
let brand1 = 'Ford'; 
let brand2 = 'Mercedes';
let longuestBrandResult = getLonguestBrand (brand1, brand2);  //We call the function
console.log ("Brand 1 = " + brand1); 
console.log ("Brand 2 = " + brand2); 
console.log ("Longuest Brand = "+ longuestBrandResult); 
function getLonguestBrand (param1, param2){ 
    let brand1; 
    if(param1.length > param2.length){ 
        brand1 = param1; 
    } else { 
       // The function will enter here and define brand1 within the scope of the function
        // as "Mercedes", although the variable brand1, outside the scope,
        // will not be modified and still contains "Ford"" 
        brand1 = param2; 
    } 
    return brand1; 
}
```

Com podem comprovar, tot i que definim la variable ***brand1*** dintre i fora de l’àmbit de la funció es comporten com dues variables diferents. Però, què passa si no declarem la variable ***brand1*** dintre de l’àmbit de la funció?. Doncs, si proveu d’eliminar d’aquest mateix codi, la línia vuitena, en què defineix la variable dintre de l’àmbit de la funció: <code>let brand1;</code> (o, simplement comentar-la) us adonareu que ara la funció redefineix el valor de la variable exterior al seu àmbit: 

```javascript
let brand1 = 'Ford'; 
let brand2 = 'Mercedes';
let longuestBrandResult = getLonguestBrand (brand1, brand2);  //We call the function
console.log ("Brand 1 = " + brand1); 
console.log ("Brand 2 = " + brand2); 
console.log ("Longuest Brand = "+ longuestBrandResult); 
function getLonguestBrand (param1, param2){ 
    // let brand1;
    if(param1.length > param2.length){ 
        brand1 = param1; 
    } else { 
       // The function will enter here and define brand1 within the scope of the function
        // as "Mercedes", although the variable brand1, outside the scope,
        // will not be modified and still contains "Ford"" 
        brand1 = param2; 
    } 
    return brand1; 
}
```
