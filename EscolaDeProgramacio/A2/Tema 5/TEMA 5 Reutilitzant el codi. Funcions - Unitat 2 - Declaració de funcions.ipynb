### Declaració de funcions

Bàsicament, si seguim aquesta estratègia, tindrem l'estructura següent:
1. Començarem sempre amb la paraula reservada <code>function</code> (obligatòria).
1. Seguidament definim el **nom de la funció** (obligatòria). Per definir aquests noms estarem sotmesos a les mateixes restriccions que per definir els noms de les variables.
1. Un llistat dels **paràmetres d’entrada** (que emprarem per passar-los al nostre codi en executar-lo) entre parèntesis i separats per comes (*paràmetre1*, *paràmetre2*, ...) (opcionals).
1.  El **codi que executarem** emmarcat entre claus <code>{}</code>.
1. Dins d’aquest codi podrem definir el **paràmetre de sortida** emprant la paraula reservada <code>return</code> (opcional).

> Sintaxi.
> 


#### 🧐 Exemple 

La millor manera de veure-ho és amb un exemple. Tot seguit definirem una funció que retorni el càlcul de la divisió donats un dividend i un divisor com a paràmetres d'entrada:


```javascript
// Declaration of a function with two input parameters and one output
function division(dividend, divider){ 
    return dividend / divider; 
}
//Calling the function with parameters 8 and 4 
let result = division(8, 4); 
console.log(result);
```

> És força habitual provar de **centralitzar totes les funcions** que declarem en un lloc concret del codi: al començament, al final... igual com amb les variables i les constants.
