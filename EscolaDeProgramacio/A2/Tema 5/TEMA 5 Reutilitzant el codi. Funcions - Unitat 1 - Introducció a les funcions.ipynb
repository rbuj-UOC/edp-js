### Introducció a les funcions

> D’una manera ràpida, podem definir una **funció** com un conjunt d’instruccions a què posem un nom per poder-les fer servir en més d’una ocasió. En aquests moments del curs, tot i que, potser no ens n'hem arribat a adonar, hem fet servir unes quantes funcions de ***JavaScript***. 

Això és el que hem estat fent, per exemple, quan fem servir ***console.log()*** o ***prompt()***: estem fent crides a funcions del sistema que executen cert codi predefinit (en aquests exemples, definit pel mateix llenguatge) quan el necessitem.

Tots els llenguatges de programació disposen d’un seguit de funcions predefinides que podem fer servir sense ni tan sols conèixer el codi que contenen. Són, de fet, programes que podem fer servir dins els nostres programes. Això ens dona molta potència perquè podem programar a partir de codi ja existent que es pot reutilitzar tantes vegades com calgui.

Per tant, tenim la possibilitat de **definir els nostres blocs de codi per executar-los quan i des d’on vulguem i tantes vegades com ho necessitem**. Així doncs, tan sols necessitarem un **nom** per definir la nostra funció, escriure el nostre **codi** a executar i definir els **paràmetres** que emprarem quan fem la crida (si és que els necessitem), així com el valor que retornarà en executar-se (si també necessitem que retorni).

Alguns punts a aclarir:

* El codi l’escriurem amb la mateixa estratègia que empràvem per definir els blocs amb les estructures de control, és a dir, amb claus <code>{}</code>.

* Els paràmetres són uns valors que posarem entre parèntesis i que farem servir dins de la funció. Al *console.log()* per exemple, el paràmetre és allò que volem que s’escrigui a la consola. 

* El valor de retorn és un valor que es podrà assignar a una variable un cop acabi la funció. Penseu en la funció *prompt()* Aquesta funció rep un paràmetre (el text que es presentarà a l’usuari) i retorna un valor (el que hagi introduït l’usuari).

Un avantatge important de les funcions és que permeten la resolució de problemes més complexos dels que hem vist fins ara. Un problema complex és el que no es pot resoldre directament, de cop, sinó que cal descompondre’l en subproblemes més petits i resoldre'n cadascun per separat.

Així doncs, una funció és només un fragment de codi, igual que un programa. L’única diferència és que pot rebre i retornar valors. D’aquesta manera es comunica amb el programa que la crida. 

Amb les funcions, tenim la possibilitat de definir els nostres propis blocs de codi per executar-los quan i des d’on vulguem i tantes vegades com ho necessitem. Per això necessitem un nom per definir la nostra funció, determinar el codi a executar (seguint la mateixa estratègia apresa per dissenyar algorismes) i definir els paràmetres que utilitzarem quan fem la crida, si és que necessitem passar valors a la funció (paràmetres d’entrada) i el valor que retornarà després d’executar-se (paràmetre de sortida).
