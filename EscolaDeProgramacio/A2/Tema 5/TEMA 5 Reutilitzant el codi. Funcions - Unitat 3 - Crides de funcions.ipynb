## <font color='#000078'><center>Unitat 3: Crides de funcions</center></font>

### Crides de funcions

A l’apartat anterior ja hem vist com procedim a fer una **crida a una funció** que tenim definida. Tot i això, hauríem de tenir en compte, amb relació a les crides de les funcions, les següents consideracions necessàries:

1. Penseu sempre que podrem definir una funció i mai no arribar a executar-la, ha de quedar-nos ben clar que:
> una funció, tot i estar definida al codi, mai no s’executarà si no la cridem des d’algun lloc del codi.

1. Com a conseqüència d’això, **l’emplaçament on es declari una funció al codi no és rellevant** (ja que no s’executarà fins que no es cridi des d’un altre lloc), pel contrari **l’emplaçament de les seves crides sí que ho és**.

1. Tot i això, com hem vist a la unitat anterior, és força habitual mirar de **localitzar totes les funcions** que declarem **en un lloc concret del codi** (al començament, al final..). 
	
1. Observeu com funcionen els **paràmetres** quan els passem a una funció a l’hora de fer les crides: bàsicament us heu de quedar amb el fet que, quan passem un paràmetre primitiu (*number*, *string*...), la funció rebrà el valor del paràmetre primitiu, mentre que si no ho és (per exemple, un tipus *Object* o *Array* que ja veurem més endavant), rebrà el paràmetre mateix, sobre el qual tindrem possibilitat de modificar-ne les propietats directament, i tindrà efecte, fins i tot, fora de l’àmbit de la funció.
