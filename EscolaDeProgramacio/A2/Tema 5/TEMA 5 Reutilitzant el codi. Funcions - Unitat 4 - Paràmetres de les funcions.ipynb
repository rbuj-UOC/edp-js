### Paràmetres de les funcions

> De vegades una funció necessita treballar amb dades que provenen de l’algorisme que la fa servir. 

Per exemple, la funció predefinida de JavaScript que ens mostra per pantalla el valor que se li passa, i que cridarem d’aquesta manera: <code>console.log("Hola")</code>.

> Altres vegades, la funció necessita tornar alguna dada a l’algorisme. 

Per exemple, la funció predefinida de JavaScript <code>parseInt</code> i que cridarem així: <code>parseInt(4.0)</code>, converteix el valor real que rep i retorna el valor enter equivalent, en aquest cas ens retornarà 4 (enter). 
Aquests valors que es passen a les funcions es diuen *paràmetres* i s’indiquen entre parèntesis just després del nom de la funció.

Els paràmetres que rep la funció es diuen **paràmetres d’entrada** i el que la funció retorna es diu **paràmetre de sortida**. Tal com es fa amb les funcions predefinides, els paràmetres de les funcions que definim nosaltres s’indiquen just després del nom de la funció entre parèntesis

#### 🧐 Exemple 
> Pensem que volem definir una funció que ens permeti accedir a informació sobre els cotxes que tenim però d'acord amb el perfil d’accés de l’usuari que empri aquesta funció es definirà d’una manera o d’una altra:

```javascript
let user = 'common'; //Let's try to change this value to admin.
let myCar = manageCar(user); 
console.log(myCar);

function manageCar (p_user){ 
    let car = ''; 
    if (p_user == 'admin'){ 
        car = 'BRAND: Honda - MODEL: Accord - YEAR: 1998';
    } else { 
        car = 'BRAND: Honda - MODEL: Accord - YEAR: unauthorized';
    } 
    return car; 
} 
```

> El codi anterior no permetrà mai que un usuari comú accedeixi a la informació de l’any de fabricació del vehicle. Si, pel contrari, accedim amb un usuari amb el perfil ‘admin’, la funció sí que ens permetrà consultar aquesta informació.
