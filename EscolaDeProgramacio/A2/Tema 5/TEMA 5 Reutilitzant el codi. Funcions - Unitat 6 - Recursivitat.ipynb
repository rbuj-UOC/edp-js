### Recursivitat

Entenem per *recursivitat* la **capacitat que té una funció de fer crides sobre si mateixa**.

Bàsicament hem d’entendre que una funció pot fer una crida a ella mateixa dintre de la seva declaració. Aquest concepte és fàcil d’entendre amb un exemple: 

#### 🧐 Exemple 
> pensem en un problema matemàtic de l’antiguitat, la **successió de Fibonacci**  https://ca.wikipedia.org/wiki/Successi%C3%B3_de_Fibonacci:  
Bàsicament, el codi d’aquest problema ens diu que:  
> * A la posició 0 d’aquesta successió el resultat ha de ser 0,  
> * A la posició 1 ha de ser 1  
> * I a la resta de posicions el resultat hauria de ser la suma del resultat de les dues posicions anteriors a l’actual.
> Amb aquests preceptes podem intuir que, el fet d'haver d’accedir al resultat de dues posicions anteriors en la crida d’una funció per donar el resultat ens facilitarà moltíssim poder cridar a la nostra funció des de la pròpia declaració. Així doncs:


```javascript
function fibonacci (position){ 
    // In the Succession we find the particularity that the first
    // position and the second are 0 and 1 respectively
    // (Unable to get value from previous two positions) 
    if (position==0 || position ==1){ 
        return position; 
    } 
    // In the other positions any value is the sum
    // of the value the two positions previous to the present one
    // which we get with a recursive call
    else { 
        return fibonacci(position-1) + fibonacci(position-2); 
    } 
} 
for (let i=0; i<=20; i++){ 
    console.log("fibonacci ", (i), " is:", fibonacci(i),"\n"); 
}
```

El fet de poder cridar a la funció des de la pròpia definició ens permet obtenir el resultat de la funció per determinats valors que han d’intervenir en el resultat que hem de tornar.
