# <font color='#000078'><center>TEMA 1: Començant a programar</center></font>
## <font color='#000078'><center>Unitat 3: Els llenguatges de programació</center></font>

A l’hora de programar, podem triar entre una pila de llenguatges, alguns més adequats per a segons quins tipus de programes o aplicacions que d’altres. Alguns més sofisticats i complicats d’aprendre, d'altres més eficients per fer segons què.



Els llenguatges de programació, tal com passa amb les llengües, són “vius”, evolucionen constantment. Així doncs, cal saber sempre on podem trobar la documentació de referència, les llibreries que hi ha, els manuals més actuals i els fòrums de dubtes per estar al corrent dels darrers canvis.
