### Introducció als Arrays

Els tipus de dades que hem vist fins ara, a mesura que treballem amb problemes més complexos són insuficients i ineficients. Per això fem servir un altre tipus de dades més estructurat que ens permetrà representar millor alguns d’aquests problemes. Es tracta de les **taules** o **Arrays**. 
Els **Arrays** serveixen per agrupar elements sota un únic nom.

Així doncs, 
> Un **Array** és un **conjunt de valors a què s’accedeix mitjançant un índex** (que indica la posició dins l’Array). 
Així mateix, a aquests Arrays els podrem donar un nom per referir-nos-hi i tindran una sèrie de propietats (com ara, per exemple, la seva longitud).
