### Iterant Arrays

Emprant les **estructures de control** que ja vam veure anteriorment, podem recórrer els elements d’un Array; així, si volem recórrer l’Array definit al nostre exemple de les visites al web, ho podrem fer de la manera següent:

#### 🧐 Exemple 

```javascript
//We create the array by defining elements:
let webVisitors1 = [256, 350, 231, 297, 183, 267, 321];

//Iterating the Array and obtaining values: 
for(i = 0; i<webVisitors1.length; i++){ //ATTENTION: the first position is ALWAYS 0
    console.log(webVisitors1[i]); 
}
```

Un dels inconvenients d’aquest mètode és que hem de conèixer la longitud de l’Array per saber exactament el nombre d’iteracions que hem de fer. Per això tenim una altra instrucció molt útil per iterar un Array: <code>forEach</code>.

D’aquesta manera, si fem la iteració d’aquest Array emprant forEach seria així


```javascript
//We create the array by defining elements:
let webVisitors2 = [256, 350, 231, 297, 183, 267, 321];

//Iterating the Array and obtaining values: 
webVisitors2.forEach(
    function(dailyVisits){ 
        console.log(dailyVisits);
    }
);
```

A priori pot semblar una mica estranya aquesta estructura, però si l’analitzem més a fons ens adonarem que simplement el que fem és agafar cada element amb el <code>forEach(</code>) i, dintre d’aquesta instrucció, definim una funció a què passem com a paràmetre el mateix element (<code>dailyVisits</code>) que després emprarem dintre de la funció.

L'avantatge principal d’aquest mètode (**forEach**) és que no ens cal conèixer a priori la longitud de l’Array ni tampoc l’índex amb què estem treballant en cada iteració, atès que **treballem directament amb cada element de l’Array**. D’altra banda, tenim la particularitat que no farem iteracions sobre elements buits, ja que aquí fem un recorregut sobre tots els elements de l’Array que hi ha (si l’Array té posicions que no estan definides **forEach** no les considerarà si no és que les hem definides nosaltres específicament com a undefined).
