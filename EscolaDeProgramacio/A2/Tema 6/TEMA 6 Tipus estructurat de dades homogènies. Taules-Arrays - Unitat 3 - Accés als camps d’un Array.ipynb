### Introducció

Una primera estratègia per omplir un Array acabem de veure-la: a l’hora de crear-lo definint-ne els elements. Tot i això és molt possible que, una vegada creat, tinguem la necessitat de definir més valors a les posicions de l’Array o canviar els que ja tenim definits o, simplement omplir un Array que hem definit com a buit.

Per referir-nos a cada element només haurem d’indicar el nom de l‘**Array** i la posició que l’element ocupa dins seu amb l'índex que tingui.

Així doncs, accedirem als valors mitjançant l’índex, tenint en compte que **l’índex del primer element d’un Array sempre és zero (0)**, és dir, les posicions d’un Array de longitud n són 0,1,2,3,...n-1. És molt habitual en els llenguatges de programació que l’índex comenci a comptar a partir del 0. 

Assignarem valors a l’Array accedint-hi de la mateixa manera:

#### 🧐 Exemple 

```javascript
//Creating the Array
let webVisitors = []; 
webVisitors.length = 7; 
//Filling our Aray
//CAUTION!: The first element is always 0 
webVisitors[0] = 256; 
webVisitors[1] = 350; 
webVisitors[2] = 231; 
webVisitors[3] = 297; 
webVisitors[4] = 183; 
webVisitors[5] = 267; 
webVisitors[6] = 321; 
console.log(webVisitors);
```

I què passa si ara definim un valor a la posició següent a la que hem establert com a longitud de l’Array en la seva definició?, és a dir, en el nostre cas, atès que l’hem definit amb longitud 7 aquesta seria la vuitena posició, és el mateix que dir <code>webVisitors[7];</code> (recordeu que comencem numerant per 0 i per això la vuitena posició la definim amb el 7).
La resposta és senzilla: doncs que l’Array s’amplia automàticament i tindrà una longitud de 8:


```javascript
webVisitors[7] = 400; 
console.log(webVisitors.length); //8
```

I si fem el mateix a la posició 10 (<code>webVisitors[9]</code>)?: doncs ara també s’amplia la seva longitud fins a 10, tot i que <code>webVisitors[8]</code> la deixarà buida:


```javascript
webVisitors[9] = 325; 
console.log(webVisitors.length); //10
console.log(webVisitors[8]); //undefined
```

> Com podeu comprovar, no tan sols per definir valors, sinó també quan obtenim els valors de les posicions d’un Array els accedim amb el seu índex per recuperar-los.
