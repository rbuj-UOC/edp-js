### Creació d’Arrays

A JavaScript tenim diverses maneres de crear un Array, tot i que bàsicament podem agrupar-les en dues:

#### 1.- Creació d’un Array definint els elements:

D’aquesta manera, inicialitzem l’Array alhora que n'estem definint els elements, dintre d’aquesta forma tenim tres estratègies diferents, i en totes procedim d’una manera semblant, definim el nom del nostre Array i la llista de valors que el compondran (que podran ser de qualsevol tipus del que hem vist fins ara).

Per posar-ne un exemple, pensem que volem definir el nombre de les visites al web durant tota una setmana:

#### 🧐 Exemple 

```javascript
//We create the array by defining elements:
let webVisitors1 = new Array(256, 350, 231, 297, 183, 267, 321); 
console.log(webVisitors1);
```

```javascript
//It is equivalent to:
let webVisitors2 = Array(256, 350, 231, 297, 183, 267, 321);
console.log(webVisitors2);
```

```javascript
//It is equivalent to:
let webVisitors3 = [256, 350, 231, 297, 183, 267, 321];
console.log(webVisitors3);
```

>La primera i segona manera són molt semblants i indica que podem prescindir de la paraula clau *new* per definir el nostre Array. La tercera és la manera més curta d’inicialitzar un Array, atès que emprem literals d’Array <code>[]</code> amb què el llenguatge interpreta que el que estem fent és definir un Array.

#### 2.- Creació d’un Array definint la longitud:

D’aquesta manera inicialitzem l’Array definint la quantitat d’elements que contindrà (malgrat que els valors en totes aquestes posicions quedaran sense omplir). 

Correlativament amb les formes anteriors:

#### 🧐 Exemple 

```javascript
//This definition will return an empty Array of length 256 
let webVisitors4 = new Array(256); 
console.log(webVisitors4.length); 
```

> Imaginem que som els gestors d’una important plataforma web i necessitem
tenir un control de les visites que rep el web cada mes. Per emmagatzemar
anv eficiència aquestes dades, segurament les enregistraríem d’una manera
similar a:

>Observeu que hem creat una taula amb el nom “Visites”, que té 30 camps, corresponents a cada dia del mes, i en cadascun d’aquests camps s’hi emmagatzema el nombre de visites rebudes aquell dia, que és un valor enter.  
>
>Observeu que fent servir un Array hem estat capaços d’emmagatzemar 30 valors en una sola variable. En canvi, sense el tipus Array hauríem d’haver creat 30 variables diferents de tipus enter, una per a cada valor. Així hem guanyat eficiència i claredat.  
>
>Fixeu-vos també que cada camp té assignat un índex, que és la posició d’aquest camp dins l’Array.  
Com representem aquest registre en JavaScript?  
>
>Podem representar aquesta estructura amb un Array i les dades a emmagatzemar (el nombre de visites diàries) seran els camps d’aquest Array:


```javascript
//Creating by defining the length of the Array (with empty values):
let webVisitors5 = new Array(30);

//We'll iterate the Array and set values: 
for(i = 0; i < webVisitors5.length; i++){ //ATTENTION: the first position is ALWAYS 0
    webVisitors5[i] = Math.floor(Math.random() * (100 - 10 + 1) + 10)
        //NOTE: You could use "window.prompt();" to define this values directly on console 
        //We have used a function of the language that calculates a number randomly: Math.random()
}
console.log(webVisitors5);
```
