## <font color='#000078'><center>Unitat 2: La gramàtica</center></font>

Pel que fa a la **gramàtica**, us hauríeu de quedar amb els conceptes següents:

### Comentaris

En JS, com en la resta de llenguatges de programació, tenim la possibilitat d’inserir comentaris dintre del codi per poder explicar parts del codi.
És un molt bon costum comentar el codi, en facilita la llegibilitat i les modificacions posteriors. Fem servir la doble barra **//** per indicar que és una línia de comentaris i que, per tant, no s’ha d’executar, i farem servir **/*   */** per englobar-hi comentaris que ocupen més d’una línia de codi


```javascript
// Online comment (no action on the code)

/*
This is a multi-line comment
nor will it produce any action on the code
*/
```

### Case-sensitive

**_JavaScript_** és **_case-sensitive_** (distingeix majúscules i minúscules). És a dir, tingueu sempre en compte com anomeneu les vostres variables perquè per a JavaScript no serà el mateix prova que Prova.

### El punt i coma (;) final

No oblideu mai el punt i coma al final de cada expressió. Tot i que a la consola ens funcionarà habitualment sense el punt i coma, penseu que no és l’únic lloc des d’on executem **_JavaScript_** i podria donar-nos problemes.
