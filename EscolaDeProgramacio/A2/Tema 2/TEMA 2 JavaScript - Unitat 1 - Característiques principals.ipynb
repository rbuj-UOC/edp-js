# <font color='#000078'><center>TEMA 2: JavaScript</center></font>
## <font color='#000078'><center>Unitat 1: Característiques principals</center></font>

Com ja s’ha comentat anteriorment, JS és un llenguatge de programació molt útil per crear codi que funcioni dins un HTML, per això es fa servir per donar interactivitat a les pàgines web i per demanar informació a l’usuari, per exemple. Actualment, també es poden crear servidors, aplicacions web i d’escriptori que funcionen amb JavaScript.

A la introducció de la “**Guia de JavaScript de Mozilla**” defineixen **_JavaScript_** formalment com a: “<span style="color:#777"><em>...un <b>llenguatge orientat a objectes</b>, lleuger i interpretat, amb funcions de primera classe, més conegut per ser el <b>llenguatge de scripting per a pàgines Web</b>, però també emprat en molts àmbits fora d'un navegador web.</em></span>”.

Fonamentalment heu d’entendre que si l’HTML s’empra per definir el contingut de les pàgines, i el CSS per especificar el disseny que tindrà aquest contingut, **_JavaScript_ l’emprarem per programar el comportament de tots aquests elements**, malgrat que avui en dia la seva evolució ha estat tan imparable que han sorgit frameworks (com ara _jQuery_) i plataformes basades en aquest llenguatge que, fins i tot, s’empren fora de l’àmbit de les pàgines web (per exemple _NodeJS_).

En aquesta unitat farem referència a la **gramàtica del llenguatge** (és a dir, com hem d’escriure per tal que el nostre codi sigui interpretat correctament) i als **tipus de dades** (les dades amb què podrem treballar en **_JavaScript_**).


Com a concepte a aclarir, caldria comentar que, parlant de JavaScript, sovint trobarem referències a l'especificació **ECMAScript**. Pel que fa a aquesta especificació, l’heu d’entendre com una mena d'estàndard (o abstracció) en el qual està basat el JavaScript, que és el que nosaltres en realitat estudiarem.
