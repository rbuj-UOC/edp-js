## Unitat 3: Classes abstractes



Una classe abstracta en programació orientada a objectes no és més que una classe que no pot ser instanciada directament. És a dir, una classe de la qual no podem crear objetes. Quan en una unitat anterior hem explicat el concepte d'herència ho hem fet a partir d'aquesta jerarquia de classes:

> `Vehicle` ← `Car`, `Motorcycle`, `Airplane`, `Rocket`, `Bicycle`

Però... té sentit crear un objecte de la classe `Vehicle`? Què és exactament alguna cosa que identifiquem simplement com un vehicle? Quines propietats i mètodes té? Si el vehicle funciona amb quatre rodes i un volant en una carrosseria tancada amb portes i finestres, segurament es tractarà més aviat d'una instància de `Car`, és a dir, un cotxe, un objecte de tipus `Car`. Si té dues rodes i es condueix mitjançant un manillar, segurament estarem davant d'un objecte de la classe `Bicycle` o `Motorcycle`, en funció de si té motor o no en té. I també trobarem algunes característiques que ens permetin identificar què és un objecte de tipus `Airplane` o `Rocket`. Però hi ha alguna cosa que simplement sigui un `Vehicle` sense més ni més?

En un altre exemple hem explorat la jerarquia de classes dels animals:

> `Animal` ← `Cat`, `Dog` (i en podríem tenir més: `Bird`, `Fish`, `Snake`, `Spider`...)

Novament, hi ha algun tipus d'animal que simplement sigui un animal, no un gat ni un gos, ni cap altre animal concret conegut, sinó simplement un animal? 

I hi ha algun tipus d'esdeveniment social que no sigui ni un concert, ni una conferència, ni un partit, ni una exposició d'art, ni un taller... ni cap altre tipus específic d'esdeveniment, sinó senzillament un esdeveniment? En què consistiria aquest esdeveniment? Si el podem explicar i definir, llavors no serà un esdeveniment genèric, sinó un tipus d'esdeveniment concret al qual podrem posar un nom més específic (és a dir, crear una subclasse). Si no el podem explicar, llavors no té sentit crear un objecte d'aquest tipus.

> `GenericEvent` ← `Concert`, `Conference`, `SportsGame`, `ArtExhibition`, `Workshop`

En tots aquests exemples, `Vehicle`, `Animal` i `GenericEvent` són classes abstractes. No tenen sentit per si mateixes, sinó que són classes base de les quals deriven altres classes més específiques. No podem crear objectes d'aquestes classes abstractes, sinó que hem de crear objectes de les classes filles, que són les que tenen sentit per si mateixes.

El propòsit principal d'una classe abstracta és servir com a classe base de la qual altres classes poden derivar. Una classe abstracta pot contenir mètodes abstractes, que són declarats però no implementats, deixant així la implementació específica a les classes derivades. Això permet establir un esquelet o una plantilla per a les classes filles, assegurant que implementin determinats comportaments, però sense especificar com ho han de fer.

Les classes abstractes tenen un paper vital en l'arquitectura de programari en facilitar un disseny modular i escalable. Emprant classes abstractes, els desenvolupadors poden establir contractes clars per a les funcionalitats compartides, al mateix temps que proporcionen la llibertat d'implementar detalls específics en les classes derivades segons calgui.

En la majoria de llenguatges de programació orientada a objectes, les classes abstractes es defineixen utilitzant la paraula clau `abstract` en la declaració de la classe. En JavaScript, tanmateix, no existeix aquesta paraula clau ni existeix el concepte de classe abstracta de manera nativa. Tanmateix, podem simular el comportament de les classes abstractes en JavaScript, per exemple implementant mètodes que llencin errors si no se sobreescriuen en les subclasses. Fer-ho d'aquesta manera és una convenció que segueixen molts programadors, però no una característica del llenguatge. Vegem-ho amb un exemple:

```javascript
// Simulated abstract class. 
class GenericEvent {
  // The constructor throws an error if the class is instantiated directly.
  constructor() {
    if (this.constructor === GenericEvent) {
      throw new Error('Abstract class GenericEvent cannot be instantiated directly.');
    }
  }

  // These methods must be implemented by subclasses.
  schedule() {
    throw new Error("Method 'schedule()' must be implemented.");
  }

  cancel() {
    throw new Error("Method 'cancel()' must be implemented.");
  }

  reschedule() {
    throw new Error("Method 'reschedule()' must be implemented.");
  }
}

// Subclasses inherit the abstract methods and can override them as needed.
class Concert extends GenericEvent {
  schedule() {
    console.log('Scheduling a concert...')
  }

  cancel() {
    console.log('Canceling the concert...')
  }
}
```

```javascript
// Example usage. Trying to instantiate the abstract class will throw an error.
const myEvent = new GenericEvent();
```

```javascript
// Example usage. Instantiating a subclass works as expected.
const myConcert = new Concert();
myConcert.schedule();
```

```javascript
// Example usage. Executing abstract methods will throw an error if not overridden.
myConcert.reschedule();
```

Com podem comprovar, el mètode constructor de `GenericEvent` dona un error si se l'intenta invocar directament. Per això quan pretenguem crear un objecte de la classe `GenericEvent` mitjançant el constructor `new GenericEvent()` obtindrem un error. Tanmateix, si creem un objecte de la classe `Concert` o de qualsevol altre subclasse de `GenericEvent`, el constructor de la classe filla s'executa sense problemes. 

Després, el mètode `schedule()` de l'objecte `myConcert` funciona amb normalitat, ja que es tracta d'un mètode implementat (o redefinit o sobreescrit) a la classe filla `Concert`. En canvi, si intentem executar el mètode `reschedule()` sobre el mateix objecte `myConcert` tornarem a obtenir un error, ja que es tracta d'un mètode abstracte que no està implementat a la classe filla `Concert`. 

La classe `Concert` redefineix els mètodes `schedule()` i `cancel()`. Així, si executem aquests mètodes partint d'un objecte d'aquesta classe s'utilitza la definició de `Concert` i funciona correctament. En canvi, el mètode `reschedule()` no s'ha redefinit a `Concert`, de manera que si executem aquest mètode s'utilitza directament el codi de la superclasse `GenericEvent`, que genera un error.

### Classes abstractes en C#

Com vàrem fer a la unitat anterior, vegem breument com quedaria la sintaxi del mateix codi anterior en un llenguatge com C# que sí que suporta classes abstractes de manera nativa:

```javascript
// 🔴 Please pay attention to the fact that this code is not JavaScript, but C#.

// Abstract class.
public abstract class GenericEvent
{
  public abstract void Schedule();
  public abstract void Cancel();
  public abstract void Reschedule();
}

// Concrete class.
public class Concert : GenericEvent
{
  public override void Schedule()
  {
    Console.WriteLine("Scheduling a concert...");
  }

  public override void Cancel()
  {
    Console.WriteLine("Canceling a concert...");
  }
}

// Example usage. Trying to instantiate the abstract class will throw an error.
GenericEvent event = new GenericEvent();

// Example usage. We can instantiate the concrete class.
GenericEvent concert = new Concert();
concert.Schedule();
concert.Cancel();
```

Novament, tot i que la sintaxi de C# és diferent a la de JavaScript, podem reconèixer amb facilitat els elements més importants en el codi anterior. Per exemple la paraula clau `abstract` en la declaració de la classe `GenericEvent` indica que es tracta d'una classe abstracta. També observem que per indicar que una classe hereta d'una altra, C# utilitza el símbol `:` en comptes de la paraula clau `extends` com en JavaScript. I finalment, la paraula clau `override` en la declaració dels mètodes `Schedule()` i `Cancel()` indica que estem sobreescrivint la implementació d'aquests mètodes a la subclasse `Concert`. Tots aquests aspects són diferències en la sintaxi, però el concepte de classe abstracta és comú a tots els llenguatges de programació orientada a objectes.
