## Unitat 4: Interfícies

Una interfície en programació orientada a objectes és un contracte que defineix un conjunt de mètodes i propietats que una classe ha d'implementar. En certa manera, una interfície actua com una plantilla que estableix com s'ha d'estructurar una part del codi, sense arribar a proporcionar una implementació concreta. Les interfícies ens permeten tenir més flexibilitat que l'herència, ja que una classe pot implementar diverses interfícies, però únicament pot heretar d'una classe. 

Les interfícies ens permeten, a més, definir comportaments que poden ser compartits per classes que no tenen cap relació entre si. Per exemple, una interfície `IFlying` pot definir un mètode `fly()` de tal manera que qualsevol classe que implementi aquesta interfície, com `Airplane` o `Bird`, haurà de proporcionar la seva pròpia implementació de `fly()`.

Tot i que també podríem definir una classe `Flying` que implementi el mètode `fly()` i fer que `Airplane` i `Bird` heretin d'aquesta, això no seria una bona pràctica, ja que `Airplane` i `Bird` no tenen cap relació entre si més enllà del fet que ambdós volen. A la primera unitat d'aquest tema, vàrem esmentar que l'herència ens permet seguir el principi de la **relació "és una"** entre objectes. En aquest cas està clar que la relació que volem modelar no és del tipus "ser alguna cosa", sinó més aviat del tipus "ser capaç d'alguna cosa".

Les interfícies permeten seguir el principi de la **relació "pot fer"** o **relació "capaç de"** entre objectes. L'herència defineix el que un objecte "és" en termes de la seva essència i característiques fonamentals. En canvi, les interfícies defineixen el que un objecte "pot fer" o les capacitats que pot implementar, independentment de la seva naturalesa o tipus. És el que passa amb la interfície `IFlying`: l'apliquem a diferents classes de molt diferents naturaleses (com un avió i un ocell), que comparteixen la capacitat de volar.

Vegem-ne un altre exemple. La interfície `IDrawable` pot requerir un mètode `draw()`, i classes com `Circle` o `Square` el poden implementar segons les seves característiques específiques. En aquest cas, aquestes dues classes sí que podrien tenir una relació d'herència amb una superclasse `Shape`, ja que ambdues són tipus de formes geomètriques. Tanmateix, la interfície `IDrawable` ens permet definir un comportament que pot ser compartit per classes que no tinguin relació d'herència amb `Shape`, com per exemple `Image`, `Chart`, o `UserInterface`.

Les interfícies són crucials per aconseguir un disseny de programari net en programació orientada a objectes, ja que ens permeten portar a terme una adequada separació i abstracció de comportaments.

> <br>🚨&nbsp; **Atenció**
>
> Per convenció, en alguns llenguatges de programació els noms de les interfícies solen començar amb una `I` majúscula, com `IFlying` o `IDrawable`. Això resulta especialment útil en llenguatges com C# o Java, on les interfícies són un tipus de dada que es pot utilitzar per declarar variables, i la `I` majúscula ayuda a distingir les interfícies de les classes. No obstant això, també hi ha corrents que defineixen que és una mala pràctica utilitzar aquesta `I` majúscula en els noms de les interfícies. Us podeu trobar, per tant, amb ambdues convencions en el món de la programació.
> <br><br>

En JavaScript, tanmateix, no hi ha les interfícies de manera nativa en la seva sintaxi o estructura de llenguatge, com sí que ho fan en altres llenguatges de programació amb tipatge estàtic com Java o C#. No obstant això, es poden simular els conceptes darrere de les interfícies en JavaScript utilizant patrons de disseny coneguts com a _duck typing_, on es posa l'enfocament en la capacitat dels objectes per dur a terme certes accions més que en el seu tipus específic. Aquest nom tan curiós ve de la frase "si camina com un ànec i sona com un ànec, llavors deu ser un ànec". Aplicat al nostre exemple anterior, si un objecte té un mètode `fly()`, llavors podem dir que és capaç de volar, independentment del seu tipus. És a dir, si té el mètode `fly()`, deu ser que implementa la interfície `IFlying`.

Tot i que aquest enfocament _duck typing_ no proporciona la verificació en temps de compilació que ofereixen les interfícies, permet una flexibilitat similar en l'estructuració del codi.

En codi es pot implementar el patró _duck typing_ de la manera següent:

```javascript
// Class Airplane with fly() method.
class Airplane {
  fly() {
    console.log("I'm an airplane and I'm flying!");
  }
}

// Class Bird with fly() method.
class Bird {
  fly() {
    console.log("I'm a bird and I'm flying!");
  }
}

// Class Superhero with fly() method.
class Superhero {
  fly() {
    console.log("I'm a superhero and I'm flying!");
  }
}

// Class Fish with swim() method.
class Fish {
  swim() {
    console.log("I'm a fish and I'm swimming!");
  }
}

// Array of objects. Some of them can fly, some of them cannot.
const randomObjects = [new Airplane(), new Bird(), new Superhero(), new Fish()];

// Iterate over the array and call fly() method if it exists.
randomObjects.forEach((object) => {
  if (typeof(object.fly) === "function") { 
    object.fly();
  }
})
```

No obstant això, aquesta construcció `typeof(object.fly) === "function"` és bastant molesta i poc llegible. Per això, la podem encapsular en una funció que ens permeti comprovar si un objecte de qualsevol tipus implementa una interfície determinada:

```javascript
Object.prototype.can = function(methodName) {
     return ((typeof this[methodName]) === "function");
};
```

En aquest codi estem estenent el prototipus de `Object` en JavaScript amb un nou mètode anomenat `can()`. Aquest mètode agafa un `methodName` com a argument i verifica si l'objecte actual (referenciat per `this`) té una propietat amb aquest nom de mètode que sigui de tipus exacte `function`. D'aquesta manera, el nou mètode `can()` ens indica si l'objecte "pot" dur a terme una determinada acció o no.

Aquest mecanisme és una forma d'implementar una comprovació de capacitat similar a la que ofereixen les interfícies en llenguatges amb tipatge estàtic, però utilitzant la flexibilitat de JavaScript. Tanmateix, hem de tenir present que modificar el prototipus de `Object` global pot portar a conflictes i efectes secundaris no desitjats, per la qual cosa és una estratègia que s'ha d'utilitzar de forma puntual i responsable, i que aquí hem utilitzat únicament amb fins didàctics.

Amb aquesta extensió sobre `Object` ara el nostre codi JavaScript queda molt més llegible:

```javascript
// Iterate over the array and call fly() method if it exists.
randomObjects.forEach((object) => {
  if (object.can("fly")) { 
    object.fly();
  }
})
```

### Interfícies en C#

Com vam fer a les unitats anteriors, vegem breument un exemple d'ús d'interfícies en un llenguatge com C# que sí que suporta aquest mecanisme de manera nativa:

```javascript
// Interface definition.
public interface IFlying {
  void Fly();
}

// Class Airplane implementing IFlying.
public class Airplane : IFlying {
  public void Fly() {
    Console.WriteLine("I'm an airplane and I'm flying!");
  }
}

// Class Bird implementing IFlying.
public class Bird : IFlying {
  public void Fly() {
    Console.WriteLine("I'm a bird and I'm flying!");
  }
}

// Class Superhero implementing IFlying.
public class Superhero : IFlying {
  public void Fly() {
    Console.WriteLine("I'm a superhero and I'm flying!");
  }
}

// Class Fish NOT implementing IFlying.
public class Fish {
  public void Swim() {
    Console.WriteLine("I'm a fish and I'm swimming!");
  }
}

// Example usage. Array of IFlying objects. Some of them can fly, some of them cannot.
var randomObjects = new IFlying[] { new Airplane(), new Bird(), new Superhero() };

// Example usage. Iterate over the array and call Fly() method.
foreach (IFlying object in randomObjects) {
  object.Fly();
}
```

Tot i que la sintaxi de C# difereix de la de JavaScript, podem identificar el concepte d'interfície tal com ho hem explicat en aquesta unitat. Primer es defineix `IFlying` com una interfície que conté un mètode `Fly()`. Després, les classes `Airplane`, `Bird` i `Superhero` implementen la interfície `IFlying` i proporcionen la seva pròpia implementació del mètode `Fly()`. Observem que s'utilitza el símbol `:` per indicar que una classe implementa una interfície, exactament igual que quan una classe hereta d'una altra. Finalment, es crea un array d'objectes indicant que tots implementen la interfície `IFlying`, i es crida el mètode `Fly()` de cada un d'ells.

> <br>🚨&nbsp; **Atenció**
>
> És important que entengueu el concepte d'interfície en el context de la programació orientada a objectes. Però tingueu en compte que cada llenguatge el pot implementar amb matisos diferents. Per exemple, en el [llenguatge Swift es parla de protocols](https://docs.swift.org/swift-book/documentation/the-swift-programming-language/protocols) i es defineixen com segueix:
>
> _Un protocol defineix un esquema de mètodes, propietats i altres requisits que s'adapten a una tasca particular o a un fragment de funcionalitat. El protocol pot ser adoptat per una classe, estructura o enumeració per proporcionar una implementació real d'aquests requisits. Qualsevol tipus que satisfaci els requisits d'un protocol es diu que està conforme amb aquest protocol._
>
> Com podeu observar, aquest concepte guarda moltes similituds amb el de la interfície tal com l'hem descrit fins ara. Tot i que hem esmentat que una interfície no arriba a proporcionar una implementació concreta, els protocols de Swift, en realitat, sí que poden proporcionar una implementació per defecte per a alguns dels seus requisits. Però la seva naturalesa essencialment és la mateixa que la d'una interfície.
