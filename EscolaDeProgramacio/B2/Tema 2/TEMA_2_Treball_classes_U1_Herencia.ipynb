En el nivell B1 vàrem tenir l'ocasió d'introduir el concepte de classes en JavaScript. ## Unitat 1: Herència

L'herència en programació, especialment en el context de classes en JavaScript, és un concepte fonamental en la programació orientada a objectes. Permet que una classe (coneguda com a subclasse) hereti propietats i mètodes d'una altra classe (la superclasse). Això facilita la reutilització de codi i la creació de jerarquies d'objectes més eficients.

Imaginem, per exemple, una classe general anomenada `Vehicle` amb propietats com `brand` i `model`, i mètodes com `start()` o `stop()`. A partir d'aquesta classe, podríem derivar subclasses més específiques com `Car` i `Motorcycle`, que heretarien aquestes quatre característiques comunes, però també podrien tenir les seves pròpies propietats i mètodes únics, com `openTrunk()` per obrir el maleter a `Car`.

L'herència ens permet crear una estructura de codi més neta i organitzada, evitant la duplicació i facilitant la mantenibilitat. També ens aporta un mecanisme que ens ajuda a construir sistemes complexos d'una manera més intuïtiva i eficient, seguint el principi de la **relació "és una"** entre objectes.

Vegem en codi com s'implementa l'herència en JavaScript. Per a això, crearem les classes esmentades anteriorment:

```javascript
// Base class. All other classes inherit from this one.
class Vehicle {
  constructor(brand, model) {
    this.brand = brand;
    this.model = model;
  }

  start() {
    console.log(`Starting ${this.brand} ${this.model}...`);
  }

  stop() {
    console.log(`Stopping ${this.brand} ${this.model}...`);
  }
}

// Subclass Car inherits from Vehicle.
class Car extends Vehicle {
  openTrunk() {
    console.log(`Opening the trunk in ${this.brand} ${this.model}...`);
  }
}

// Subclass Motorcycle inherits from Vehicle.
class Motorcycle extends Vehicle {
  revEngine() {
    console.log(`Revving the engine in ${this.brand} ${this.model}...`);
  }
}

// Example usage.
const myCar = new Car('Toyota', 'Corolla');
// Inherited method from Vehicle.
myCar.start();
// Specific method from Car.
myCar.openTrunk();

// Example usage.
const myMotorcycle = new Motorcycle('Harley-Davidson', 'Street')
// Inherited method from Vehicle.
myMotorcycle.start();
// Specific method from Motorcycle.
myMotorcycle.revEngine();
```

Podem observar diversos punts importants en el codi anterior:

- La classe `Vehicle` és la superclasse. Les classes `Car` i `Motorcycle` són subclasses de `Vehicle`. Indiquem que una classe és subclasse d'una altra mitjançant la paraula reservada `extends`.
- Les subclasses hereten les propietats i els mètodes de la superclasse. En aquest cas, les subclasses hereten les propietats `brand` i `model`, i els mètodes `start()` i `stop()`. Podem observar que les subclasses no tenen definits aquests mètodes, però els hereten de la superclasse i, per tant, els poden emprar. És el que està passant en el codi d'exemple quan s'invoca el mètode `start()` sobre els objectes `myCar` o `myMotorcycle`, que són instàncies de les classes `Car` i `Motorcycle`, respectivament.
- Les classes `Car` i `Motorcycle` poden definir els seus propis mètodes, com `openTrunk()` en el cas de `Car` o `revEngine()` en el cas de `Motorcycle`. Aquests mètodes no són heretats, sinó que són propis de cada subclasse.
- En el codi dels mètodes que són propis de cada subclasse es pot accedir a les propietats heretades de la superclasse mitjançant la paraula reservada `this`.

A més del que hem explicat anteriorment, una subclasse també pot sobreescriure un mètode heretat de la superclasse. Això és útil quan volem que una subclasse tingui un comportament diferent al de la superclasse. Per exemple, podem sobreescriure el mètode `start()` de la classe `Vehicle` a la subclasse `Airplane` perquè, en comptes d'imprimir per consola el missatge "Starting...", imprimeixi el missatge "Started. Let's fly!".

```javascript
// Subclass Airplane inherits from Vehicle.
class Airplane extends Vehicle {
  start() {
    console.log(`${this.brand} ${this.model} started. Let's fly!`);
  }
}

// Example usage.
const myAirplane = new Airplane('Boeing', '747');
// Overridden method from Airplane.
myAirplane.start();         
```

Fins i tot podem sobreescriure un mètode heretat de la superclasse perquè, a més d'executar el codi propi de la subclasse, executi també el codi del mètode heretat des de la superclasse. Per a això, podem emprar la paraula reservada `super` per referir-nos al mètode heretat. Per exemple, podem sobreescriure el mètode `stop()` de la classe `Vehicle` a la subclasse `Rocket` perquè, a més d'imprimir per consola el missatge "Stopping...", imprimeixi el missatge "Stopped. No space oddity anymore!".

```javascript
// Subclass Rocket inherits from Vehicle.
class Rocket extends Vehicle {
  stop() {
    super.stop();
    console.log('Stopped. No space oddity anymore!');
  }
}

// Example usage.
const myRocket = new Rocket('SpaceX', 'Falcon Heavy');
// Overridden method from Rocket.
myRocket.stop();   
```

La paraula clau `super` té un rol crucial en l'herència de classes en JavaScript. Quan sobreescribim un mètode en una subclasse, podem usar `super` per anomenar el mètode corresponent de la superclasse. Això ens permet extendre la funcionalitat del mètode heretat en comptes de reemplaçar-la completament. Utilitzar `super` ens assegura que la nostra subclasse manté una connexió amb la lògica de la superclasse, que permet construir sobre la funcionalitat existent en comptes de partir de zero.

Una subclasse també pot afegir noves propietats a les heretades de la superclasse. Per exemple, podem afegir la propietat `weight` a la subclasse `Bicycle` perquè, a més de les propietats heretades `brand` i `model`, tingui també aquesta propietat `weight` que pot resultar rellevant en el cas de bicicletes de competició.

```javascript
// Subclass Bicycle inherits from Vehicle. It adds weight property.
class Bicycle extends Vehicle {
  constructor(brand, model, weight) {
    super(brand, model);
    this.weight = weight;
  }
}

// Example usage.
const myBicycle = new Bicycle('Specialized', 'Tarmac', 6.8);
// Inherited method from Vehicle.
myBicycle.start();
// Property from Bicycle.
console.log(`My ${myBicycle.brand} ${myBicycle.model} weighs ${myBicycle.weight} kg.`);
```

En el codi anterior observem que quan volem agregar noves propietats a una subclasse hem d'emprar el mètode `constructor()` de la subclasse per inicialitzar-les. En el cas de la subclasse `Bicycle`, el mètode `constructor()` rep com a paràmetres les propietats heretades `brand` i `model`, i la propietat nova `weight`. En el cos del mètode `constructor()` s'invoca, en primer lloc, el mètode `constructor()` de la superclasse mitjançant la paraula reservada `super` per inicialitzar les propietats heretades `brand` i `model`. A continuació, s'inicialitza la nova propietat `weight` amb el valor rebut com a paràmetre.

Quan utilitzem l'herència de classes és possible tenir múltiples nivells, de tal manera que una subclasse hereti d'una altra subclasse, creant una cadena d'herència. Per exemple, fins ara tenim la cadena `Vehicle` > `Rocket`, però podem tenir una altra subclasse que hereti directament de `Rocket`, per exemple `ToyRocket`, que heretarà les propietats i els mètodes tant de `Rocket` com de `Vehicle`, permetent una especialització i extensió més detallada de les característiques heretades i l'addició de noves.

```javascript
// Subclass ToyRocket inherits from Rocket.
class ToyRocket extends Rocket {
  stop() {
    super.stop();
    console.log('Please remove the batteries now.');
  }

  play() {
    console.log('Playing with the toy rocket...');
  }
}

// Example usage.
const myToyRocket = new ToyRocket('Lego', 'Space Shuttle');
// Inherited method from Rocket.
myToyRocket.start();
// Overridden method from ToyRocket.
myToyRocket.stop();
// Specific method from ToyRocket.
myToyRocket.play();
```

L'herència és un pilar de la programació orientada a objectes que ens permet construir estructures de codi eficients i reutilitzables. Trobem aquest mecanisme en molts altres llenguatges de programació. En fer-ne ús, podem crear jerarquies de classes a les nostres aplicacions i reutilitzar i estendre funcionalitats. Amb un enteniment sòlid de l'herència, estem ben equipats per modelar sistemes complexos d'objectes de la vida real als nostres programes.

> <br>🧐&nbsp; **Una persona hereta del seu pare i de la seva mare. Una classe pot heretar de dues superclasses?**
>
> En JavaScript una classe no pot heretar directament de dues superclasses a causa de la seva naturalesa d'herència única. Això significa que cada classe pot heretar només d'una classe pare. Tanmateix, hi ha tècniques com la composició o els mixins que podeu utilitzar per incorporar funcionalitats de múltiples fonts. Tot i que això és comú a la majoria de llenguatges de programació, n'hi ha d'altres, com C++, en els quals sí que es permet l'herència múltiple.
>
> No els estudiarem en aquest nivell B2, però heu de saber que els mixins en JavaScript consisteixen a crear funcions que afegeixen certes capacitats o mètodes a una classe existent. D'aquesta manera, els mixins permeten reutilitzar un conjunt de funcionalitats entre diferents classes sense necessitat de crear una jerarquia de classes basada en l'herència.
