## Unitat 2: Control de versions amb Git

Git és un sistema de control de versions distribuït, dissenyat per manejar eficientment projectes de programació i altres tipus de contingut basats en fitxers de text pla. La funció principal és facilitar la feina col·laborativa entre desenvolupadors, permetent un seguiment detallat dels canvis duts a terme en el codi font. Això ens permet integrar modificacions de diferents branques de feina de manera fluïda. També té un paper crucial en la prevenció de la pèrdua de codi i en la conservació d'un historial complet de les modificacions efectuades.

Git s'ha establert com una eina indispensable en l'àmbit del desenvolupament de programari modern, no només per l'eficàcia en la gestió de versions, sinó també per la flexibilitat i adaptabilitat a diversos fluxos de treball. Bé en projectes individuals o en col·laboracions de gran escala, Git ofereix un marc robust per al maneig de codi. Avui dia és el sistema de control de versions més utilitzat en el món.

### Conceptes bàsics de Git

Git és una eina fonamental per als qui desenvolupen programari, tot i que pot resultar una mica desconcertant al principi si no s'està familiaritzat amb el concepte de control de versions. En essència, Git és com un sistema de seguiment per als canvis en els nostres fitxers de codi font. Quan redactem un document o preparem una presentació, sovint volem desar diferents versions a mida que avancem. Git fa alguna cosa similar amb els nostres fitxers de codi.

Cada vegada que duem a terme un conjunt de canvis significatius en el nostre codi, Git ens permet desar aquests canvis mitjançant uns punts de desament que es coneixen com a commits. Es pot entendre un commit com una espècie de fotografia de l'estat del nostre projecte en un moment determinat. Aquesta fotografia inclou, a més del contingut del codi font, informació sobre què es va canviar respecte a la fotografia anterior i qui fa fer aquests canvis.

Git també ens permet treballar amb diferents branques en els nostres projectes. Això resulta molt útil quan volem experimentar o desenvolupar noves característiques sense alterar la versió principal del projecte. Podem treballar en una branca separada i després decidir si volem combinar els canvis amb la versió principal del nostre codi o si, al contrari, preferim descartar aquest nou codi i no integrar-lo en el projecte.

Si treballem en un equip amb diverses persones, Git facilita que cada una d'elles desenvolupi la seva part per separat i després es puguin unir de manera coherent. També ens permet resoldre els conflictes que puguin sorgir quan dues persones han modificat el mateix fitxer de codi. El disseny de Git permet als equips col·laborar de manera més coordinada, revisant i fusionant canvis, resolent conflictes i mantenint un historial clar i ordenat del desenvolupament del projecte.

### Comandaments imprescindibles de Git

A continuació, mostrem els comandaments més bàsics per a un flux de feina eficient amb Git. Per a una descripció més detallada de tots els comandaments disponibles, podeu consultar el llibre esmentat anteriorment o la [documentació oficial](https://git-scm.com/docs).

1. Inicialitzar un repositori Git en el directori actual.

```javascript
git init
```

2. Clonar un repositori Git remot a la nostra màquina.

```javascript
git clone https://github.com/username/repository-name.git
```

3. Afegir fitxers a l'índex de Git perquè s'incloguin en el commit següent (instantània del projecte).

```javascript
git add file1.txt
git add subdirectory/file2.txt
git add .
```

4. Crear un commit amb els fitxers afegits a l'índex, indicant un missatge descriptiu.

```javascript
git commit -m "Fix issue #1: Change hello to goodbye"
```

5. Mostrar l'estat actual del repositori, incloent els fitxers modificats, afegits o eliminats, així com els fitxers que encara no s'han afegit a l'índex.

```javascript
git status
```

6. Descarregar els canvis duts a terme en el repositori remot i fusionar-los amb el nostre repositori local.

```javascript
git pull origin main
```

7. Pujar els canvis duts a terme en el nostre repositori local al repositori remot.

```javascript
git push origin main
```

8. Mostrar l'historial de commits duts a terme al repositori.

```javascript
git log
git log --oneline
```

9. Mostrar les branques de treball existents en el repositori, així com la branca actual.

```javascript
git branch
```

10. Canviar de branca de treball, creant una nova branca si cal.

```javascript
git switch -c new-branch-name
git switch existing-branch-name
git switch main
```

11. Fusionar una branca de treball amb la branca actual.

```javascript
git merge branch-name-to-merge-with-current-branch
```

12. Consultar la documentació general de Git i la sintaxi d'un comandament en particular.

```javascript
git -h
git switch -h
```

> <br>🧐&nbsp; **El nom de la branca principal és _main_ o és _master_?**
>
> A la comunitat de desenvolupament de programari hi ha hagut un canvi significatiu en la nomenclatura de la branca principal de Git. Històricament, aquesta s'anomenava _master_, però actualment el seu nom és _main_. Aquest canvi es va originar com un esforç per eliminar les referències que es podrien considerar insensibles o carregar amb connotacions històriques negatives relacionades amb l'esclavitud. GitHub, un dels principals serveis d'allotjament de repositoris Git, va adoptar oficialment _main_ com el terme predeterminat per a la branca principal a partir de 2020. Aquesta decisió va ser part d'un moviment més ampli a la indústria tecnològica per promoure un llenguatge més inclusiu i conscient en les eines i plataformes de desenvolupament.

### Treball amb GitHub

Si utilitzem Git en els nostres projectes, és molt recomanable utilitzar un servei d'allotjament de repositoris Git com [GitHub](https://github.com). Aquest tipus de serveis ens permeten emmagatzemar els nostres repositoris de codi al núvol, que facilita la feina col·laborativa i la gestió de projectes. GitHub és el servei d'allotjament de repositoris Git més popular, però hi ha alternatives com [GitLab](https://gitlab.com) o [Bitbucket](https://bitbucket.org) que també ofereixen funcionalitats similars.

Git i GitHub són dues eines diferents, tot i que estretament relacionades. Git és el sistema de control de versions, mentre que GitHub és un servei d'allotjament de repositoris Git. Git treballa a la nostra màquina local, mentre que GitHub treballa al núvol. Git ens permet fer commits, crear branques de treball i fusionar-les, mentre que GitHub ens permet emmagatzemar els nostres repositoris al núvol, compartir-los amb altres persones i col·laborar en projectes de manera eficient.

El flux de feina a GitHub generalment inclou els passos següents:

1. **Creació d'un repositori** – S'inicia un projecte creant un repositori a GitHub. Aquest repositori serà el centre d'emmagatzematge en línia per al nostre projecte i pot ser públic o privat.

2. **Clonació del repositori** – Tots els col·laboradors poden clonar el repositori a les seves màquines locals, fet que els permet treballar en el projecte de forma independent. Tot i que GitHub destaca per la seva funcionalitat de feina col·laborativa, també es pot utilitzar per a projectes individuals.

3. **Realització de canvis i commits locals** – Després de fer canvis en el codi de manera local, els desenvolupadors duen a terme commits per registrar les seves modificacions.

4. **Push de canvis al repositori remot** – Una vegada dut a terme el commit local, els canvis es pugen (push) al repositori a GitHub, fent-los accessibles a la resta de col·laboradors.

5. **Revisió de canvis i pull requests** – Els col·laboradors d'un projecte poden revisar els canvis proposats i discutir-los. Les _pull requests_ són una forma de sol·licitar la integració d'aquests canvis a la branca principal del projecte.

6. **Fusió de canvis i resolució de conflictes** – Finalment, els canvis poden ser fusionats en el repositori principal. Si hi ha conflictes, s'han de resoldre abans de completar la fusió.

7. **Gestió d'issues i col·laboració** – GitHub també permet als equips rastrejar problemes (issues), planificar tasques i col·laborar a través d'eines integrades amb els taulers de projectes i les wikis.

Mitjançant la feina amb GitHub millorem la col·laboració i la gestió dels nostres projectes. A més, posa a la nostra disposició una plataforma que ens permet compartir codi de manera oberta, contribuir a projectes d'altres i construir un portafolis visible de feina per a futurs ocupadors o col·laboradors.
