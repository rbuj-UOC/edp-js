# TEMA 5: Tècniques per millorar l'estructura i robustesa del codi

## Unitat 1: Gestió d'errors i fiabilitat del codi

Una gestió efectiva d'errors i excepcions resulta fonamental per crear un codi robust i fiable. Quan programem qualsevol funcionalitat d'una aplicació, ens solem centrar en l'algoritme que resol el problema, però sovint cometem l'error de no considerar en profunditat tots els casos límit o casos extrem (an anglès, _edge cases_) en els quals el nostre codi també ha de donar un resultat correcte, així com els diferents tipus d'error que es poden produir.

Hi ha moltes situacions en les quals el comportament d'un programa pot ser inusual o imprevist, sovint en condicions poc comunes. És important prestar atenció a aquests casos per garantir la robustesa i fiabilitat del programari.

Enfrontar i manejar adequadament les situacions inesperades o errònies en el nostre codi ens permet preveure errors en les aplicacions i assegurar una experiència d'usuari més fluida i segura. A més, una bona pràctica en la gestió d'errors ens permet depurar i mantenir el codi de manera més eficaç. Quan aconseguim anticipar i planificar respostes davant de possibles errors, incrementem la qualitat i la confiabilitat de les nostres aplicacions.

Per a això disposem de diverses eines fonamentals que estudiarem en aquesta unitat: 

- Gestió d'excepcions mitjançant el mecanisme try-catch-finally.
- Ús d'assercions.
- Registre i monitoratge d'errors.

> <br>🧐&nbsp; **Errors i excepcions són el mateix?**
> 
> Tot i que sovint s'usen de manera intercambiable, els termes "error" i "excepció" tenen matisos diferents. Un error es refereix a un problema que impedeix que el programa funcioni correctament. Pot ser un error de sintaxi, un error en el temps de compilació, o un error en el temps d'execució que no s'ha previst ni manejat adequadament. Per contra, una excepció és un tipus d'error que pot ser previst i manejat dins del flux d'execució del programa. Les excepcions s'utilitzen per manejar situacions anòmales o inesperades que poden ser recuperables, com un fitxer que no es troba o un format de dades incorrecte. Així doncs, no tots els errors són excepcions, però les excepcions sí que són un subconjunt d'errors. En qualsevol cas, sovint trobareu que s'usen ambdós termes de manera indistinta. També en aquests materials utilitzarem ambdós termes de manera intercambiable.
> <br><br>

### Maneig d'excepcions mitjançant el mecanisme try-catch-finally

El mecanisme `try-catch-finally` ens proporciona una sintaxi fonamental en programació per controlar el flux d'execució d'un programa quan ocorren errors o comportaments inesperats que, si no es controlen, poden provocar que el programa s'aturi o es comporti de manera inadequada.

- `try` – Encercla el codi que pot generar una excepció. Si dins d'aquest bloc es produeix un error en una línia, l'execució no continua a la línia següent com esperaríem, sinó que salta immediatament al bloc `catch`.
- `catch` – Captura l'error que s'hagi pogut generar dins del bloc `try`. Aquí es defineix com respondre davant de l'error, per exemple mostrant un missatge a l'usuari o duent a terme alguna acció correctiva. 
- `finally` – Aquest bloc s'executa sempre, independentment de si s'ha produït o no una excepció en el bloc `try`. Resulta molt útil per dur a terme operacions de neteja o finalització de processos, com tancar una connexió a una base de dades o resetejar el valor d'algunes variables.

Vegem-ho mitjançant un exemple en el qual tenim dos objectes corresponents a dos usuaris i volem accedir a la seva propietat `name`. Tot i que aquí definim els objectes de manera literal, en un cas real podríem rebre les dades d'una font externa, com un fitxer o una base de dades, i no tindrem garantia que les dades siguin correctes. És el que en aquest exemple simulem amb l'objecte `user2`.

```javascript
const user1 = {name: 'Richard', age: 42};
const user2 = null;

// user1 is not null, so the name property can be accessed
console.log(user1.name);
// user2 is null, so trying to access the name property causes an error
// TypeError: Cannot read property 'name' of null
console.log(user2.name); 
```

Observem que, en intentar accedir a la propietat `name` de l'objecte `user2`, es produeix un error que deté l'execució del programa. Això és precisament el que volem evitar: en una aplicació real no podem permetre que un error faci que el nostre programa deixi de funcionar. Per a això, utilitzarem el mecanisme `try-catch-finally`.

```javascript
try {
  // user1 is not null, so the name property can be accessed
  console.log(user1.name);
  // user2 is null, so trying to access the name property causes an error
  console.log(user2.name); 
} catch (error) {
  // This time the error is caught and the code in the catch block executes
  console.error("Oooops!!! " + error.message);
} finally {
  console.log("This is always executed");
}
```

Aquest exemple mostra de manera molt senzilla el mecanisme `try-catch-finally`. Però presenta un problema important: estem utilitzant un bloc `try` per encerclar un codi que, en general, no hauria de produir errors, ja que accedir a una propietat d'un objecte no és una operació crítica. Hem de tenir en compte que el mecanisme `try-catch-finally` té un cert cost computacional, per la qual cosa no l'hauríem d'utilitzar per encerclar codi que no hauria de produir errors. En aquest cas, el que hauríem de fer es assegurar-nos que l'objecte `user2` té la propietat `name` abans d'intentar accedir-hi.

Resulta, doncs, una mala pràctica fer un ús indiscriminat i preventiu del mecanisme `try-catch-finally`. Aquest mecanisme mai ha de servir per fer més segur un codi que no s'ha dissenyat correctament. Al contrari, l'hem d'utilitzar només en aquells escenaris en els quals hi hagi factors que no podem controlar i que poden produir errors o comportaments inesperats. Per exemple, una mala connexió a una base de dades, una xarxa inestable o un fitxer corrupte.

> <br>🚨&nbsp; **Atenció**
>
> Heu de tenir molt clara la diferència entre un codi que no funciona com s'espera i un error o una excepció. Un error és una situació excepcional que es produeix durant l'execució d'un programa i que impedeix que aquest continuï el seu normal funcionament. Per contra, si el vostre codi inclou un algoritme incorrecte (per exemple, un bucle que no es deté mai o un càlcul matemàtic erroni), el programa no donarà el resultat esperat, però no es produirà cap error ni excepció.
> 
> Per exemple, quan porteu a terme una petició `GET` a un servidor web, si el recurs sol·licitat no existeix, el servidor us retornarà un resultat `404` (pàgina no trobada). Però si el codi de la crida `fetch()` i el posterior processament és correcte, no s'haurà produït cap error ni excepció. El codi haurà funcionat amb normalitat, però el resultat no serà l'esperat.
> 
> Per tant, per cobrir els dos motius pels quals el vostre codi pot no funcionar com espereu, necessiteu dues eines diferents: gestió d'excepcions i proves unitàries. 
>
> De nou, no confongueu un resultat incorrecte per un error a l'algoritme seguit (el que es coneix com a _bug_) amb un error (excepció). Un _bug_ és un error en el codi que provoca que el programa no funcioni com s'espera. Per exemple, un _bug_ en un algoritme d'ordenació pot provocar que el resultat no estigui ordenat correctament. Però no s'haurà produït cap error (excepció). El codi haurà funcionat amb normalitat, però el resultat no serà l'esperat. En canvi, si una operació `fetch()` no es pot dur a terme perquè no hi ha connexió a internet, sí que es produirà un error (excepció). En aquest cas, el codi no haurà funcionat amb normalitat i el programa no podrà continuar l'execució. Tot i que l'ús de la paraula _error_ és ambigu, convé diferenciar entre els seus dos significats: _bug_ i excepció.
> <br><br>

### Ús d'assercions

L'ús d'assercions també ens pot ajudar en la gestió d'errors i excepcions. Les assercions són una eina que ens permet verificar que una condició específica es compleix en un punt del programa. Si la condició avaluada en l'asserció és certa, el programa continua l'execució de manera normal i silenciosa. Si, en canvi, l'asserció és falsa, el programa llança un error i típicament es deté. Això pot semblar un inconvenient, però en realitat ens facilita identificar problemes i comportaments inesperats durant la fase de desenvolupament.

> <br>🚨&nbsp; **Atenció**
>
> Hi ha una filosofia de desenvolupament de programari que emfatitza la detecció primerenca i explícita d'errors. Aquesta idea es basa en el principi que com més aviat es detecti un problema en el codi, més fàcil i ràpid serà solucionar-lo. En comptes de permetre que un error es propagui a través del sistema i causi problemes més complexos i difícils de rastrejar, es prefereix que el programa falli immediatament quan alguna cosa va malament. Aquest enfocament permet una detecció primerenca d'errors, en facilita la depuració i prevé errors més importants. En aquest sentit, les assercions són una eina molt útil per detectar errors en el codi de manera primerenca i explícita.
> <br><br>

En JavaScript, podem utilitzar assercions mitjançant l'ús del mètode estàtic `assert()` de l'objecte `console`, o optar per implementar una funció d'asserció simple pròpia:

```javascript
function assert(condition, message) {
  if (!condition) {
    throw new Error(message || "Assertion failed");
  }
}

assert(value > 10, "value must be greater than 10");
```

El funcionament és molt senzill: `assert()` verifica que una condició sigui certa i llança un error amb un missatge específic si no ho és. Com hem comentat, les assercions són útils durant el desenvolupament per validar supòsits i detectar errors ràpidament. No obstant això, s'han d'usar amb molta precaució en un entorn de producció, ja que la detenció del programa podria tenir efectes no desitjats.

### Registre i monitoratge d'errors

El registre i monitoratge d'errors són pràctiques clau en el desenvolupament de JavaScript per identificar i abordar problemes a les aplicacions. El registre (en anglès, _logging_) implica capturar informació sobre esdeveniments anòmals o errors en el codi, mentre que el monitoratge es refereix a l'observació contínua de l'estat i comportament d'una aplicació.

En JavaScript disposem d'eines com `console.log()`, `console.error()` i `console.warn()` que utilitzem de manera habitual per registrar missatges a la consola del navegador o de l'entorn d'execució. Tanmateix, per a un monitoratge més avançat, es poden implementar solucions professionals com [Sentry](https://sentry.io) o [LogRocket](https://logrocket.com), que ofereixen anàlisis en temps real, seguiment d'excepcions i mètriques de rendiment.

Aquestes eines permeten no només registrar errors, sinó també recopilar el context detallat amb informació sobre l'estat de l'aplicació en el moment de l'error, facilitant així la depuració i la millora contínua del codi. El monitoratge proactiu ajuda a detectar patrons i tendències en els errors, fet que és essencial per al manteniment i l'optimització a llarg termini de les aplicacions web.

Mentrestant, els mètodes esmentats de l'objecte `console` us seran de gran utilitat per registrar i monitorar errors en els vostres projectes acadèmics. I si necessieu analitzar el rendiment de les vostres aplicacions, disposeu també de `console.time()`, `console.timeLog()` i `console.timeEnd()`, que permten mesurar el temps d'execució d'un bloc de codi.

### Altres tècniques per millorar la robustesa del codi

En un entorn de producció és important que les aplicacions web siguin robustes i fiables. Per a això, necessitem desenvolupar habilitats que ens permetin identificar i resoldre errors de manera eficient. Hi ha eines que ens poden ajudar a escriure un codi més confiable.

#### Depuració de codi

La depuració de codi (en anglès, _debugging_) és una competència clau en el desenvolupament de programari. En un primer moment, podem utilitzar els diferents mètodes de l'objecte `console` que acabem d'esmentar. Però la depuració avançada suposa anar més enllà d'aquests logs i implica entendre profundament com funciona el codi i com interactuen els diferents components.

En els navegadors moderns disposem d'eines integrades, com els panels de depuració DevTools de Chrome o Firefox, que ofereixen funcions com punts d'interrupció, inspecció de variables i visualització de la pila de crides. Amb aquestes eines podem marcar punts del nostre codi on volem que l'execució es detingui per inspeccionar l'estat de les variables i el flux d'execució. Fins i tot podem modificar el valor de les variables en el temps d'execució per comprovar com afecta el comportament del programa. Tot i que una mica més complex, aquestes eines també ens permeten depurar codi asíncron. Disposem, a més, de llibreries externes que poden facilitar la detecció i resolució de problemes complexos en aplicacions JavaScript més grans i en entorns de producció.

> <br>🧐&nbsp; **Quan cal _depurar el nostre codi_?**
>
> A vegades us trobareu davant d'un problema en el qual el vostre codi no es comporta com esperàveu i potser ho fa de manera intermitent. Per exemple, si la vostra aplicació web presenta un error només els dimecres o si unes dades no es processen correctament quan l'usuari ha seleccionat una determinada combinació d'opcions. En aquests escenaris rars, aturar-se i examinar detalladament el flux del programa i l'estat de les variables en punts crítics mitjançant la depuració us pot ajudar a identificar i resoldre el problema.
> <br><br>

#### Proves unitàries

En la mateixa línia d'intentar que el nostre codi sigui més robust, les proves unitàries emergeixen com una tècnica fonamental. Aquest tipus de proves s'enfoquen en verificar la funcionalitat de les unitats més petites del nostre codi, com funcions o mètodes, de forma aïllada. Aquesta pràctica no només ens ajuda a detectar i corregir errors de manera primerenca en el cicle de desenvolupament, sinó que també ens garanteix que cada component funciona segons l'esperat abans d'integrar-se en el sistema més gran.

Les proves unitàries, per tant, són crucials per mantenir un codi net, eficient i confiable. A cada llenguatge de programació hi ha eines específiques per dur a terme aquestes proves unitàries. En el cas de JavaScript, un dels marcs de proves més populars i eficients és [Jest](https://jestjs.io). Aquest framework facilita la creació de proves concises i mantenibles, que ens permeten verificar de manera individual cada part de la nostra aplicació.

A més, Jest s'integra amb projectes JavaScript per automatitzar tot el procés de creació de proves, incloent tècniques avançades com el _mockeo_ de mòduls i API. El _mockeo_, en el context de la programació i les proves de programari, es refereix a la simulació de comportaments o respostes de certs components o sistemes per poder provar altres parts de l'aplicació. Essencialment, consisteix a "fingir" certes funcionalitats, com crides a bases de dades o a serveis externs, per aïllar i provar específicament el codi que ens interessa a cada moment, sense dependre de components externs o que són difícils de controlar durant les proves.

> <br>🧐&nbsp; **Què _comproven_ les proves unitàries?**
>
> Per preparar una prova unitària primer necessiteu identificar la funcionalitat específica que voleu provar. Pot ser una funció o un mètode en el vostre codi. Per exemple, si teniu una funció que calcula el total d'un carretó, una prova unitària es pot enfocar en aquesta funció. Dins d'aquesta prova haureu d'escriure tots els casos de prova que incloguin diferents escenaris i dades d'entrada: un carretó buit, un carretó amb diversos articles, un carretó amb articles repetits, un carretó amb descomptes aplicats, etc. A continuació, haureu de comprovar que la sortida de la funció (l'output) en cada cas coincideix amb l'esperat (els inputs). D'aquesta manera, obtindreu una raonable seguretat que la funció es comporta correctament en condicions diverses.
>
> Penseu també en els _edge cases_ o casos límit. Què passa si l'objecte o array que recull els articles del carretó és `null` o `undefined`? Què passa si un article té el seu preu establert a `null`, `undefined`, un nombre negatiu o algun tipus de dada que no és un nombre? Les proves unitàries han de cobrir tots aquests casos límit per garantir que el codi funciona correctament en qualsevol escenari.
> <br><br>

Tant la depuració avançada com les proves unitàries són tècniques que requereixen un aprenentatge i una pràctica continuada. En aquest nivell B2 no entrem en detall en aquestes tècniques, però qualsevol programador professional les ha de conèixer i dominar.
