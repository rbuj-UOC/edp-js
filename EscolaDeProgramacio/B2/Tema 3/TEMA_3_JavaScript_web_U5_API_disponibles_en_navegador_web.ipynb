## Unitat 5: API disponibles en el navegador web

Els navegadors web moderns proporcionen una varietat d'API que permeten als desenvolupadors web interactuar de manera més rica i dinàmica amb l'entorn del navegador i el sistema de l'usuari. Aquestes API no formen part del llenguatge JavaScript, sinó que són exclusives de la programació d'aplicacions web, en tant que, com acabem de mencionar, és el navegador web qui hi dona accés.

En aquesta unitat explorarem tres de les API més comunes i útils que podem trobar en els navegadors web moderns. En concret, veurem com podem accedir a les API de _Geolocation_, _Broadcast Channel_ i _Speech_.

Actualment, l'API Fetch s'ha convertit en l'estàndard de facto per dur a terme peticions de manera asíncrona. Tot i que en aquest moment, amb l'ànim de simplificar, la vàrem presentar simplement com una API pròpia de JavaScript, la realitat és que Fetch és una API proporcionada pel navegador web, com les que veurem en aquesta unitat. No obstant això, altres entorns de programació, com Node.js, també han incorporat ja la seva pròpia API Fetch. En qualsevol cas, tot i que les diferències pràctiques siguin insignificants, convé tenir present que Fetch no és una API pròpia del llenguatge JavaScript, sinó de l'entorn en el qual s'executa el codi JavaScript.

### Geolocation API

Aquesta API permet que una aplicació web pugui accedir a la ubicació geogràfica del dispositiu de l'usuari. Resulta útil per a una gran varietat d'aplicacions: serveis de mapes, jocs basats en la ubicació, cadenes de botigues que poden mostrar contingut personalitzat segons on es troba l'usuari, etc. Fins i tot ens pot permetre ajustar el contingut i la publicitat de la nostra pàgina web basant-nos en la ubicació de l'usuari. 

L'API de geolocalització brinda opcions tant per obtenir la ubicació actual com per seguir els canvis d'ubicació en temps real. No obstant això, com a desenvolupadors hem de tenir present que sempre es necessita el permís de l'usuari, per la qual cosa és important que expliquem clarament per què necessitem accedir a la seva ubicació i què farem amb aquesta. I també hem de tenir contemplat què haurem de fer si l'usuari no ens dona permís per accedir a aquesta informació.

El codi següent mostra un exemple de com podem utilitzar aquesta API per obtenir la ubicació de l'usuari i mostrar-la en una pàgina web.

> <br>🎲&nbsp; **Exemple disponible**
>
> Trobareu els fitxers HTML, CSS i JavaScript complets per reproduir aquest exemple a la carpeta **code35/geolocation**.
> <br><br>

```javascript
// Check if the browser has geolocation capabilities.
if ('geolocation' in navigator) {
  // Get the current position of the device.
  navigator.geolocation.getCurrentPosition(success, error);

  // Start watching for changes in the device's location.
  const watcher = navigator.geolocation.watchPosition(success, error);

  // Stop watching for changes in the device's location.
  // Commented out because we don't want to stop watching.
  // But at some point we will want to stop watching.
  // navigator.geolocation.clearWatch(watcher);

  // What to do when we have the position.
  function success(position) {
    const latitude = position.coords.latitude;
    const longitude = position.coords.longitude;
    console.log(`Latitude: ${latitude}, Longitude: ${longitude}`);
  }

  // What to do when we have an error.
  function error(err) {
    console.warn(`ERROR(${err.code}): ${err.message}`);
  }
} else {
  console.log('Geolocation is not supported by this browser.');
}
```

Les quatre parts més rellevants d'aquest codi són les següents:

- Comprovar que el navegador web suporta l'API de geolocalització. Per a això, verifiquem si la propietat `geolocation` dins de l'objecte `navigator` ens retorna un objecte de tipus `Geolocation`. Si no és el cas, significa que l'API no està disponible i mostrem un missatge d'error a la consola del navegador web.
- Quan volem aconseguir una lectura puntual de la ubicació de l'usuari utilitzem el mètode `getCurrentPosition()` de l'objecte `Geolocation`. Aquest mètode rep dues funcions callback com a paràmetres: la primera s'executa si la lectura de la ubicació es du a terme correctament, i la segona si es produeix algun error.
- Quan volem seguir els canvis d'ubicació de l'usuari utilitzem el mètode `watchPosition()` de l'objecte `Geolocation`. Aquest mètode també rep dues funcions callback com a paràmetres: la primera s'executa cada vegada que es produeix un canvi d'ubicació, i la segona si es produeix algun error. És important conservar en una variable l'identificador que el mètode ens retorna, ja que posteriorment el podem utilitzar per detenir el seguiment de la ubicació de l'usuari.
- Per detenir aquest seguiment utilitzem el mètode `clearWatch()` de l'objecte `Geolocation`, passant-li com a paràmetre l'identificador que ens va retornar el mètode `watchPosition()`.

### Broadcast Channel API

Aquesta API ens permet establir una comunicació senzilla entre múltiples contextos de navegació (com diferents pàgines o diferents pestanyes) que comparteixen el mateix origen. Mitjançant aquest canal de comunicació podem portar a terme l'enviament de missatges i dades entre diferents instàncies d'una aplicació obertes en un navegador. S'utilitza per sincronitzar dades o estats entre aquestes instàncies, i permet que els canvis fets en una pestanya es reflecteixin automàticament a les altres. Resulta especialment útil en aplicacions web que requereixen que diverses finestres o pestanyes que estan mostrant la mateixa pàgina es mantinguin actualitzades en temps real.

El mecanisme de comunicació que proporciona aquesta API és molt senzill. En primer lloc, hem de crear un canal de comunicació mitjançant `new BroadcastChannel()`, al qual li hem de passar com a paràmetre el nom del canal que volem crear. Aquest nom ha de ser el mateix en tots els contextos de navegació que vulguem que comparteixin el canal de comunicació. En segon loc, ens hem de subscriure al canal mitjançant el mètode `onmessage`, que ens permetrà rebre els missatges que s'enviïn a través del canal. Finalment, per enviar un missatge a través del canal, utilitzarem el mètode `postMessage()`, al qual li passarem com a paràmetre el missatge que volem enviar.

El codi següent mostra tot això en acció. Per provar-lo, cal obrir dues finestres o dues pestanyes del navegador i carregar en ambdues el fitxer HTML corresponent. A la primera premem el botó per escoltar. A la segona, premem el botó per enviar. En ambdues consoles podrem veure els missatges que s'envien i reben a través del canal de comunicació.

> <br>🎲&nbsp; **Exemple disponible**
>
> Trobareu els fitxers HTML, CSS i JavaScript complets per reproduir aquest exemple a la carpeta **code35/broadcastchannel**.
> <br><br>

```javascript
// Add event listener for the poster button.
document.querySelector('#poster').addEventListener('click', () => {
  // Create a new BroadcastChannel with the name 'test_channel'.
  const bc = new BroadcastChannel('test_channel');
  // Post a message to the channel.
  bc.postMessage('Hello World!');
  console.log('Message posted');
});

// Add event listener for the listener button.
document.querySelector('#listener').addEventListener('click', () => {
  // Create a new BroadcastChannel with the name 'test_channel'.
  const bc = new BroadcastChannel('test_channel');
  // Listen for messages on the channel.
  bc.onmessage = (event) => {
    console.log('Message received');
    console.log(event);
    console.log(event.data);
  }
});
```

Aquesta API pot resultar útil en diferents escenaris en projectes web professionals:

- En una aplicació web d'edició de text, la podem utilitzar perquè els canvis fets en una pestanya d'un document es reflecteixin automàticament a les altres pestanyes obertes amb el mateix document. 
- En una aplicació web de missatgeria, ens pot servir perquè els missatges que enviem en una pestanya es mostrin automàticament a les altres pestanyes obertes amb la mateixa conversa.
- En una aplicació web que periòdicament obté dades del servidor, si un usuari obre l'aplicació en diverses pestanyes, cada una obtindrà la seva pròpia còpia de les dades, desaprofitant l'amplada de banda. Podem fer que una pestanya obtingui les dades i les comparteixi amb les altres pestanyes.
- En qualsevol aplicació web que requereixi obrir una sessió d'usuari, si l'usuari ha tancat sessió en una pestanya, podem transmetre aquest esdeveniment a totes les altres pestanyes perquè actualitzin el seu estat en conseqüència.

En general, aquesta API resulta útil en qualsevol aplicació web que requereixi que diverses finestres o pestanyes que estan mostrant la mateixa pàgina es mantinguin actualitzades entre elles en temps real.

### Speech API

Aquesta API està, en realitat, composta per dues parts: Speech Synthesis i Speech Recognition. De manera conjunta permeten que una aplicació web pugui interactuar amb l'usuari mitjançant la parla. En concret, el component Speech Synthesis s'utilitza per convertir text en parla, permetent que l'aplicació "parli" a l'usuari. En canvi, la part de Speech Recognition permet que sigui l'usuari qui parli i l'aplicació reconegui i interpreti les paraules de l'usuari. 

Resulta evident que aquests dos components, junts o per separat, obren un ampli rang de possibilitats per millorar l'accessibilitat de les nostres aplicacions, crear experiències interactives, desenvolupar assistents virtuals o aplicacions educatives, etc. Ese tracta, en definitiva, d'oferir mètodes alternatius d'entrada de dades i interacció.

> <br>🎲&nbsp; **Exemple disponible**
>
> Trobareu els fitxers HTML, CSS i JavaScript complets per reproduir aquest exemple a la carpeta **code35/speech**.
> <br><br>

```javascript
// Speech Synthesis.
const speechSynthesis = window.speechSynthesis;
// Text to speech.
const speak = (text) => {
  // Check if already speaking.
  if (speechSynthesis.speaking) {
    console.error('speechSynthesis.speaking');
    return;
  }

  const utterance = new SpeechSynthesisUtterance(text);
  speechSynthesis.speak(utterance);
}

// Speech Recognition.
const SpeechRecognition = window.SpeechRecognition || window.webkitSpeechRecognition;
const recognition = new SpeechRecognition();

// What to do when we have the result of the recognition.
recognition.onresult = (event) => {
  const transcript = event.results[0][0].transcript;
  console.log(transcript);
  speak(transcript);
}

// Start recognition (speech to text).
recognition.start();
```

Totes les API del navegador estan subjectes a la disponibilitat i els permisos que atorgui l'usuari. Això és especialment crític en el cas de l'API Speech. La compatibilitat i el funcionament d'aquesta API poden variar de manera significativa entre diferents navegadors i sistemes operatius. A més, la configuració de privacitat i els permisos de l'usuari tenen un paper crucial, ja que l'API requereix accés al micròfon del dispositiu per portar a terme el reconeixement de veu i al sistema d'àudio per a la síntesi de veu. A vegades l'usuari pot haver deshabilitat de manera global l'accés a aquests dispositius i no ser-ne ni tan sols conscient.

Aquests factors poden suposar inconsistències en l'experiència de l'usuari i la necessitat d'implementar solucions de suport o detectar la disponibilitat de l'API abans del seu ús. En cas contrari, la nostra aplicació podria no funcionar correctament i l'usuari podria no entendre per què passa això.
