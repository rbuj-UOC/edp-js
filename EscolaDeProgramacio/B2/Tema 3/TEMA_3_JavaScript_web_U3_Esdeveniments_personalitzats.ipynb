## Unitat 3: Esdeveniments personalitzats

Els esdeveniments personalitzats són una tècnica que ens permet manejar interaccions i comunicacions específiques a les nostres aplicacions web. Amb la creació d'esdeveniments únics en el nostre codi aconseguim tenir un control més detallat de les activitats que porta a terme l'usuari i podem millorar la coordinació entre diferents components de l'aplicació.

Un primer exemple d'esdeveniment personalitzat el podem imaginar en un joc en línia. Imaginem que cada vegada que un jugador arriba a un nou nivell optem per disparar un esdeveniment de "nivell aconseguit". Aquest esdeveniment podria activar diverses accions, com el desbloqueig de noves habilitats o un anunci de la gesta aconseguida a la resta de jugadors.

Amb un esdeveniment personalitzat podem crear una comunicació clara i específica entre diferents parts de l'aplicació, que facilita així la implementació de respostes dinàmiques a accions de l'usuari.

Un segon exemple el trobem en una aplicació de comerç electrònic. Quan l'usuari afegeix un producte al carretó es pot llançar un esdeveniment de "producte afegit". Aquest esdeveniment personalitzat pot actualitzar la interfície d'usuari per mostrar el nombre actualitzat d'articles al carretó i desencadenar recomanacions d'altres productes relacionats.

> <br>🚨&nbsp; **Atenció**
>
> Els esdeveniments personalitzats en JavaScript solen estar associats amb accions específiques dins de l'aplicació. Generalment, és el codi JavaScript qui els desencadena, a diferència del que passa amb els esdeveniments que hem estudiat fins ara, que es llancen a partir d'accions de l'usuari com moviments del ratolí o pulsacions de teclat. Podeu pensar en els esdeveniments personalitzats com una forma de comunicació entre diferents parts de la vostra aplicació web.
> <br><br>

Vegem el codi bàsic d'un exemple d'esdeveniment personalitzat:

```javascript
// Create a custom event.
const myCustomEvent = new CustomEvent('myEvent', {
  detail: { message: 'This is my custom event!' }
});

// Listen and handle the custom event.
document.addEventListener('myEvent', function(e) {
  console.log('Event received:', e.detail.message);
});

// Dispatch the event.
document.dispatchEvent(myCustomEvent);
```

Analitzem aquest codi pas a pas:

- En primer lloc, es crea un esdeveniment personalitzat anomenat `myEvent`. Aquest nom és l'equivalent a, per exemple, `click` a l'esdeveniment més habitual amb el qual solem treballar en una aplicació web. Es tracta del nom que després utilitzarem per enllaçar el controlador d'esdeveniments.
- El nostre esdeveniment pot portar unes dades addicionals a la propietat `detail`, tan àmplies com necessitem. Hem optat per afegir un missatge de text, però podríem incloure qualsevol tipus d'informació, com un objecte amb dades d'usuari o un array amb els productes del carretó.
- A continuació, es defineix un controlador d'esdeveniments amb `addEventListener()` per escoltar l'esdeveniment `myEvent`. Aquesta part funciona exactament igual com l'hem vist fins ara per exemple per escoltar els esdeveniments `click`. Utilizem el nom de l'esdeveniment personalitzat que hem creat i, com a segon paràmetre, una funció callback que s'executarà quan es dispari l'esdeveniment.
- Finalment, es dispara l'esdeveniment de manera específica amb `dispatchEvent()`. Aquest mètode és l'equivalent a fer clic en un botó o a prémer una tecla.
  
En el cas dels esdeveniments natius és l'usuari qui desencadena l'esdeveniment. En canvi, els esdeveniments personalitzats es llancen des del codi JavaScript. No obstant això, és perfectament possible combinar ambdues coses: tot i que en aquest exemple disparem l'esdeveniment en carregar la pàgina, el podríem disparar quan l'usuari faci clic en un botó.

Mostrem a continuació un exemple més elaborat d'esdeveniment personalitzat en una aplicació web que incorpora un carretó clàssic. En aquest cas, l'esdeveniment personalitzat es dispara quan l'usuari afegeix un producte al carretó i s'encarrega d'actualitzar la interfície d'usuari per mostrar el contingut d'aquest carretó. El codi HTML és el següent:

```javascript
<div id="products">
  <h2>Products</h2>
  <div class="product">
    <span>Artisan Symphony</span>
    <button class="add" data-product-id="1" data-product-description="Artisan Symphony">Add to cart</button>
  </div>
  <div class="product">
    <span>Glowing Aurora</span>
    <button class="add" data-product-id="2" data-product-description="Glowing Aurora">Add to cart</button>
  </div>
  <div class="product">
    <span>Stellar Voyager</span>
    <button class="add" data-product-id="3" data-product-description="Stellar Voyager">Add to cart</button>
  </div>
  <div class="product">
    <span>Time Guardian</span>
    <button class="add" data-product-id="4" data-product-description="Time Guardian">Add to cart</button>
  </div>
  <div class="product">
    <span>Zen Elixir</span>
    <button class="add" data-product-id="5" data-product-description="Zen Elixir">Add to cart</button>
  </div>
</div>

<div id="shopping-cart">
  <h2>Shopping cart</h2>
  <ul id="cart-items"></ul>
</div>
```

Els estils aplicats no revesteixen més importància i els podeu trobar al fitxer CSS corresponent. En canvi, el codi JavaScript és més interessant i l'analitzarem pas a pas:

```javascript
// Listen for clicks on the "Add to cart" buttons.
document.querySelectorAll('.product .add').forEach((button) => {
  button.addEventListener('click', function () {
    // Get the product ID and description from data attributes.
    const productId = this.getAttribute('data-product-id');
    const productDescription = this.getAttribute('data-product-description');
    // Create a custom event with the product ID and description.
    const event = new CustomEvent('addToCart', { detail: { productId, productDescription } });
    // Dispatch the event.
    document.dispatchEvent(event);
  })
})

// Listen and handle the custom event.
document.addEventListener('addToCart', function (e) {
  const cartItem = document.createElement('li');
  cartItem.textContent = e.detail.productDescription;
  document.getElementById('cart-items').appendChild(cartItem);
  console.log('Product added to cart:', e.detail.productId, e.detail.productDescription);
});
```

En el primer bloc, afegim un controlador d'esdeveniments per a l'esdeveniment `click` de cada botó. Aquest controlador s'encarrega de recuperar algunes dades, crear un esdeveniment personalitzat i disparar-lo. En aquest cas, l'esdeveniment personalitzat es diu `addToCart` i porta com a dades addicionals el nom del producte i l'identificador.

En el segon bloc, indiquem el desig d'escoltar l'esdeveniment `addToCart` i definim què ha de passar quan aquest esdeveniment es dispari, que en aquest cas no és una altra cosa que actualitzar la interfície d'usuari per mostrar el contingut del carretó i mostrar un missatge a la consola.

Els esdeveniments personalitzats són una eina potent a l'arsenal d'un desenvolupador web, ja que permeten una interacció detallada i controlada a l'aplicació. Des de millorar l'experiència de l'usuari fins a facilitar la comunicació entre components, el seu ús adequat pot aportar una millora en la funcionalitat i eficiència de les aplicacions web.
