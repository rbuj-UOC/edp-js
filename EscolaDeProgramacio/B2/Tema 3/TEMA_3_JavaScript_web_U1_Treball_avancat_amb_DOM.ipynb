## Unitat 1: Treball avançat amb el DOM

En aquesta unitat explorarem algunes tècniques avançades, o menys habituals, per treballar amb el DOM. Veurem com podem utilitzar un element de HTML per crear plantilles de contingut que podem clonar i reutilitzar a la nostra pàgina. També descobrirem de quina manera podem utilitzar atributs personalitzats per emmagatzemar dades pròpies en els elements de HTML. ### Etiqueta `<template>`

Quan programem aplicacions web és habitual trobar-nos amb la necessitat de mostrar un conjunt de dades que segueixen una estructura repetitiva, bé utilitzant un format tabular amb `<table>`, o bé mitjançant un disseny amb contenidors `<div>` o altres etiquetes semàntiques.

En aquests casos, podem emprar els mètodes habituals per manipular el DOM i afegir nous elements, per exemple `createElement()` i `appendChild()`. Tanmateix, si cada un dels elements té una estructura HTML d'una certa complexitat (per exemple, amb presència de molts atributs i classes), aquesta tasca es pot tornar tediosa. Suposem, per exemple, que volem mostrar un grup de fotografies i que per a cada una hem d'utilitzar aquest codi HTML:

```javascript
<div class="picture">
  <img src="--url here--" class="picture" alt="--title here--" />
  <p class="title">--title here--</p>
  <p class="author">--author here--</p>
</div>
```

Amb els mètodes tradicionals de manipulació del DOM a JavaScript, hauríem d'usar `createElement()` quatre vegades: per a `<div>`, `<img>` i `<p>` en dues ocasions. A més, hauríem d'assignar quatre classes CSS afegint-les a les corresponents `classList` i ajustar altres propietats com `src`, `alt` o `textContent`. Aquest enfocament és perfectament vàlid, però implica escriure massa línies de codi per aconseguir un resultat bastant simple.

Si observem el bloc HTML anterior podem veure que, en realitat, només hi ha quatre paràmetres per informar: el títol de la fotografia dues vegades, l'autor i la URL corresponent a la imatge. La resta d'elements són fixos i no hauria de caldre repetir-los cada vegada.

Per a aquests casos, en HTML disposem de l'etiqueta `<template>`. Aquesta etiqueta ens permet definir un bloc de codi HTML que el navegador no mostrarà a la pàgina, però que podem utilitzar com a plantilla per crear nous elements des de JavaScript. Amb el seu ús millorem l'eficiència del codi en evitar la repetició i facilitar la clonació d'estructures HTML complexes. Aquest mètode és especialment útil en aplicacions web dinàmiques on el contingut es genera o actualitza freqüentment basat en dades d'usuari o respostes de servidor.

```javascript
<template id="template-picture">
  <div class="picture">
    <img src="--url here--" class="picture" alt="--title here--" />
    <p class="title">--title here--</p>
    <p class="author">--author here--</p>
  </div>
</template>
```

Com que els navegadors web ignoren aquesta etiqueta (li asignen un estil `display: none`), aquest bloc pot aparèixer en qualsevol lloc del document HTML. Tanmateix, per mantenir el document ordenat, és recomanable acumular totes les etiquetes `<template>` a la part final, per exemple, just abans del tancament de l'etiqueta `<body>`.

A partir d'aquí, des de JavaScript podem utilitzar aquesta plantilla per mostrar la informació relativa a una fotografia d'una manera molt més abreujada i amb menys possibilitat d'errors. Ho faríem de la manera següent:

```javascript
// Let's grab a reference to <div> with class `pictures`.
const container = document.querySelector('.pictures');

// Let's grab a reference to our <template> for each picture.
const template = document.querySelector('#template-picture').content;

// Let's loop through all pictures.
for (const picture of pictures) {
  // Create a clone based on template.
  const clone = template.cloneNode(true);

  // Customize properties.
  clone.querySelector('img').src = picture.url;
  clone.querySelector('img').alt = picture.title;
  clone.querySelector('.title').textContent = picture.title;
  clone.querySelector('.author').textContent = picture.author;

  // And add clone to container.
  container.appendChild(clone);
}
```

La part significativa d'aquest codi i la que realment mostra el benefici d'utilitzar `<template>` és la formada per les primeres cinc línies del bucle, que substitueixen les prop de 20 línies que necessitem per fer el mateix si optem per emprar els mètodes `createElement()` i `appendChild()` per portar a terme la manipulació del DOM.

Finalment, afegiríem el fragment a la pàgina amb un únic `appendChild()`. D'aquesta manera, el navegador només hauria d'actualitzar el DOM una vegada, en comptes de fer-ho cada vegada que afegim un nou element. > 
> `fragment.appendChild(clone)`
> 
> `container.appendChild(fragment)`

### Atributs data-*

Tots els elements HTML poden incloure atributs que proporcionen informació addicional útil per definir la seva representació o el seu comportament. Aquests atributs sempre es defineixen a l'etiqueta d'obertura i normalment estan formats per parelles de nom i valor amb la sintaxi `name="value"`.

Consultant la documentació de qualsevol element podem esbrinar quins són els atributs que té definits, que seran els que els navegadors web tindran en compte a l'hora de representar aquest element en el document web. Així, per a l'element `img` tenim disponible l'atribut `src`, però en canvi aquest atribut no està definit per als elements `p` o `li`. Nosaltres podem informar atributs addicionals no documentats, però els navegadors no els tindran en compte a l'hora de representar l'element.

Tanmateix, hi ha uns atributs especials, que anomenem de manera genèrica atributs `data-*`, que ens permeten afegir qualsevol tipus d'informació addicional que vulguem a un element, sovint amb la intenció de poder-la recuperar des de JavaScript quan accedim a aquest element a través del DOM.

Suposem que en el nostre document HTML tenim definida una imatge així:

```javascript
<img src="dyson-v8.png" alt="Dyson v8" id="dyson-v8" class="vacuum">
```

Sabem que des de JavaScript la podem recuperar de diverses maneres: mitjançant l'identificador, a través de la seva classe CSS o fins i tot utilitzant la seva etiqueta i posició relativa en el document. La manera més directa segurament seria la següent:

```javascript
const myDyson = document.querySelector('#dyson-v8');
```

Arribats a aquest punt, pot ser que necessitem que aquest element `<img>` contingui informació addicional, no relacionada amb la seva representació o comportament, sinó per a ús exclusiu per part del nostre codi JavaScript. Potser necessitem que inclogui un altre identificador, una categoria, la puntuació mitjana sobre aquest producte per part dels clients, un enllaç a un altre document, etc. En cada cas concret es pot tractar d'informació totalment diferent. És en aquest escenari on podem usar els atributs `data-*` de manera lliure, tal com es pot veure a continuació:

```javascript
<img src="dyson-v8.png" alt="Dyson v8" id="dyson-v8" class="vacuum" 
  data-sql-id="42" data-category="Cordless" data-score="4.8">
```

El navegador ignorarà tots els atributs que comencen per `data-`, però nosaltres hi podem accedir des de JavaScript de manera molt senzilla mitjançant `getAttribute()`:

```javascript
console.log(myDyson.getAttribute('data-sql-id'));
console.log(myDyson.getAttribute('data-category'));
console.log(myDyson.getAttribute('data-score'));
```

O també hi podem accedir a través de la propietat `dataset` de l'element. Aquesta propietat és un objecte que conté tots els atributs `data-*` de l'element i ens permet accedir-hi de manera directa. Aquest segon mètode és més net i recomanat actualment. Per accedir als atributs a partir d'aquest objecte `dataset` hem de tenir en compte que els noms no inclouen el prefix `data-` i que qualsevol caràcter `-` s'elimina i se substitueix per una majúscula, seguint una notació _camelCase_ habitual. Per tant, l'atribut `data-sql-id` queda redefinit com la propietat `sqlId`. Amb aquest segon sistema recuperaríem els valors anteriors de la manera següent:

```javascript
console.log(myDyson.dataset.sqlId);
console.log(myDyson.dataset.category);
console.log(myDyson.dataset.score);
```

L'ús dels atributs `data-*` pot ser especialment interessant, per exemple, quan definim un mateix esdeveniment per a molts elements al mateix temps, però en el moment d'executar el codi corresponent a aquest esdeveniment volem tenir accés a algunes propietats que ens permetin saber quin element en concret l'ha desencadenat. Els atributs `data-*` proporcionen una manera flexible d'emmagatzemar i recuperar dades específiques relacionades amb un element, i faciliten la identificació i el maneig individualitzat dins d'un controlador d'esdeveniments comú.

### Web components

En el món del desenvolupament web modern, els _web components_ representen un avenç significatiu en la creació d'interfícies d'usuari personalitzades i reutilitzables. Aquests components ens permeten als desenvolupadors crear elements HTML propis, amb una aparença i funcionalitat específiques. Mitjançant web components podem dissenyar widgets o components complexos que poden ser utilitzats de manera consistent en diferents parts d'una aplicació web. Es tracta, doncs, d'una solució modular i mantenible per construir aplicacions web escalables.

Per crear un nou component web a JavaScript necessitem definir una classe que estengui `HTMLElement` i després registrar el nou element amb el nom que vulguem. A partir d'aquí, el nou element podrà ser utilitzat en l'HTML com qualsevol altre element de HTML estàndard.

```javascript
class AwesomeParagraph extends HTMLElement {
  constructor() {
      super();
      this.innerHTML = "<p>This is an awesome paragraph!</p>";
  }
}

customElements.define('awesome-paragraph', AwesomeParagraph);
```

Anem una mica més enllà i suposem que en la nostra aplicació (algun tipus de joc) necessitem utilitzar sovint un dau típic de sis cares, como el que s'utilitza en el parxís. Podem crear un web component i definir-lo com l'element `<dice-roller>` que després podrem utilitzar directament en els nostres documents HTML.

A la classe `DiceRoller` corresponent, definim un constructor que crea un element `button` i l'afegeix al component. Aquest botó té un controlador d'esdeveniments de tal manera que cada vegada que l'usuari fa clic es genera un nombre aleatori entre 1 i 6. El codi complet d'aquest nou element és el següent:

```javascript
class DiceRoller extends HTMLElement {
  constructor() {
    super();
    this.number = 1;
    this.innerHTML = `<button>${this.number}</button>`;
    this.button = this.querySelector('button');
    this.button.addEventListener('click', () => this.roll());
  }

  roll() {
    // Roll the dice!
    const interval = setInterval(() => {
      this.number = Math.ceil(Math.random() * 6);
      this.button.textContent = this.number;
    }, 100);

    // Stop rolling after 2 seconds.
    setTimeout(() => {
      clearInterval(interval);
      console.log('You rolled a', this.number, '(DiceRoller)');
    }, 2000);
  }
}

customElements.define('dice-roller', DiceRoller);
```

Podem fins i tot personalitzar aquest dau i crear-ne una variant `<trick-dice>` que ens permeti fer trampes en el nostre joc. En aquest cas, el dau admet un atribut `favorite` mitjançant el qual li podem indicar quin és el nombre que hauria de sortir amb més freqüència. Aquest seria el codi complet d'aquest nou element:

```javascript
class TrickDice extends HTMLElement {
  constructor() {
    super();
    this.number = 1;
    this.favorite = parseInt(this.getAttribute('favorite') || 0);
    this.innerHTML = `<button>${this.number}</button>`;
    this.button = this.querySelector('button');
    this.button.addEventListener('click', () => this.roll());
  }

  roll() {
    const interval = setInterval(() => {
      this.number = Math.ceil(Math.random() * 6);
      if (this.favorite > 0 && this.favorite <= 6) {
        this.number = Math.random() < 0.5 ? this.number : this.favorite;
      }
      this.button.textContent = this.number;
    }, 100);

    setTimeout(() => {
      clearInterval(interval);
      console.log('You rolled a', this.number, '(TrickDice)');
    }, 2000);
  }
}

customElements.define('trick-dice', TrickDice);
```

Les úniques diferències entre `DiceRoller` i `TrickDice` són les següents:

- El constructor de `TrickDice` busca un atribut `favorite` i el desa en una propietat amb el mateix nom. Si no el troba, estableix el valor per defecte a 0.
- El mètode `roll()` de `TrickDice` genera un nombre aleatori entre 1 i 6, igual que fa `DiceRoller`, però a continuació decideix al 50 % si la propietat `number` queda assignada a aquest nombre aleatori acabat de calcular o se li assigna de manera directa el valor de la propietat `favorite`. És a dir, que el nostre dau trucat té un 50 % de possibilitats de mostrar el nombre que li hem indicat com a favorit.
- Si no informem de cap atribut `favorite` o li assignem un valor per sobre de 6, el dau trucat es comportarà exactament igual que el dau normal.
