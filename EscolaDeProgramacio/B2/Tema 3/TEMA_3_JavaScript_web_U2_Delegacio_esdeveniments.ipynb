## Unitat 2: Delegació d'esdeveniments

La delegació d'esdeveniments en JavaScript és un mecanisme que ens permet una gestió més eficient i efectiva dels esdeveniments, sobretot en situacions on hi ha múltiples elements que requereixen un maneig d'esdeveniments similar.

En termes generals sabem que quan volem gestionar un esdeveniment en un element del DOM, ho podem fer a través del mètode `addEventListener()`:

```javascript
document.querySelector('#save').addEventListener('click', myEventHandler);
```

En el codi anterior hem afegit un controlador d'esdeveniments a l'element que té l'identificador `save`. En concret, controlem què ha de passar quan es produeixi l'esdeveniment `click`: la funció `myEventHandler()` s'executarà quan es produeixi aquest esdeveniment. És a dir, quan un usuari faci clic en el botó que té l'atribut `id` informat al valor `save` s'executarà el codi de la funció `myEventHandler()`.

També vam veure que la funció que passem com a callback a `addEventListener()` rep, com a primer paràmetre, les metadades de l'esdeveniment que s'ha produït (normalment utilitzem el nom `e`, `evt` o `event`). Sovint, tanmateix, necessitem gestionar esdeveniments en múltiples elements del DOM. Per exemple, suposem que tenim un contenidor amb un rast de deu estrelles i volem que, quan l'usuari faci clic a qualsevol d'elles, es mostri un missatge a la consola del navegador indicant la valoració que l'usuari ha seleccionat. Es tracta d'un sistema de valoració que podem trobar en moltes pàgines web, amb una interfície formada per unes quantes estrelletes (normalment cinc o deu) que l'usuari pot marcar per indicar la seva valoració sobre un producte o servei. En HTML podem crear un contenidor amb deu estrelles de la manera següent:

```javascript
<div class="star-container">
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
  <div class="star star-gray"></div>
</div>
```

Partint d'aquest DOM, si optem per utilitzar el mètode `addEventListener()` per a cada una de les estrelles, el codi podria quedar com es pot veure tot seguit:

```javascript
// Adds a different event listener to each star.
function addEventsToAllStars() {
  const stars = document.querySelectorAll('.star')
  stars.forEach((star) => star.addEventListener('click', updateRatingWithManyEvents))
}

// Updates rating selected by user using many events.
function updateRatingWithManyEvents(event) {
  // Get a reference to `<div>` element clicked by user.
  const clickedStar = event.target

  // Get a reference to all `<div>` elements with class `star`.
  const stars = document.querySelectorAll('.star')

  // Remove `star-pink` class from all `<div>` elements.
  stars.forEach((star) => star.classList.remove('star-pink'))

  // Add `star-pink` class to all `<div>` elements until we reach the one clicked by user.
  let rating = 0

  for (const star of stars) {
    star.classList.add('star-pink')
    rating++
    if (star === clickedStar) break
  }

  console.log('User selected rating', rating)
}

// Example usage.
addEventsToAllStars()
```

Aquest codi funciona perfectament. Comença afegint un controlador d'esdeveniments a cada una de les estrelles. Quan l'usuari fa clic en una, s'executa la funció `updateRatingWithManyEvents()`, que mostra un missatge a la consola del navegador indicant el número d'estrella que s'ha premut. La lògica interna d'aquesta funció es pot seguir a través dels comentaris que acompanyen el codi font.

Tanmateix, aquest enfocament té un problema: estem afegint un controlador d'esdeveniments a cada una de les estrelles. Si tenim deu estrelles, estem afegint deu controladors d'esdeveniments. Si tinguéssim cent elements, estaríem afegint cent controladors d'esdeveniments. No sembla un enfocament net i eficient.

Per solventar-ho, podem utilitzar la propagació d'esdeveniments i la delegació d'esdeveniments.

> <br>🧐&nbsp; **Són el mateix la _delegació d'esdeveniments_ i la _propagació d'esdeveniments (event bubbling)_?**
>
> No exactament. Tot i que és cert que ambdós conceptes estan molt relacionats.
> 
> La propagació d'esdeveniments és un mecanisme pel qual un esdeveniment en un element fill es propaga a través dels seus ancestres en el DOM. Per exemple, si feu clic en un botó, l'esdeveniment de clic no només passa en el botó, sinó que també "bombolleja" cap amunt en el DOM, i pot ser captat pels seus elements pares. Per això es coneix com a _event bubbling_ (de _bubble_, bombolla en anglès).
> 
> La delegació d'esdeveniments, en canvi, és una tècnica que s'aprofita d'aquest bombolleig cap amunt dels esdeveniments per afegir un únic controlador d'esdeveniments en un element pare, en comptes de múltiples controladors en elements individuals. Per exemple, en comptes d'assignar un gestor de clic a cada botó, podeu afegir un sol gestor en el seu contenidor.
> 
> Així doncs, el _bubbling_ descriu com els esdeveniments es propaguen a través del DOM, mentre que la delegació s'aprofita d'aquesta característica per manejar esdeveniments de manera més eficient.
> <br><br>

Vegem de quina manera podem aprofitar aquesta característica del DOM per millorar el nostre codi anterior. En comptes d'afegir un controlador d'esdeveniments a cada estrella, podem afegir un controlador d'esdeveniments al contenidor que les conté. D'aquesta manera, quan l'usuari faci clic en una estrella, l'esdeveniment es propagarà cap amunt en el DOM i serà capturat pel contenidor. En el controlador d'esdeveniments del contenidor podem esbrinar quina estrella s'ha premut i actuar en conseqüència. El codi podria quedar com es pot veure tot seguit:

```javascript
// Adds a single event listener to container.
function addEventToStarContainer() {
  const starContainer = document.querySelector('.star-container')
  starContainer.addEventListener('click', updateRatingEventDelegation)
}

// Updates rating selected by user using event delegation.
function updateRatingEventDelegation(event) {
  // Check whether user has actually clicked on a star or the container itself.
  const starContainer = document.querySelector('.star-container')
  if (event.target === starContainer) return

  // Get a reference to `<div>` element clicked by user.
  const clickedStar = event.target

  // Get a reference to all `<div>` elements with class `star`.
  const stars = document.querySelectorAll('.star')

  // Remove `star-pink` class from all `<div>` elements.
  stars.forEach((star) => star.classList.remove('star-pink'))

  // Add `star-pink` class to all `<div>` elements until we reach the one clicked by user.
  let rating = 0

  for (const star of stars) {
    star.classList.add('star-pink')
    rating++
    if (star === clickedStar) break
  }

  console.log('User selected rating', rating)
}

// Example usage.
addEventToStarContainer()
```

En aquest cas no comencem afegint un controlador d'esdeveniments a cada una de les estrelles, sinó un d'únic al contenidor. Després comprovem si l'element que ha disparat l'esdeveniment és una estrella (per descartar que el clic s'hagi produït directament en el contenidor, però fora de qualsevol estrella). Si és el cas, seguim la mateixa lògica que a l'exemple anterior.

Amb tots dos enfocaments acabem obtenint el mateix resultat, però el segon és molt més eficient, ja que només afegim un controlador d'esdeveniments en comptes de deu (o de tants com elements iguals tinguem a la nostra interfície d'usuari).

### Alguns casos d'ús en aplicacions reals

La delegació d'esdeveniments és una tècnica molt utilitzada en aplicacions web professionals. Vegem-ne alguns exemples:

- **Gestió de llistes dinàmiques** – En una aplicació web on els elements d'una llista (com comentaris, tuits o articles en un blog) s'afegeixen o s'eliminen dinàmicament, la delegació d'esdeveniments resulta molt útil. En comptes d'assignar un controlador d'esdeveniments a cada element individualment, es pot assignar un sol controlador a l'element contenidor de la llista. Aquest enfocament assegura que fins i tot els elements afegits dinàmicament responguin als esdeveniments sense necessitat d'assignar nous controladors cada vegada que s'afegeix un element.

- **Taules editables** – En una taula amb files que poden ser editades, eliminades o seleccionades, en comptes d'adjuntar un controlador d'esdeveniments a cada botó o camp a les files, es pot utilitzar la delegació d'esdeveniments. Si assignem el controlador a l'element `table` o al `tbody`, podem controlar accions com clics en botons d'edició o eliminació. Això permet reduir el nombre de controladors d'esdeveniments i facilita la gestió del codi i el seu manteniment.

- **Menús desplegables amb múltiples opcions** – En un menú desplegable amb múltiples opcions, com un menú de navegació, la delegació d'esdeveniments ens permet gestionar les interaccions d'usuari de manera eficient. Podem assignar un únic controlador d'esdeveniments al contenidor del menú que s'encarregui de gestionar els clics sobre cada una de les opcions; així s'evita la necessitat d'adjuntar controladors d'esdeveniments a cada opció individual del menú.

- **Galeries d'imatges dinàmiques** – En un lloc web que presenta una galeria d'imatges on es poden afegir noves imatges o canviar o eliminar les existents, usar la delegació d'esdeveniments per gestionar diferents accions (per exemple, l'usuari passant el ratolí per sobre de cada una de les imatges) és una estratègia eficaç. Si assignem el controlador d'esdeveniments al contenidor de la galeria, qualsevol interacció amb cada una de les imatges es podrà gestionar sense necessitat d'adjuntar controladors a cada imatge nova que s'afegeixi.

Així doncs, la delegació d'esdeveniments no només millora el rendiment de les nostres aplicacions en reduir la quantitat de controladors d'esdeveniments necessaris, sinó que també facilita aspectes com l'escalabilitat i el manteniment del codi. La seva comprensió i aplicació adequada ens permeten simplificar les nostres aplicacions.
