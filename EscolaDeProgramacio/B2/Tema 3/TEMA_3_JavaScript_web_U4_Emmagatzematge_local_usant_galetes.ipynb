## Unitat 4: Emmagatzematge local usant galetes

En el nivell B1 vàrem tenir ocasió d'estudiar el funcionament de l'emmagatzematge local mitjançant _local storage_. En aquest nivell B2 estudiarem el funcionament d'un altre mecanisme per emmagatzemar algunes dades a la màquina de l'usuari: les galetes.

Les galetes s'utilitzen en aplicacions web per emmagatzemar informació en el navegador de l'usuari i poden servir per a finalitats com mantenir l'estat de la sessió, personalitzar experiències d'usuari i rastrejar dades per a anàlisis. En aquesta unitat aprendrem com crear, llegir i eliminar galetes, així com algunes bones pràctiques i consideracions de seguretat en manejar-les, per garantir una gestió eficient i responsable de les dades de l'usuari.

Les galetes consisteixen en petits fitxers de text que s'utilitzen per emmagatzemar informació sobre la sessió de l'usuari, les preferències i altres dades que necessiten persistir entre diferents visites a una pàgina web. Cada vegada que el mateix usuari torna a visitar la nostra pàgina web, el navegador envia les galetes rellevants al servidor; d'aquesta manera, permet que aquest recordi informació sobre l'usuari i la seva interacció prèvia amb la pàgina. 

A causa de la mida limitada i, sobretot, de qüestions cada vegada més importants relacionades amb la privacitat, és important usar-les de manera responsable i segura. Les galetes (o algun mecanisme anàleg com el que vàrem tenir ocasió d'estudiar en el nivell B1) resulten fonamentals per oferir algunes funcionalitats web, com el manteniment de sessions d'usuari i la personalització de continguts.

Vegem un primer exemple de codi per crear una galeta:

```javascript
// Set a cookie with a name, value and number of days until expiration.
function setCookie(name, value, days) {
  // By default, there is no expiration so the cookie is temporary.
  let expires = '';
  // If we have a number of days, we calculate the expiration date.
  if (days) {
    let date = new Date();
    date.setTime(date.getTime() + days * 24 * 60 * 60 * 1000);
    expires = '; expires=' + date.toUTCString();
  }
  // Set the cookie.
  document.cookie = name + '=' + (value || '') + expires + '; path=/';
}
```

Aquesta funció rep com a paràmetres el nom de la galeta, el valor que li volem assignar i el nombre de dies que volem que duri. La part més rellevant de la funció és la crida a `document.cookie`, que és la que realment crea la galeta.

A continuació necessitem una altra funció que ens permeti recuperar el valor d'una galeta:

```javascript
// Get the value of a cookie by its name.
function getCookie(name) {
  // Create a key with the cookie name, followed by '='
  const nameEqual = name + '=';

  // Split document.cookie on semicolons into an array.
  const cookies = document.cookie.split(';');

  // Loop through the cookies array and check if the cookie name is found.
  for (let cookie of cookies) {
    cookie = cookie.trim();
    if (cookie.startsWith(nameEqual)) {
      return cookie.substring(nameEqual.length);
    }
  }

  // Return null if the cookie has not been found.
  return null;
}
```

Aquesta funció rep com a paràmetre el nom de la galeta que volem recuperar. Si la galeta existeix, la funció retorna el seu valor. Si no existeix, retorna un valor nul (tot i que podríem retornar qualsevol altre valor que ens interessi). Novament, la part més rellevant de la funció és la crida a `document.cookie`, que és la que recupera el valor de la galeta.

Una vegada disposem d'aquestes dues funcions, vegem com podem crear i recuperar galetes:

```javascript
// Do not attempt to run this code in a notebook. It requires `document` to be defined.
// Example usage. Set a cookie.
setCookie('user', 'John Doe', 7)
// setCookie('age', 45, 7)

// Do not attempt to run this code in a notebook. It requires `document` to be defined.
// Example usage. Get a cookie that exists.
const user = getCookie('user')
console.log(user)

// Do not attempt to run this code in a notebook. It requires `document` to be defined.
// Example usage. Get a cookie that does not exist.
const age = getCookie('age')
console.log(age)
```

Finalment, per eliminar una galeta en JavaScript, la pràctica habitual consisteix a sobreescriure la galeta per establir la data d'expiració en un moment passat. Això es fa mitjançant l'ajustatge del paràmetre `expires`:

```javascript
// Delete a cookie by its name.
function deleteCookie(name) {
  // Set the expiration date to the past.
  document.cookie = name + '=; expires=Thu, 01 Jan 1970 00:00:01 GMT;';
}

// Do not attempt to run this code in a notebook. It requires `document` to be defined.
// Example usage. Delete a cookie.
deleteCookie('user');
```

Tot i que temps enrere les galetes s'usaven per aconseguir aquest emmagatzematge al costat del client que acabem de veure, ara es recomana utilitzar altres API més modernes per a aquesta finalitat, com la Web Storage (`localStorage` i `sessionStorage`) i també IndexedDB. Fer-ho d'aquesta manera millora el rendiment de la nostra pàgina web, perquè les galetes s'envien amb cada sol·licitud, mentre que les API d'emmagatzematge local no ho fan.

En canvi, les galetes segueixen tenint un paper rellevant en aspectes que incumbeixen al servidor, com per exemple la gestió de sessions d'usuari. En aquest cas, la galeta s'utilitza per emmagatzemar un identificador únic de sessió, i la resta de dades es poden emmagatzemar en el servidor. Gràcies a això, el servidor pot identificar l'usuari a cada sol·licitud, i recuperar les dades de sessió associades a aquest identificador.

En rebre una sol·licitud HTTP, el servidor web pot enviar una o més capçaleres de `Set-Cookie` en la resposta. El navegador normalment emmagatzemarà les galetes pertinents i les enviarà de tornada en posteriors sol·licituds al mateix servidor. En aquest context, aquestes galetes són petits blocs de dades que el servidor web envia al navegador de l'usuari. Aquest navegador les rep, les desa i s'encarrega d'adjuntar-les de tornada al servidor en cada sol·licitud posterior, permetent així que el servidor identifiqui si dues sol·licituds provenen del mateix navegador. Amb això el servidor aconsegueix mantenir una sessió d'usuari activa i, en general, pot simular una certa gestió de l'estat, que és una característica que el protocol HTTP no ofereix, en ser un tipus de protocol sense estat.
