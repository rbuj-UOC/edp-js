# TEMA 1: Ús avançat de funcions

## Unitat 5: Closures

El concepte de _closure_ (sense traducció amb ús real en català) és una característica potent de JavaScript que ens permet accedir a l'àmbit d'una funció exterior des d'una funció interior, mantenint així viu l'estat i el context d'aquesta funció exterior. Dit d'una manera més senzilla, un closure ens dona accés a l'abast d'una funció externa des d'una funció interna. Cada vegada que es crea una funció s'està creant un closure.

### Exemple introductori: salutació personalitzada

En tant que el concepte de closure és una mica abstracte, resulta més senzill d'explicar i d'entendre mitjançant exemples. Vegem-ne un primer cas molt senzill:

```javascript
function createGreeter() {
  const greet = "Hello";

  function sayHello(name) {
    console.log(`${greet} ${name}`);
  }

  return sayHello;
}

const greeter = createGreeter();
greeter('Dolly');
```

Analitzem pas a pas el que fa el codi anterior:

- En primer lloc, declarem una funció externa anomenada `createGreeter()` que no rep cap paràmetre. Aquesta funció té una variable interna `greet` que té assignat el valor "Hello". A continuació, declarem una funció interna anomenada `sayHello()` que rep un nom per paràmetre i que mostra per consola el valor de la variable `greet` juntament amb aquest nom. És a dir, si cridem la funció `sayHello("John")`, mostrarà per consola "Hello John". Finalment, la funció `createGreeter()` retorna una referència a la funció interna `sayHello()`.

- Observem que la variable `greet` està definida dins de la funció `createGreeter()`, però no dins de la funció `sayHello()`. Tanmateix, la funció `sayHello()` pot accedir a la variable `greet` i mostrar-ne el valor per consola. Això és possible gràcies als closures. Quan s'executa la funció `createGreeter()`, es crea un closure que conté l'àmbit de la funció `createGreeter()`. Aquest closure s'assigna a la funció `sayHello()`. Quan s'executa aquesta funció `sayHello()`, busca la variable `greet` dins del seu propi àmbit. Si no la troba, busca en l'àmbit del seu closure, és a dir, en l'àmbit de la funció `createGreeter()`. Gràcies a això, la funció `sayHello()` pot accedir a la variable `greet` i mostrar-ne el valor per consola.

- Fora de les funcions, creem una variable `greeter` que té assignat el valor retornat per la funció `createGreeter()`. És a dir, la variable `greeter` conté la funció `sayHello()`. D'aquesta manera podem executar la funció `sayHello()` a través de la variable `greeter`. Quan executem `greeter("Dolly")` es mostra per consola "Hello Dolly". Aquest "Hello" inicial està disponible gràcies al fet que la funció `sayHello()` té accés a l'àmbit de la funció `createGreeter()` a través del closure que es va crear quan es va executar la funció `createGreeter()`.

Podria semblar poc intuïtiu que aquest codi funcioni correctament. En alguns llenguatges de programació, les variables locals dins d'una funció (en el nostre cas, `greet`) existeixen únicament durant l'execució d'aquesta funció. Una vegada que `createGreeter()` s'acaba d'executar s'assumeix que la variable `greet` ja no estarà disponible. Tanmateix, en JavaScript el codi funciona correctament, perquè en JavaScript les funcions formen closures. 

Un closure és la combinació d'una funció i l'entorn lèxic dins del qual es va declarar aquesta funció. Aquest entorn està format per qualsevol variable local que estigués dins de l'abast en el moment en què es va crear aquest closure. En el nostre cas, `greeter` és una referència a la funció `sayHello()` que es crea quan s'executa `createGreeter()`. Aquesta funció `sayHello()` manté una referència al seu entorn lèxic, dins del qual hi ha la variable `greet`. Així, quan executem la funció `sayHello()` a través de la variable `greeter`, aquesta funció pot accedir a la variable `greet` que ha conservat dins del seu entorn lèxic.

Dit en termes molt senzills, quan es crea una funció, aquesta funció manté una referència a totes les variables a les quals tenia accés en el moment de la seva creació, encara que aparentment aquestes variables ja no existeixin. Això es el que permet que el codi anterior funcioni de la manera que ho fa.

### Cas pràctic: comptador

Imaginem que volem crear una funció que ens retorni un comptador. Cada vegada que cridem aquesta funció, ens retornarà un comptador independent de qualsevol altre comptador. I tots els comptadors tindran les funcions per incrementar, decrementar i resetejar. Podríem fer-ho de la manera següent:

```javascript
function createCounter() {
  // This works as a private variable.
  let _counter = 0;

  function increment() {
    _counter++;
  }

  function decrement() {
    _counter--;
  }

  function reset() {
    _counter = 0;
  }

  function value() {
    return _counter;
  }

  // This is what the outside world can see.
  return { increment, decrement, reset, value };
}
```

```javascript
// Create two counters.
const firstCounter = createCounter();
const secondCounter = createCounter();

// Log the counter value to the console. Expected 0 0.
console.log(firstCounter.value());
console.log(secondCounter.value());

// Increment the first counter five times.
for (let i = 1; i <= 5; i++) {
  firstCounter.increment();
}

// Increment the second counter nine times.
for (let i = 1; i <= 9; i++) {
  secondCounter.increment();
}

// Log the counter value to the console. Expected 5 9.
console.log(firstCounter.value());
console.log(secondCounter.value());

// Reset the first counter.
firstCounter.reset();

// Decrement the second counter three times.
secondCounter.decrement();
secondCounter.decrement();
secondCounter.decrement();

// Log the counter value to the console. Expected 0 6.
console.log(firstCounter.value());
console.log(secondCounter.value());
```

Aquest codi defineix una funció `createCounter()` que utilitza un closure per encapsular una variable privada `_counter` i proporcionar un conjunt de funcions per manipular-la. Aquestes funcions són `increment()`, `decrement()`, `reset()` i `value()`. Més en detall:

- `_counter` manté l'estat del comptador i no és accessible directament des de fora de `createCounter()`.
- `increment()` i `decrement()` modifiquen `_counter` incrementant-ne o decrementant-ne el valor, respectivament.
- `reset()` reinicia `_counter` a 0.
- `value()` retorna el valor actual de `_counter` en cada moment.

En cridar la funció `createCounter()`, es crea una nova instància de comptador amb el seu propi àmbit i estat; això permet que `firstCounter` i `secondCounter` operin de manera independent. Això és possible gràcies al funcionament dels closures. Cada un manté el seu propi valor de `_counter`, i les operacions en un no afecten l'altre, demostrant com els closures poden ser utilitzats per crear dades privades i funcions relacionades en JavaScript.

Altres llenguatges de programació disposen de modificadors d'accés com `private`. Tot i que JavaScript no els té, podem simular la privacitat utilitzant closures. En aquest exemple del comptador, `_counter` actua com una variable privada perquè no és accessible directament des de fora de `createCounter()`. Només les funcions definides dins de `createCounter()` poden accedir a `_counter`, gràcies al closure que es forma quan `createCounter()` és invocada. Aquest patró s'utilitza de manera comuna per encapsular l'estat de manera segura i evitar accessos o modificacions no autoritzades.

### Altres possibles casos pràctics

1. **Gestionar la privacitat en objectes de configuració** – De manera similar al comptador que acabem de veure, els closures es poden utilitzar per crear objectes de configuració en els quals volem que certes propietats romanguin privades i només es puguin modificar a través de mètodes específics. Per exemple, podríem tenir un objecte de configuració que emmagatzemi el nom d'usuari i la contrasenya, però que només permeti modificar la contrasenya a través d'un mètode `setPassword()`.

2. **Fàbriques de funcions** – Els closures permeten la creació de funcions personalitzades basades en arguments. Per exemple, podríem tenir una funció que creï noves funcions internes per calcular diferents tipus d'impostos, on cada funció interna generada manté el seu propi conjunt de normes fiscals.

3. **Control d'accés a recursos limitats** – Podem utilitzar closures per controlar l'accés a recursos que han de ser limitats o regulats. Si la nostra aplicació treballa amb una base de dades, ens pot interessar programar una funció que gestioni les connexions a aquesta base de dades mitjançant un closure, per mantenir així un comptador de quantes connexions estan actives, que permet la creació de noves només si no se supera un límit establert.

4. **Memorització de resultats de funcions costoses** – Els closures també es poden utilitzar per memoritzar (registrar) resultats d'operacions costoses. Imaginem que hem programat una funció que fa càlculs matemàtics complexos i que és probable que la cridem repetidament, moltes vegades passant-li els mateixos paràmetres. Aquesta funció podria emmagatzemar els seus resultats que va calculant utilitzant un closure, de tal manera que quan la cridem de nou amb els mateixos arguments, pugui retornar ràpidament el resultat emmagatzemat en comptes d'haver-lo de recalcular.

Els closures en JavaScript ens permeten millorar l'encapsulació i la gestió de dades privades. També ens permeten implementar característiques avançades com el maneig d'estat i la memorització de resultats. Mitjançant closures podem programar funcions més dinàmiques i personalitzades que ens permetin crear codi més net. I el concepte principal al voltant del qual gira tot aquest mecanisme és el d'àmbit lèxic. La idea que hem de tenir clara és que una funció pot accedir a totes les variables que estaven disponibles en el moment de la seva creació, fins i tot encara que aquestes variables ja no existeixin en el moment que s'executa la funció. Aquesta distinció és clau: moment en què es crea la funció i moment en què s'executa la funció. Els closures capturen l'àmbit lèxic de la funció en el moment en què es crea la funció.
