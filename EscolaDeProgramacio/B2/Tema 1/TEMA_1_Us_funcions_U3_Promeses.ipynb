## Unitat 3: Promeses

En el nivell B1 vàrem tenir un primer contacte amb les promeses. En concret, vàrem exposar el següent:

_[...] la funció `fetch()` és una funció asíncrona que no retorna directament una resposta, sinó que retorna només una garantia... una promesa (és a dir, un objecte de tipus `Promise`) del qual més endavant es rebrà una resposta._

_Quan escrivim el nostre codi [asíncron] hem de tenir present que les línies posteriors a l'enviament de la petició no poden utilitzar la resposta que s'espera rebre (si tot va bé, unes dades concretes), perquè quan s'executin aquestes línies posteriors encara no s'haurà rebut la resposta amb les dades per part del servidor. El que sí que es rep de manera immediata és una **promesa** que més endavant es rebrà una resposta real. Aquesta promesa consisteix en un objecte de tipus `Promise`. Diferents llenguatges de programació ofereixen solucions a aquest problema mitjançant l'ús de conceptes com futurs, promeses, retards o diferits, que sovint s'utilitzen de manera similar malgrat que la naturalesa de cada un presenta algunes diferències._

_Podem entendre el concepte de promesa mitjançant una analogia amb el rebut que ens donen quan portem un vestit a la tintoreria. Aquest rebut el podrem canviar pel nostre vestit rentat i planxat en algun moment futur, però ara mateix només és un tros de paper i no el podem utilitzar per vestir-nos. El rebut és, doncs, una promesa que més endavant tindrem el nostre vestit llest, però evidentment el rebut no és un vestit. Quan el rebut, al cap d'unes hores o uns dies, es converteixi en el vestit, el podrem per fi utilitzar per vestir-nos. Amb un objecte `Promise` passa el mateix. Quan enviem una petició a un servidor, obtenim de manera immediata una promesa que ens indica que més endavant tindrem dades disponibles, però ara mateix no tenim res real. Si intentem portar a terme alguna operació amb aquesta promesa, segurament tindrem problemes, perquè es tracta només d'una garantia que més endavant tindrem dades, però en aquest moment només és això... una promesa, un rebut. Al cap d'uns centenars de mil·lisegons, la promesa es traduirà en unes dades reals que, ara ja sí, podrem utilitzar per mostrar una nova informació a l'usuari de la nostra aplicació web._

Les promeses constitueixen un mecanisme essencial per manejar operacions asíncrones d'una manera més efectiva i organitzada en comparació amb els enfocaments tradicionals basats en callbacks. Les promeses ens permeten escriure codi asíncron més ben estructurat i més llegible i fàcil de mantenir.

Una promesa en JavaScript és essencialment un objecte que representa el resultat potencial d'una operació asíncrona. Aquest resultat pot ser l'èxit de l'operació (complerta) o el seu fracàs (refusada). Una promesa, per tant, només pot estar en un d'aquests tres estats: pendent, complerta o refusada. Això ens ajuda a controlar el flux d'operacions asíncrones i els possibles resultats.

Una promesa es crea utilitzant el constructor `Promise`, que requereix una funció executora com a argument. Aquesta funció executora s'invoca automàticament en construir la promesa i ha de tenir dos arguments: `resolve` i `reject`. Aquests arguments també són, al seu torn, funcions que es poden cridar dins de la funció executora per, respectivament, resoldre la promesa (indicar èxit) o refusar-la (indicar fracàs). L'estructura bàsica d'una promesa és la següent:

```javascript
// Create a new Promise object. The executor function is passed two parameters: resolve and reject.
const promise = new Promise((resolve, reject) => {
  // Place your asynchronous operation here.
  // This might involve fetching data from a server, reading a file, etc.
  
  // For demonstration, let's assume this asynchronous operation can either succeed or fail,
  // which is determined by the `success` variable.
  // In a real scenario, this could be determined by the result of the asynchronous operation.
  // Let's assume the operation was successful for this example.
  let success = true;  
  
  // Check the outcome of the asynchronous operation.
  if (success) {
    // If the operation succeeded, resolve the promise with a value.
    // `value` should be replaced with the actual data you want to return in case of success.
    resolve("Operation succeeded");
  } else {
    // If the operation failed, reject the promise with an error.
    // `error` should be replaced with an actual error object or message indicating the failure.
    reject("Operation failed");
  }
});
```

La funció `resolve(value)` es crida amb el valor resultant si l'operació asíncrona es completa amb èxit. En aquest moment l'estat de la promesa canvia a completa. En canvi, la funció `reject(error)` es crida amb un objecte d'error si l'operació falla, canviant l'estat de la promesa a refusada. Posteriorment podem utilitzar els mètodes `then()` i `catch()` per manejar els resultats de la promesa.

```javascript
// To handle the fulfilled or rejected state of the promise, use `then()` and `catch()` methods.
promise
  .then((value) => {
    // This code block is executed if the promise is resolved/succeeded.
    console.log(value);
  })
  .catch((error) => {
    // This code block is executed if the promise is rejected/failed.
    console.error(error);
  });
```

Una vegada vista l'estructura bàsica d'una promesa, aprofundirem en el seu ús pràctic. Comencem amb un exemple senzill. Suposem que tenim una funció asíncrona que retorna una promesa que es resol al cap de dos segons. Aquesta promesa es resol amb un valor numèric aleatori entre 1 i 10. La funció `generateRandomNumber()` es defineix de la manera següent:

```javascript
// Async function to generate a random number between 1 and 10 (2 seconds).
function generateRandomNumber() {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      const randomNumber = Math.floor(Math.random() * 10) + 1;
      resolve(randomNumber);
    }, 2000);
  });
}
```

Quan la vulguem utilitzar, no obstant això, hem de tenir en compte que la funció no retorna un nombre aleatori (no retorna dades, no ens lliura el vestit en la nostra analogia anterior) sinó que retorna una promesa (un rebut que ens garanteix que més endavant tindrem un nombre aleatori... o el nostre vestit net). Ho podem comprovar mostrant el resultat de cridar la funció `generateRandomNumber()`:

```javascript
console.log(generateRandomNumber())
```

Com veiem, es tracta d'una promesa. Si volem utilitzar el nombre aleatori que es genera al cap de dos segons, ho hem de fer dins de la funció `then()` que s'executa quan la promesa es resol. A l'exemple següent, la funció `generateRandomNumber()` s'invoca diverses vegades, i en cada invocació es mostra per consola el nombre aleatori que es genera al cap de dos segons:

```javascript
generateRandomNumber().then(result => console.log('First random number', result))
generateRandomNumber().then(result => console.log('Second random number', result))
generateRandomNumber().then(result => console.log('Third random number', result))
generateRandomNumber().then(result => console.log('Fourth random number', result))
```

En aquest cas els quatre nombres aleatoris probablement arriben en l'ordre esperat, perquè totes les promeses es resolen al cap de dos segons. Però en tractar-se d'una programació asíncrona, no tenim la garantia que sempre es produeixi d'aquesta manera. Si, per exemple, qui genera les promeses és la funció `fetch()` i les promeses es resolen quan es reben les dades del servidor, pot ser que la primera promesa es resolgui després que la segona, encara que s'hagi generat abans. Podem simular aquest fet adaptant l'exemple anterior perquè la promesa no es resolgui al cap de dos segons, sinó en un temps aleatori entre 0 i 5 segons:

```javascript
// Async function to generate a random number between 1 and 10 after (random time).
function generateRandomNumberInRandomTime() {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      const randomNumber = Math.floor(Math.random() * 10) + 1;
      resolve(randomNumber);
    }, Math.floor(Math.random() * 5000));
  });
}
```

```javascript
generateRandomNumberInRandomTime().then(result => console.log('First random number', result))
generateRandomNumberInRandomTime().then(result => console.log('Second random number', result))
generateRandomNumberInRandomTime().then(result => console.log('Third random number', result))
generateRandomNumberInRandomTime().then(result => console.log('Fourth random number', result))
```

Anteriorment hem comentat que una promesa només pot estar en un d'aquests tres estats: pendent, complerta o refusada. Quan una promesa es resol, es diu que la promesa es compleix. Podem detectar en quin moment passa perquè és quan s'executa el mètode `then()`. Quan una promesa es refusa, al contrari, s'executa el mètode `catch()`. En els exemples anteriors no hem tingut en compte el cas que la promesa es refusi. Per simular aquest cas, podem modificar la funció `generateRandomNumber()` perquè la promesa es refusi en comptes de resoldre's si el nombre aleatori generat és més petit que 5:

```javascript
// Async function to generate a random number between 1 and 10.
// If the number is lower than 5, promise will be rejected.
function generateRandomNumberHigherThan5() {
  return new Promise((resolve, reject) => {
    setTimeout(() => {
      const randomNumber = Math.floor(Math.random() * 10) + 1;
      if (randomNumber < 5) {
        reject('REJECTED!!!');
      }
      resolve(randomNumber);
    }, 2000);
  });
}
```

```javascript
generateRandomNumberHigherThan5()
  .then(result => console.log('First random number', result))
  .catch(error => console.log('First random number', error));
generateRandomNumberHigherThan5()
  .then(result => console.log('Second random number', result))
  .catch(error => console.log('Second random number', error));
generateRandomNumberHigherThan5()
  .then(result => console.log('Third random number', result))
  .catch(error => console.log('Third random number', error));
generateRandomNumberHigherThan5()
  .then(result => console.log('Fourth random number', result))
  .catch(error => console.log('Fourth random number', error));
```

En els exemples anteriors estem utilitzant el mètode constructor de l'objecte `Promise`, tal com es va exposar en el nivell B1 (l'ús de `new` ens ho indica clarament). Aquest mètode constructor rep com a paràmetre una funció que, al seu torn, rep dos paràmetres: `resolve` i `reject`. Aquests dos paràmetres són funcions que s'executen quan la promesa es resol o es refusa, respectivament. 

Dit d'una altra manera, el constructor de `Promise` exigeix que li passem un únic paràmetre. Aquest paràmetre és de tipus funció i ens hi solem referir com a executor. Aquest executor, que és una funció, rep dos paràmetres, `resolve` i `reject`, i s'encarrega de portar a terme qualsevol tipus d'operació que haguem d'escriure. Escriure el codi d'aquesta funció executora és la nostra responsabilitat. I dins d'aquesta funció haurem de decidir en quins escenaris volem considerar que la promesa queda complerta i en quins casos s'ha de considerar refusada. Aquesta qüestió és totalment arbitrària i depèn de la nostra voluntat com a programadors. Però, un cop decidit en quins casos es considerarà complerta o refusada, hem d'invocar les funcions `resolve()` i `reject()` rebudes com a paràmetres. Si ho fem així, el codi que posteriorment utilitzi aquesta promesa podrà detectar si s'ha complert o refusat i actuar en conseqüència, mitjançant els mètodes `then()` i `catch()` respectivament.

A més de `then()` i `catch()`, les promeses en JavaScript també proporcionen el mètode `finally()`, que s'executa després que les promeses es resolguin o refusin. Aquest mètode és especialment útil per dur a terme operacions de neteja o finalització, independentment del resultat de la promesa. Utilitzar `finally()` assegura que certes accions es duran a terme després de completar les operacions asíncrones, fet que ajuda a mantenir el codi net i clar.
