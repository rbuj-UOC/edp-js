## Unitat 4: Asincronisme i `await`

En l'últim tema del nivell B1 vàrem poder veure una introducció a l'asincronisme. A tall de repàs ràpid, recordem alguns conceptes importants que vàrem mencionar en aquest tema:

_El fet que aquest mecanisme sigui asíncron significa que quan l'aplicació web envia una petició al servidor no es queda esperant fins a rebre una resposta, sinó que segueix executant les línies de codi posteriors, si n'hi ha._

_Això suposa un avantatge clar: l'aplicació web no quedarà bloquejada mentre espera rebre una respuesta que pot trigar un temps indeterminat, potser llarg, sinó que l'usuari en tot moment podrà seguir interactuant amb la pàgina mentre, en segon pla, l'aplicació segueix esperant que la resposta arribi. [...] Si l'aplicació es quedés bloquejada fins a rebre la resposta, l'usuari podria pensar que l'aplicació no funciona correctament i la podria tancar o abandonar. Si, en canvi, l'aplicació segueix funcionant mentre espera la resposta, l'usuari podrà seguir interactuant-hi i, quan la resposta arribi, l'aplicació podrà mostrar a l'usuari la informació que li ha sol·licitat._

_Però aquest comportament també té un inconvenient important: quan escrivim el nostre codi hem de tenir present que les línies posteriors a l'enviament de la petició no poden utilitzar la resposta que s'espera rebre [...] perquè quan s'executin aquestes línies posteriors encara no s'haurà rebut la resposta amb les dades per part del servidor. El que sí que es rep de manera immediata és una promesa que més endavant es rebrà una resposta real. Aquesta promesa consisteix en un objecte de tipus `Promise`._ 
 
_Solució mitjançant `async` i `await`: per solucionar [aquest inconvenient] podem demanar al nostre codi que cada vegada que es trobi amb un mètode que retorna una promesa (objecte `Promise`), renunciï a aquesta promesa i sol·liciti esperar fins que hi hagi una resposta (objecte `Response`) disponible, sense importar el temps que trigui._ 

Exposat de manera una mica més formal, l'asincronisme en programació es refereix a la capacitat d'executar processos o tasques de forma que no bloquegin l'execució seqüencial del programa. En un entorn asíncron, certes operacions poden iniciar i després continuar l'execució en segon pla, permetent que el programa principal segueixi funcionant sense esperar que aquestes operacions finalitzin. Això és particularment útil en tasques que impliquen esperes, com les peticions a un servidor web o la feina amb fitxers desats en un disc dur. L'asincronisme permet una eficiència més gran i una millor gestió de recursos, especialment en aplicacions web on l'experiència de l'usuari depèn de respostes ràpides i no bloquejadores.

En JavaScript les promeses són un mecanisme que facilita la gestió de les operacions asíncrones, permetent manejar de forma més eficient les tasques que es poden no completar de manera immediata. Una promesa representa un valor que pot estar disponible ara, en el futur, o mai. Podem gestionar aquestes promeses mitjançant l'ús dels mètodes `then()` i `catch()`. Vàrem tenir ocasió de veure'n algun exemple en el nivell B1 i trobarem més exemples més endavant en aquest nivell B2, quan treballem les diferents peticions que podem efectuar en una API. Tanmateix, l'ús del mecanisme `async` / `await` és tan comú en JavaScript i, en general, en qualsevol llenguatge de programació modern, que s'ha convertit en un estàndard de facto. No en va, resulta una de les formes més senzilles de treballar amb codi asíncron en JavaScript. En aquesta unitat aprofundirem en aquest mecanisme i en les implicacions que té marcar una funció com a asíncrona.

Quan en el nostre codi trobem un mètode que ha de portar a terme alguna operació que requereix temps per completar-se, com les sol·licituds de xarxa o lectures de fitxers esmentades més amunt, aquest mètode sol retornar una promesa, amb l'objectiu de no bloquejar el flux principal del programa i permetre que se segueixin executant les línies següents.

Davant d'una promesa podem utilitzar `await` per indicar a l'intèrpret de JavaScript que esperi que la promesa es resolgui i que, només a partir d'aquest moment, continuï executant el codi. Funciona, per tant, de manera diferent al mecanisme `then()` esmentat anteriorment. Fins i tot amb el risc de repetir conceptes ja introduïts, convé distingir molt bé entre l'ús de `await` i l'ús de `then()`: 

- Amb `then()` indiquem, mitjançant una funció callback, què s'ha de dur a terme quan la promesa es resolgui. Però el nostre codi es continua executant de manera immediata a la instrucció següent, sense esperar aquesta resolució. 

- Amb `await`, en canvi, indiquem que volem esperar fins que la promesa es resolgui, sense que s'executi mentrestant cap altra línia de codi. Marcar amb `await`, tanmateix, no fa que la promesa es resolgui abans. El temps que trigui dependrà d'un servidor web extern, de la mida d'un fitxer que s'estigui descarregant, etc. Marcar amb `await` no té cap implicació en el temps que trigui en resoldre's la promesa. Simplement serveix per indicar el desig d'esperar que la promesa es resolgui abans de continuar executant el codi. 

Si dins d'una funció pròpia optem per utilitzar `await` per esperar que es resolgui una promesa, el codi que cridi aquesta funció es podria trobar amb la desagradable sorpresa d'observar que la funció es demora un temps indeterminat, potencialment llarg, en retornar un resultat. Per evitar aquest efecte sorpresa, és requisit imprescindible marcar la funció que conté el `await` com `async`. D'aquesta forma, el codi que utilitzi aquesta funció sabrà que la funció és asíncrona i retorna una promesa, i la podrà gestionar adequadament al seu torn, ja sigui mitjançant `await` o mitjançant `then()`.

> <br>🚨&nbsp; **Atenció**
> 
> Recordeu aquest punt perquè és realment important: **utilitzar `await` en una funció exigeig marcar-la com a asíncrona mitjançant la paraula clau `async`.** Amb això informem els que utilitzin aquesta funció que és asíncrona i retorna una promesa. D'aquesta forma, podran gestionar la promesa mitjançant `await` o mitjançant `then()`.

En primer lloc creem una funció `getPokemonsPromise()` que es connecta a aquesta API per obtenir una llista de berries. Per a això, utilitza el mètode `fetch()`. Atès que d'una operació que requereix temps (indeterminat, depèn del servidor on es troba l'API), `fetch()` opta per retornar una promesa. Per tant, la nostra funció `getPokemonsPromise()` retorna també una promesa, en tant que està retornant directament el que obté de `fetch()`.

```javascript
// Example 1.

// This is a syncronous function.
function getPokemonsPromise() {
  // This code returns a Promise.
  const response = fetch('https://pokeapi.co/api/v2/berry');
  return response;
}

// We call a syncronous function and get exactly 
// what the function returns: a Promise.
result = getPokemonsPromise();
console.log(result);
```

Si no volem que la nostra funció retorni directament una promesa, podem marcar la crida a `fetch()` amb `await`. D'aquesta forma, la línia `return` de la nostra funció `getPokemonsResponse()` no retorna una promesa sinó una resposta real. No obstant això, com ja hem comentat, utilitzar `await` dins del cos d'una funció exigeix marcar la funció com a asíncrona mitjançant la paraula clau `async`. Això automàticament fa que la nostra funció `getPokemonsResponse()` retorni també una promesa quan se la invoqui. 

Aquest matís és molt important. El codi interior de la funció retorna una resposta, no una promesa. Però la funció en si mateixa retorna una promesa, perquè l'hem marcat com a asíncrona. Per tant, qui invoqui aquesta funció haurà de gestionar la promesa, bé mitjançant `await` o mitjançant `then()`.

```javascript
// Example 2.

// This is an asyncronous function.
async function getPokemonsResponse() {
  // This code DOES NOT return a Promise but a real Response.
  const response = await fetch('https://pokeapi.co/api/v2/berry');
  return response;
}

// We call an asyncronous function.
// Even when the function returns a Response,
// as the function itself is asyncronous, we get a Promise.
result = getPokemonsResponse();
console.log(result);
```

Finalment, si quan invoquem la funció `getPokemonsResponse()` utilitzem `await`, el que obtindrem serà una resposta real en comptes d'una promesa. Això, com ja hem comentat, no significa en cap moment que la resposta s'obtingui abans. Simplement, el codi que invoqui la funció `getPokemonsResponse()` esperarà que la resposta estigui disponible abans de continuar-se executant.

```javascript
// Example 3.

// We call an asyncronous function.
// This time, we use `await` as we do not want a Promise.
// As the function returns a Response and we use `await`,
// we get a real Response. But it takes a while.
result = await getPokemonsResponse();
console.log(result);
```

> <br>🚨&nbsp; **Atenció**
> 
> Quan en el codi d'un fitxer JavaScript enllaçat amb una pàgina web utilitzem `await` en el nivell més alt del nostre codi, és a dir, fora de qualsevol funció, obtenim un error que ens indica que _await is only valid in async functions and the top level bodies of modules_. Per tant, per poder utilitzar `await` fora d'una funció marcada amb `async`, ho hem de fer dins d'un mòdul. Per a això, el fitxer JavaScript s'ha d'enllaçar afegint la notació `type="module"` a l'etiqueta `<script>` corresponent del fitxer HTML.
