## Unitat 6: Mòduls

Aquesta unitat s'enfoca en l'ús i maneig de mòduls en JavaScript, un concepte clau per organitzar i estructurar el codi en projectes més grans i complexos. Els mòduls permeten als desenvolupadors dividir el seu codi en unitats més petites i reutilitzables, fet que facilita l'organització i el manteniment del codi. 

### Antecedents i necessitat de mòduls en JavaScript

Al principi, JavaScript s'usava principalment per programar petits scripts, destinats a incrementar la interactivitat a les pàgines web, per la qual cosa no era comú treballar amb extenses bases de codi. Amb el temps, tanmateix, JavaScript va anar guanyant terreny en popularitat i versatilitat, evolucionant fins a convertir-se en el nucli d'aplicacions completes corrent tant en el marc d'un navegador web com fora d'aquest, a través per exemple de l'entorn Node.js. Aquest auge va impulsar la necessitat de millors pràctiques en l'estructuració del codi, per evitar així col·lisions de noms i millorar l'organització i el rendiment dels nostres programes.

El concepte de mòduls en JavaScript va sorgir, doncs, com una resposta a la necessitat d'estructurar i organitzar millor el codi en aplicacions grans i complexes. Els mòduls permeten als desenvolupadors dividir el seu codi en unitats més petites i reutilitzables. Cada mòdul encapsula el seu propi conjunt de variables i funcions, evitant així les col·lisions de noms i facilitant la reutilització del codi. Els mòduls poden importar funcionalitats d'altres mòduls i exportar les seves pròpies funcionalitats, fet que permet crear aplicacions grans i complexes d'una manera més estructurada i mantenible.

Amb l'adopció d'ES6 (ECMAScript 2015), els mòduls es van convertir en una part integral del llenguatge JavaScript. ES6 va introduir una sintaxi estàndard per a la importació i exportació de mòduls, millorant significativament la capacitat dels programadors per organitzar i mantenir el nostre codi. Això ha facilitat i impulsat el desenvolupament d'aplicacions JavaScript a gran escala, al mateix temps que en millora la llegibilitat i el manteniment, i permet afrontar amb garanties qüestions tan importants com l'escalabilitat del codi. 

Els mòduls, a més, ajuden a millorar el rendiment de les aplicacions en permetre la càrrega diferida i l'optimització dels scripts. Gràcies a aquest mecanisme podem carregar només els mòduls necessaris per a una pàgina o un component específic, reduint així el temps de càrrega i millorant l'experiència de l'usuari.

Un avantatge clau d'usar mòduls en JavaScript és la promoció de la separació de responsabilitats en el desenvolupament d'aplicacions. Cada mòdul es pot enfocar en una part específica de la funcionalitat de l'aplicació, fet que facilita la feina en equip i la col·laboració, ja que diferents desenvolupadors poden treballar en mòduls separats simultàniament sense interferències. Aquesta organització modular del codi ajuda a prevenir problemes de manteniment i fa que el codi sigui més fàcil d'entendre, provar i depurar.

### Exportació i importació de mòduls

Els mòduls ES6 es defineixen en fitxers JavaScript separats. Cada fitxer és un mòdul i pot exportar les seves pròpies funcionalitats. Per exportar una funcionalitat, hem d'usar la paraula clau `export` seguida d'allò que volem exportar, que pot ser una funció, una classe, un objecte o fins i tot un valor primitiu. Per exemple, el codi següent correspon a un fitxer de nom `math.js` i exporta una funció anomenada `sum()`:

```javascript
// math.js
export function sum(a, b) {
  return a + b;
}
```

Per importar una funcionalitat d'un mòdul, hem d'usar la paraula clau `import` seguida del nom de la funcionalitat que volem importar (entre claus) i el nom del mòdul. Per exemple, el codi següent correspon a un fitxer `main.js` des del qual volem importar la funció `sum()` del mòdul `math.js`:

```javascript
// main.js
import { sum } from './math.js';

// Now we can use `sum()` function in this file as if it was defined in it.
console.log(sum(25, 17));
```

No estem limitats a exportar un únic element, sinó que és possible exportar múltiples entitats des d'un sol mòdul. De la mateixa manera, també podem importar múltiples elements d'un mòdul en una sola declaració. Per exemple, el codi següent exporta dues funcions, `sum()` i `multiply()`, i les importa en un sol comandament. Mostrem el contingut dels dos fitxers implicats perquè pugueu veure com s'estructura el codi:

```javascript
// math.js
export function sum(a, b) {
  return a + b;
}

export function multiply(a, b) {
  return a * b;
}
```

```javascript
// main.js
import { sum, multiply } from './math.js';

console.log(sum(25, 17));
console.log(multiply(2, 21));
```

En JavaScript, els mòduls poden exportar elements de dues maneres principals: mitjançant exportacions nomenades i exportacions per defecte. Ambdós enfocaments tenen propòsits específics i poden ser utilitzats segons les necessitats del mòdul i la claredat del codi.

Les exportacions nomenades, que són les que hem vist en els dos exemples anteriors, permeten exportar múltiples valors (funcions, classes, objectes o primitives) des d'un mateix mòdul. Cada un d'aquests valors pot ser importat pel seu nom específic des d'altres mòduls. Les exportacions nomenades promouen un accés selectiu als components del mòdul, facilitant la importació d'únicament aquells elements que es necessiten. 

En el cas de les exportacions per defecte, en canvi, un mòdul només pot exportar un únic element d'aquesta manera. Aquest tipus d'exportació està pensada per als casos en els quals un mòdul se centra en exportar un únic valor principal, que es considera l'element principal o "per defecte" del mòdul.  

Tanmateix, és important destacar que un mòdul pot contenir, al mateix temps, tant exportacions nomenades com una única exportació per defecte. Això permet que el mòdul ofereixi un element principal de fàcil accés (per defecte), al mateix temps que proporciona elements addicionals que poden ser importats de manera selectiva. 

A l'hora de dur a terme la importació, en el cas de les exportacions nomenades hem d'usar la sintaxi de claus per importar els elements desitjats, indicant el nom de l'element tal com va ser exportat en el mòdul, tot i que tenim la possibilitat de canviar-li el nom en importar-lo mitjançant la sintaxi `importedName as someNewName`. En el cas de les exportacions per defecte no usem claus per importar l'element, ja que només hi ha un element exportat per defecte. A més, precisament perquè l'element exportat per defecte és únic, podem usar qualsevol nom per importar-lo, sense necessitat d'utilitzar la paraula clau `as` per canviar-ne el nom.

Ho veurem més clar mitjançant un exemple. En el codi següent, en el mòdul `math.js` exportem una funció amb nom `random()` com a exportació per defecte (única, no n'hi pot haver una altra marcada amb `default`) i dues funcions anomenades `subtract()` i `divide()` com a exportacions nomenades. Posteriorment, des de `main.js` importem la funció `random()` amb el nom `randomNumber()`, la funció `subtract()` amb el seu nom original i la funció `divide()` amb l'àlies `divideNumbers()`:

```javascript
// math.js
export default function random() {
  return Math.random();
}

export function subtract(a, b) {
  return a - b;
}

export function divide(a, b) {
  return a / b;
}
```

```javascript
// main.js
import randomNumber, { subtract, divide as divideNumbers } from './math.js';

console.log(randomNumber());
console.log(subtract(10, 5));
console.log(divideNumbers(10, 5));
```

> <br>🚨&nbsp; **Atenció**
> 
> Fixeu-vos en els matisos de la sintaxi. En l'exportació, la funció exportada per defecte requereix la paraula clau `default`. Posteriorment, durant la importació no l'heu d'indicar entre claus i li podeu assignar qualsevol nom, que pot coincidir o no amb el nom original. En el cas de les exportacions nomenades, no heu d'usar la paraula clau `default` i sí que les heu d'indicar entre claus durant la importació, usant el nom original.

### Resolució de conflictes de noms i renomenament en importacions i exportacions

Quan treballem amb importacions i exportacions en JavaScript, ens podem trobar amb diferents elements que han sigut exportats amb el mateix nom des de mòduls diferents. Per exemple, el mòdul `numbers.js` pot exportar la funció `random()` que retorna un nombre aleatori entre 0 i 100 quan la invoquem, mentre que el mòdul `letters.js` pot exportar una altra funció, amb el mateix nom `random()`, que retorna una lletra aleatòria de l'alfabet quan la cridem. Si aquestes exportacions es duen a terme amb la modalitat per defecte no tindrem cap problema, perquè en importar-les els podem assignar nom diferents: `randomNumber()` i `randomLetter()`. Però què passa si són exportacions nomenades? En importar-les podem generar un error de nom duplicat, ja que ambdues funcions es diuen `random()`:

```javascript
// main.js
import { random } from './numbers.js';
import { random } from './letters.js';
```

Per sort, podem solucionar aquest problema renomenant el que importem. JavaScript ens permet assignar un nom diferent a les entitats que importem, tot i que s'hagin exportat de manera nomenada. Per a això utilitzem la paraula clau `as`, amb la qual no només evitem confusions, sinó que també ens permet ser més descriptius amb els nom:

```javascript
// main.js
import { random as randomNumber } from './numbers.js';
import { random as randomLetter } from './letters.js';
```

### Integració de mòduls en aplicacions HTML

En l'àmbit del desenvolupament web, la integració de mòduls JavaScript en aplicacions HTML constitueix un enfocament fonamental per estructurar eficientment el codi. No obstant això, per poder utilitzar aquesta característica de JavaScript i tenir la capacitat d'importar i exportar funcionalitats entre diferents fitxers, hem d'indicar explícitament que el fitxer JavaScript es un mòdul. Per a això, a l'etiqueta `<script>` del fitxer HTML hi hem d'afegir l'atribut `type="module"`.

Per exemple, el fitxer següent `index.html` enllaça amb el fitxer de scripts `index.js`. En portar incorporat l'atribut `type="module"`, el navegador interpreta que el fitxer `index.js` és un mòdul i, per tant, pot importar i exportar funcionalitats d'altres mòduls:

```javascript
<!-- index.html -->
<!DOCTYPE html>
<html lang="en">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <script type="module" src="index.js"></script>
    <title>Demo - How to use JavaScript modules in our HTML</title>
</head>
<body>
    <h1>Demo - How to use JavaScript modules in our HTML</h1>
    <p>Open the console to see the result of the sum of two numbers.</p>
</body>

</html>
```

```javascript
// index.js
import { sum } from './math.js';

console.log(sum(18, 24));
```

```javascript
// math.js
export function sum(a, b) {
  return a + b;
}
```

En implementar mòduls mitjançant l'etiqueta `<script type="module">` en els nostres fitxers HTML, establim un marc de treball que afavoreix la modularitat i la claredat en la construcció de l'aplicació.

Aquesta metodologia permet que cada mòdul gestioni les seves pròpies dependències, fomentant així la reutilització de codi i una coherència més gran en el projecte. A més, facilita el manteniment i l'escalabilitat de les aplicacions, que s'alineen amb les pràctiques modernes de desenvolupament web.

Per tant, la integració adequada de mòduls en HTML no és només una qüestió d'organització del codi, sinó que representa una estratègia essencial per al desenvolupament d'aplicacions web robustes i mantenibles.
