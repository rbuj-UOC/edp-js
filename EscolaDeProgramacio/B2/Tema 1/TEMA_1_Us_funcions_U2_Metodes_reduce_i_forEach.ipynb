## Unitat 2: Mètodes `reduce` i `forEach`

En el primer tema del nivell B1 vàrem veure que el correcte maneig d'arrays resulta una habilitat imprescindible per a qualsevol programador. I vàrem tenir l'ocasió d'aprendre sobre dos dels mètodes més utilitzats per treballar amb arrays en JavaScript: `map()` i `filter()`. En concret, `map()` ens permet crear un nou array a partir de la conversió dels elements d'un array original en nous elements diferents, mentre que `filter()` obre les portes al filtratge o selecció d'elements.

En aquesta unitat, explorarem uns altres dos mètodes molt versàtils per treballar amb arrays, `reduce()` i `forEach()`, que no només són fonamentals per entendre la manipulació d'arrays, sinó que també ens permeten escriure codi més expressiu, funcional i eficient.

### Acumulació o reducció d'arrays amb `reduce`

En el món de la programació amb JavaScript, sovint ens trobem amb la necessitat de transformar un array de dades en un únic valor acumulatiu. Això pot passar, per exemple, quan volem sumar tots els elements d'un array de nombres, compactar una sèrie de cadenes en una de sola, o fins i tot calcular un valor estadístic com una mitjana. Aquí és on `reduce()` es converteix en una eina invalorable.

Aquest mètode treballa sobre un array i aplica una funció reductora a cada un dels seus elements, acumulant el resultat en un valor únic. El procés comença amb un valor inicial, que pot ser qualsevol valor que definim, i després procedeix a iterar a través de l'array. En cada iteració, la funció reductora agafa dos arguments: el valor acumulat de les iteraccions anteriors i el valor actual de l'array. Al final del procés, `reduce()` retorna aquest valor acumulat.

Un aspecte fascinant de `reduce()` és la versatilitat. Es pot utilitzar per a operacions simples com les sumes o mitjanes ja esmentades, però també per a tasques més complexes com convertir un array en un objecte o fins i tot en un nou array amb una estructura completament diferent. És aquesta flexibilitat la que fa de `reduce()` una eina essencial per a qualsevol desenvolupador de JavaScript, permetent abordar una àmplia gamma de problemes de processament i manipulació de dades de manera elegant i eficient.

Vegem un primer exemple senzill de `reduce()` en acció. Suposem que tenim un array de nombres i volem calcular la suma de tots els elements. Ho podríem fer de la manera següent:

```javascript
const numbers = [1, 2, 3, 4, 5];

// Use `reduce()` to sum the numbers in the array.
const sum = numbers.reduce((accumulator, currentValue) => {
  return accumulator + currentValue;
}, 0);

console.log(sum);
```

El valor inicial de l'acumulador és 0, així que li ho indiquem com a segon paràmetre. En el primer paràmetre li indiquem la funció reductora, que en aquest exemple consisteix a sumar el valor actual de l'array `currentValue` al valor acumulat `accumulator`. Al final del procés, el valor acumulat és 15, que és el resultat que buscàvem.

Ara bé, què passa si volem calcular la mitjana dels elements de l'array? En aquest cas, podem optar per una solució en la qual el valor inicial de l'acumulador consisteixi en un objecte amb dues propietats: `sum` i `count`. La primera propietat emmagatzemarà la suma de tots els elements de l'array, mentre que la segona emmagatzemarà la quantitat d'elements. La funció reductora, per part seva, ha de sumar el valor actual de l'array a la propietat `sum` de l'acumulador, i augmentar en 1 la propietat `count`. Al final del procés, el valor acumulat serà un objecte amb les propietats `sum` i `count`, i la mitjana s'obtindrà dividint la primera entre la segona.

```javascript
const salaries = [75000, 55000, 88000, 125000, 64000];

// Use `reduce()` to calculate the average salary.
const averageSalary = salaries.reduce(
  (accumulator, currentValue) => {
    accumulator.sum += currentValue;
    accumulator.count++;
    return accumulator;
  },
  { sum: 0, count: 0 }
);

console.log(averageSalary.sum / averageSalary.count);
```

En aquest cas, la funció `reduce()` és la que rep com a argument la funció reductora. A la unitat anterior hem estudiat les funcions callback, així que us recomanem que la repasseu si teniu dubtes.

### Iterant arrays amb `forEach`: eficiència i claredat

En nivells anteriors hem vist que la iteració sobre els elements d'un array és una tasca comuna i crucial en qualsevol llenguatge de programació. Tot i que en JavaScript tenim diferents maneres de dur a terme aquest procés, el mètode `forEach()` brilla amb llum pròpia, oferint una forma clara i eficient de recórrer cada element d'un array. Aquest mètode forma part essencial del conjunt d'eines de qualsevol desenvolupador, ja que proporciona una solució elegant i funcional en escenaris que requereixen processar cada element d'un array.

Mitjançant `forEach()` executem una funció específica per a cada un dels elements d'un array. A diferència dels bucles tradicionals com `for` o `while`, el mètode `forEach()` encapsula la lògica de la iteració, el que dona com a resultat un codi més net i fàcil d'entendre. Cada vegada que `forEach()` processa un element de l'array, invoca la funció proporcionada, i passa l'element actual com a argument. Això significa que podem dur a terme qualsevol operació que necessitem sobre cada element.

> <br>🚨&nbsp; **Atenció**
> 
> Aquesta funció proporcionada és, novament, una funció callback. Els mètodes `map()`, `filter()`, `reduce()` i `forEach()` utilitzen funcions callback. En general, els mètodes de programació funcional de JavaScript les solen utilitzar. Per això és important que us familiaritzeu amb aquest concepte i amb la sintaxi de les arrow functions, que és la que s'utilitza de manera més habitual. 
> <br><br>

El que fa que `forEach` sigui particularment poderós és la capacitat de manejar no només l'element actual, sinó també el seu índex i l'array complet. Aquesta flexibilitat ens permet escriure codi que es no només concís sinó també expressiu. Per exemple, es pot usar `forEach()` per registrar cada element juntament amb la seva posició a l'array, o per dur a terme comprovacions i actualitzacions condicionals en els elements.

Vegem-ho amb un exemple. Suposem que tenim un array de pokémons i volem imprimir a la consola el nom de cada un, juntament amb la seva posició a l'array. Ho podríem fer de la manera següent:

```javascript
const pokemons = [
  { name: "Bulbasaur", type: "grass" },
  { name: "Charmander", type: "fire" },
  { name: "Squirtle", type: "water" },
];

// Use `forEach()` to print each pokemon name and position.
pokemons.forEach((pokemon, index) => {
  console.log(`#${index + 1}: ${pokemon.name}`);
});
```

Com en el cas dels mètodes esmentats anteriors, `forEach()` és un clar exemple de la filosofia de programació funcional en JavaScript. En separar la lògica d'iteració del cos de la funció, fomenta un estil de codificació més declaratiu i modular. Això millora la llegibilitat del codi i facilita la depuració i el manteniment a llarg termini.

> <br>💫&nbsp; **Ampliació**
> 
> Tot i que `reduce()` i `forEach()` s'utilitzen per a propòsits diferents, la combinació pot ser molt potent. Per exemple, es pot utilitzar `forEach()` per aplicar una operació a cada element i després `reduce()` per combinar els resultats en un valor final.

### Immutabilitat dels arrays

En JavaScript, els arrays són estructures de dades flexibles i dinàmiques que es comporten com objectes mutables. Això implica que el seu contingut pot ser alterat mitjançant l'addició, eliminació o modificació d'elements. La mutabilitat dels arrays pot ser molt útil, però també presenta desafiaments en certs contextos de programació.

La programació funcional, un paradigma que enfatitza la immutabilitat i les operacions sense efectes secundaris, afavoreix l'ús de dades que no canvien un cop creades. Els objectes immutables són més predictibles i menys propensos a errors, la qual cosa és especialment important en entorns de programació concurrent o multifil, on múltiples processos poden accedir i modificar dades al mateix temps.

Respecte als mètodes d'array com `map()`, `filter()`, `reduce()` i `forEach()` resulta clau entendre que per si mateixos no modifiquen l'array original sobre el qual s'invoquen. En concret, `map()` i `filter()` retornen un nou array, mentre que `reduce()` s'utilitza per obtenir un valor acumulat sense alterar l'array original. Al seu torn, `forEach()` s'utilitza principalment per generar efectes secundaris, també sense alterar l'array de partida.

Tanmateix, la funció callback proporcionada a aquests mètodes sí que té el potencial de modificar l'array original. Per exemple, si s'usa `forEach()` per iterar sobre un array, la funció callback pot canviar-ne els elements, alterant així l'array original. Això pot portar a comportaments inesperats si no es maneja amb compte.

```javascript
const oddNumbers = [1, 3, 5, 7, 9];

// Use `forEach()` to double all numbers in the array.
oddNumbers.forEach((number, index, array) => {
  array[index] = number * 2;
});

console.log(oddNumbers);
```

Per tant, tot i que els mètodes en si mateixos siguin funcionals i no modifiquin l'array directament, la immutabilitat de les dades depèn de l'ús que es faci de les funcions callback. És important adoptar pràctiques que respectin la immutabilitat, especialment en contextos on la predictibilitat i la seguretat en la concurrència són crítiques.
