# TEMA 1: Ús avançat de funcions

## Unitat 1: Funcions de retorn de crida (callbacks)

En el nivell B1 vam tenir la necessitat d'utilitzar en un parell d'ocasions les anomenades funcions de retorn de crida (callbacks). Però no vàrem arribar a desenvolupar-ne el concepte en profunditat. En aquest nivell B2 n'estudiarem l'ús i la utilitat, en tant que resulten un concepte clau en JavaScript, per exemple a l'hora de treballar amb operacions asíncrones i per al maneig d'esdeveniments.

Una funció callback en JavaScript és una funció que es passa a una altra funció com a argument perquè aquesta segona funció (la que rep la funció callback) l'executi quan calgui, per exemple després que es completi alguna operació. Les funcions callback són fonamentals en la feina amb operacions asíncrones, com veurem en una unitat posterior, ja que permeten que el codi es continuï executant mentre s'espera algun tipus de resposta.

> <br>🧐&nbsp; **Analogia**
>
> Imagineu-vos que esteu organitzant un banquet nupcial i contracteu una orquestra per amenitzar el sopar. Lliureu al director de l'orquestra un sobre que indica quines cançons han de tocar quan arribi el moment de començar el ball. El director rep aquest sobre, però no l'obre fins que arriba el moment indicat. En aquest cas, el sobre seria la funció callback, i el director de l'orquestra seria la funció que rep la funció callback. Fixeu-vos que quan el director rep el sobre, no l'obre i es posa a tocar les cançons esmentades en aquell mateix instant, sinó que el guarda fins que arriba el moment d'obrir-lo. De la mateixa manera, la funció que rep la funció callback no l'executa immediatament, sinó que la guarda fins que més endavant arribi el moment d'executar-la.
> <br><br>

Dit d'una altra manera, una funció callback consisteix en un mecanisme mitjançant el qual passem una sèrie d'instruccions a una altra funció perquè les executi quan calgui. És molt important tenir en compte que la funció callback es pot executar immediatament o es pot executar més tard... o fins i tot es podria no arribar a executar mai. És la funció que rep la funció callback la que decideix quan executar-la.

Vegem-ho amb l'exemple següent. Imaginem-nos una funció `processTask()` que ha de portar a terme una operació complexa que algunes vegades acaba bé i d'altres no. Volem passar unes instruccions addicionals a `processTask()` perquè sàpiga què ha de fer quan la tasca acaba bé i quines accions ha d'executar en cas contrari. Per a això, definim dues funcions `onTaskCompleted()` i `onTaskFailed()`. La primera s'executarà quan la tasca complexa acabi bé. La segona, quan acabi malament. La funció `processTask()` s'encarregarà d'executar una de les dues funcions callback depenent del resultat de l'operació.

```javascript
// Function that performs a task and then executes a callback function.
function processTask(successCallback, errorCallback) {
  console.log("Performing difficult task...");

  // Simulating a condition to determine success or failure.
  const taskSuccess = Math.random() > 0.5;
  if (taskSuccess) {
    successCallback();
  } else {
    errorCallback();
  }
}

// First possible callback function - this one is executed if the task is successful.
function onTaskCompleted() {
    console.log("Task successfully completed.");
    // Additional operation - update UI (pseudocode).
    updateUI();
}

// Second possible callback function - this one is executed if the task fails.
function onTaskFailed() {
    console.log("Task failed.");
    // Additional operation - log error (pseudocode).
    logError();
}

// Executing the function `processTask()` with both callbacks.
// Output: "Performing difficult task..." followed by "Task successfully completed." (if task is successful).
// Output: "Performing difficult task..." followed by "Task failed." (if task fails).
processTask(onTaskCompleted, onTaskFailed); 
```

> <br>🚨&nbsp; **Atenció**
>
> Observeu que quan passem una funció com a argument a una altra funció no incloem els parèntesis. Si els incloguéssim, estaríem executant la funció en aquest moment i no estaríem passant la funció com a argument, sinó que estaríem passant el resultat de l'execució de la funció (o `undefined` si la funció no retorna res). Aquest és un error habitual que hem d'evitar. A l'exemple anterior, a l'última línia de codi escrivim només els noms de les funcions `onTaskCompleted` i `onTaskFailed`, sense incloure els parèntesis. Si els incloguéssim, estaríem passant el resultat de l'execució de les funcions, en comptes de les funcions mateixes.
>
> Recordeu sempre aquest detall: **quan passem una funció com a argument a una altra funció no incloem els parèntesis**.
> <br><br>

### Funcions d'ordre superior i funcions de primer ordre

L'ús de funcions callback ens permet treballar amb les anomenades funcions d'ordre superior. Una funció d'ordre superior és una funció que rep com a argument una altra funció. A l'exemple anterior, la funció `processTask()` és una funció d'ordre superior, ja que rep com a argument una funció (en realitat, dues). En canvi, les funcions `onTaskCompleted()` i `onTaskFailed()` són funcions de primer ordre, ja que no reben com a argument cap funció.

> <br>🧐&nbsp; **Petit matís**
>En realitat una funció és d'ordre superior no només si rep com a argument una altra funció, sinó també si retorna una funció. De tota manera, aquest segon cas és menys habitual i no l'estudiarem en aquest nivell B2. En la gran majoria dels casos, quan treballeu amb funcions d'ordre superior us estareu referint a funcions que reben com a argument una altra funció.
> <br><br>

Sovint trobem aquests dos conceptes (funcions d'ordre superior i funcions de primer ordre) en la literatura en anglès amb els termes _higher-order functions_ i _first-order functions_, o fins i tot amb les sigles HOF i FOF.

En el nivell B1 ja vam tenir ocasió de treballar amb algunes funcions d'ordre superior, com `map()`, `filter()` o `addEventListener()`. A la unitat següent veurem també que les funcions `reduce()` i `forEach()` són també funcions d'ordre superior. En realitat JavaScript disposa de moltes funcions d'ordre superior que ens permeten treballar de forma molt còmoda i eficient, per exemple en el cas dels arrays.

Si analitzem el cas de `map()`, veiem que aquesta funció rep com a argument una altra funció, que és la que s'encarrega de transformar cada un dels elements de l'array. En el cas de `filter()`, la funció que rep com a argument és la que s'encarrega de decidir si cada un dels elements de l'array ha de ser inclòs en el nou array resultant o no. En el cas de `addEventListener()`, la funció que rep com a argument és la que s'encarrega de manejar l'esdeveniment que s'ha produït (per exemple un clic d'un usuari en un botó d'una pàgina web). En tots aquests casos, la funció que es rep com a argument és una funció callback.

Vegem-ho amb un exemple nou. Crearem una funció `calculate()` que rep dos arguments: un array amb dos nombres i una funció que s'encarrega de dur a terme una operació matemàtica. La funció `calculate()`, per tant, és una funció d'ordre superior, ja que rep com a argument una altra funció. La resta de funcions, que s'utilitzaran com a funcions callback, són funcions de primer ordre, ja que no reben com a argument cap funció i es limiten a operar amb dos nombres. Podem observar que la funció `calculate()` s'encarrega d'executar la funció callback amb els dos nombres que rep com a argument, però no sap quina operació matemàtica es durà a terme. És la funció callback la que du a terme l'operació matemàtica.

```javascript
// Higher-order function.
function calculate(numbers, operation) {
  return operation(numbers[0], numbers[1]);
}

// First-order function.
function add(a, b) {
  return a + b;
}

// First-order function.
function subtract(a, b) {
  return a - b;
}

// First-order function.
function multiply(a, b) {
  return a * b;
}

// First-order function.
function divide(a, b) {
  return a / b;
}

console.log(calculate([2, 3], add));      // 5
console.log(calculate([2, 3], subtract)); // -1
console.log(calculate([2, 3], multiply)); // 6
console.log(calculate([2, 3], divide));   // 0.666...
```

### Funcions callback mitjançant arrow functions

### Callback hell

Quan treballem amb funcions callback es pot donar l'escenari en què la funció que passem com a argument a una altra funció també rebi, al seu torn, una funció com a argument, i aquesta última funció al seu torn rebi una altra funció, i així successivament. És a dir, podem tenir una funció `hof1()` que rep com a argument una funció `hof2()`, que al seu torn rep com a argument una funció `hof3()`, que al seu torn rep com a argument una funció `hof4()`, i així successivament. En aquest cas, diem que tenim un _callback hell_, o un infern de funcions callback. Això pot passar perquè, tot i que en els exemples vistos fins ara totes les funcions callback eren funcions simples de primer ordre, en realitat no hi ha cap restricció que impedeixi que una funció callback sigui una funció d'ordre superior. 

El terme _callback hell_ es refereix a un problema comú en JavaScript i altres llenguatges de programació on s'utilitzen múltiples funcions de retorn de crida (callbacks) imbricades. Això sol passar de manera molt habitual en operacions que requereixen diverses tasques asíncrones, com les peticions de dades a un servidor web, on cada tasca depèn del resultat de l'anterior.

```javascript
// Simulating an asynchronous task that takes 1 second to complete.
function asyncTask(callback) {
  setTimeout(() => {
      console.log("Task completed");
      callback();
  }, 1000);
}

// Example of callback hell.
asyncTask(() => {
  console.log("First level");
  asyncTask(() => {
      console.log("Second level");
      asyncTask(() => {
          console.log("Third level");
          // We could continue nesting more calls here...
      });
  });
});
```

Visualment, aquest _callback hell_ es caracteritza per tenir diversos nivells de funcions imbricades, fet que porta a un codi amb una estructura en forma de piràmide o triangle cap a la dreta. Això dificulta la lectura i el manteniment del codi, a més de conduir a problemes de gestió i depuració d'errors.

Per evitar aquest problema s'han desenvolupat diverses solucions. Una de les més comunes és l'ús de promeses (en JavaScript, objectes de tipus `Promise`) que permeten escriure codi asíncron d'una manera més plana i llegible. Les promeses proporcionen una forma d'organitzar les tasques asíncrones en una cadena, en comptes d'imbricar-les. Una altra solució més recent és l'ús del mecanisme format per `async` i `await`, que simplifiquen encara més la feina amb operacions asíncrones, fent que el codi s'assembli més a una estructura síncrona tradicional.

En unitats posteriors estudiarem amb més detall aquests dos mecanismes, que resulten fonamentals en la feina amb operacions asíncrones.
