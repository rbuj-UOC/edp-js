## Unitat 3: Peticions PUT, DELETE i PATCH amb Fetch

A la unitat anterior hem esmentat que:

- Les peticions `GET` s'utilitzen principalment per sol·licitar dades d'un recurs específic. 
- Les peticions `POST` s'usen per enviar dades a un servidor i crear un recurs.

Aquests dos tipus de peticions són les més habituals en qualsevol aplicació que treballi amb dades: crear nova informació i llegir informació ja existent. Tanmateix, hi ha altres tipus d'operacions que també són d'ús habitual: actualitzar registres i eliminar registres. Per portar a terme aquestes operacions en el marc del treball amb una API REST, s'utilitzen les peticions `PUT` i `PATCH` per actualitzar i `DELETE`, és clar, per eliminar.

> <br>💫&nbsp; **Ampliació**
>
> Sovint trobareu que s'utilitza l'acrònim **CRUD** per fer referència a les operacions bàsiques que es poden dur a terme sobre les dades d'una aplicació. Aquest acrònim recull els quatre tipus d'operacions esmentades: Create, Read, Update i Delete. Quan treballem amb els verbs HTTP, `POST` s'utilitza per a Create, `GET` per a Read, `PUT` i `PATCH` per a Update i `DELETE` per a Delete. Així, si un desenvolupador comenta que una determinada aplicació no es més que un CRUD, el que realment vol dir és que aquesta aplicació es limita a crear, llegir, actualitzar i eliminar dades, bé directament en una base de dades o bé través d'una API.
> <br><br>

### Peticions PUT

Les peticions `PUT` s'utilitzen per actualitzar un recurs ja existent en el servidor. Tal com passa amb les peticions `POST`, les peticions `PUT` també inclouen un cos amb les dades que es volen enviar al servidor, però les intencions i els efectes a l'estat del recurs en el servidor són diferents.

En una petició `PUT` s'espera que el cos de la sol·licitud inclogui tots els camps del recurs que s'estan actualitzant, no només els que canvien. Això és degut al fet que `PUT` reemplaça tot l'estat del recurs existent amb el nou estat proporcionat. 

> <br>🚨&nbsp; **Atenció**
>
> Si executeu una petició `POST` repetides vegades, es crearan tants recursos com peticions s'hagin executat. En canvi, si executeu una petició `PUT` repetides vegades, el recurs s'actualizarà tantes vegades com peticions envieu, però no es crearan recursos duplicats. Això és el que es coneix com a **idempotència**: dur a terme la mateixa petició diverses vegades no altera el resultat final aconseguit després de la primera petició.
> <br><br>

Vegem el funcionament d'una petició `PUT` amb un exemple. Tal com vàrem comentar a la unitat anterior, utilitzarem l'API de [JSONPlaceholder](https://jsonplaceholder.typicode.com), que ens permet fer peticions no només `GET`, sinó també `POST`, `PUT`, `PATCH` i `DELETE` sobre un conjunt de dades de prova. En concret, utilitzarem la ruta `/posts/12` per actualitzar el recurs amb l'identificador 12.

```javascript
// We first read post with id 12 using GET method.
fetch('https://jsonplaceholder.typicode.com/posts/12')
  .then((response) => response.json())
  .then((data) => console.log(data))
  .catch((error) => console.error('Error:', error));
```

```javascript
// Then we update this post using PUT method.
const updatedPost = {
  userId: 2,
  id: 12,
  title: 'A NEW UPPERCASE TITLE HERE',
  body: 'itaque id aut magnam\n' +
    'praesentium quia et ea odit et ea voluptas et\n' +
    'sapiente quia nihil amet occaecati quia id voluptatem\n' +
    'incidunt ea est distinctio odio'
};

const headers = { 'Content-Type': 'application/json' };

fetch('https://jsonplaceholder.typicode.com/posts/12', {
    method: 'PUT',
    headers: headers,
    body: JSON.stringify(updatedPost)
})
  .then((response) => response.json())
  .then((data) => console.log(data))
  .catch((error) => console.error('Error:', error));
```

Observem que les respostes que obtenim després de les peticions `GET` i `PUT` inclouen el mateix recurs, el que té l'identificador 12, però en el segon cas el títol del recurs s'ha actualitzat. La resta de camps (l'identificador de l'usuari i el cos de l'article) s'han mantingut intactes.

Què passaria si a la petició `PUT` no hi incloguéssim el camp `title`? Doncs que el recurs s'actualitzaria, però el títol s'eliminaria. Això passa perquè, com hem comentat, en una petició `PUT` s'espera que el cos de la sol·licitud inclogui tots els camps del recurs, tant si canvien com si no.

```javascript
// PUT request with incomplete data.
const incompletePost = {
  userId: 2,
  id: 12,
  body: 'itaque id aut magnam\n' +
    'praesentium quia et ea odit et ea voluptas et\n' +
    'sapiente quia nihil amet occaecati quia id voluptatem\n' +
    'incidunt ea est distinctio odio'
};

fetch('https://jsonplaceholder.typicode.com/posts/12', {
    method: 'PUT',
    headers: headers,
    body: JSON.stringify(incompletePost)
})
  .then((response) => response.json())
  .then((data) => console.log(data))
  .catch((error) => console.error('Error:', error));
```

> <br>🚨&nbsp; **Atenció**
>
> Si executeu novament una petició `GET` per llegir el mateix recurs (l'article amb identificador 12) veureu que en el servidor les dades no han canviat. Això passa perquè aquesta API no emmagatzema les dades que se li envien, sinó que les descarta. En una API real, les dades s'emmagatzemen en una base de dades i, per tant, es mantenen actualitzades, de tal manera que si executeu una petició `GET` abans i després d'una petició `PUT` veureu que les dades s'han actualitzat.
> <br><br>

### Peticions PATCH

Les peticions `PATCH` s'utilitzen també per actualitzar un recurs ja existent en el servidor. I tal com passa amb les peticions `POST` i les peticions `PUT`, les peticions `PATCH` també inclouen un cos amb les dades que es volen enviar al servidor. Tanmateix, en aquest cas s'espera que aquest cos inclogui únicament els camps del recurs que canvien perquè, en realitat, `PATCH` s'utilitza per dur a terme actualitzacions parcials en recursos existents, modificant només els camps especificats a la petició.

Aquesta diferència fa que `PATCH` sigui ideal per a situacions on només cal canviar alguns aspectes del recurs, sense haver d'enviar tot el contingut. Les peticions `PATCH` són més eficients en termes d'amplada de banda i precisió, ja que només transmeten i actualitzen les dades necessàries.

```javascript
fetch('https://jsonplaceholder.typicode.com/posts/12', {
  method: 'PATCH',
  headers: headers,
  body: JSON.stringify({ title: 'A NEW UPPERCASE TITLE HERE' }),
})
  .then((response) => response.json())
  .then((data) => console.log(data))
  .catch((error) => console.error('Error:', error));
```

Observem que la resposta que obtenim després de la petició `PATCH` és la mateixa que anteriorment hem obtingut després de la petició `PUT`, però en aquest cas només hem hagut d'enviar el camp `title` perquè el recurs s'actualitzi correctament. La resta de camps (l'identificador de l'usuari i el cos de l'article) s'han mantingut intactes.

> <br>🚨&nbsp; **Atenció**
>
> Tal com passa amb les peticions `PUT`, si executeu la petició `PATCH` anterior repetides vegades, el recurs s'actualizarà tantes vegades com peticions envieu, però no es crearan recursos duplicats. `PATCH`, igual que passa amb `PUT`, té naturalesa **idempotent**: fer la mateixa petició més d'una vegada no canvia res.
> <br><br>

La resposta de les peticions `PUT` i de les peticions `PATCH` inclou el recurs complet actualitzat. Vegem a continuació un altre tipus de petició que no segueix aquest patró.

### Peticions DELETE

Les peticions `DELETE` són l'últim tipus entre les cinc peticions bàsiques que es poden fer sobre una API REST. Com el seu nom indica, s'utilitzen per eliminar un recurs existent en el servidor, per la qual cosa exigeixen tenir precaució. A diferència de les peticions `POST`, `PUT` i `PATCH` vistes fins ara, les peticions `DELETE` no inclouen un cos amb les dades que es volen enviar al servidor, ja que no cal enviar cap dada per eliminar un recurs, simplement cal identificar el recurs que es vol eliminar.

```javascript
fetch('https://jsonplaceholder.typicode.com/posts/12', {
  method: 'DELETE',
})
  .then((response) => response.json())
  .then((data) => console.log('DELETED', data))
  .catch((error) => console.error('Error:', error));
```

> <br>🚨&nbsp; **Atenció**
>
> `DELETE` també és una operació **idempotent**: enviar diverses vegades seguides una petició per eliminar un determinat recurs no té cap efecte més enllà de la primera vegada que s'envia la petició. Així doncs, l'única petició que no és idempotent és `POST`.
> <br><br>

Hem de tenir present que enviar una petició de tipus `DELETE` no implica necessàriament que el recurs s'elimini a la base de dades del servidor. Serà decisió de l'equip de desenvolupament de l'API determinar si el recurs s'elimina físicament o no. En alguns casos, com passa amb l'API de JSONPlaceholder, els recursos no s'eliminen. Altres vegades es pot dur a terme el que es coneix com un esborrament lògic: el recurs es marca com a eliminat, però no s'elimina físicament de la base de dades. En qualsevol cas, el resultat vist des de l'exterior d'una petició `DELETE` hauria de ser que el recurs en qüestió ja no estarà disponible per llegir posteriorment.
