## Unitat 1: Fonaments d'interacció amb una API REST

A l'últim tema del nivell B1 vàrem tenir ocasió de presentar una introducció a AJAX i al treball amb l'API Fetch de JavaScript. En aquest tema aprofundirem en el treball amb API de tipus REST i veurem els diferents tipus de peticions que podem dur a terme en un servidor web.

> <br>🧐&nbsp; **Què significa _API_?**
> 
> En sentit estricte, API significa _Application Programming Interface_ (en català, _Interfície de Programació d'Aplicacions_). Una API defineix com els components de programari han d'interactuar entre si, permetent que diferents programes es comuniquin de manera eficient. Les API s'utilitzen àmpliament per permetre la integració entre diferents sistemes i serveis. De tota manera, a efectes pràctics el terme API té dos significats (o, si ho preferiu, matisos) lleugerament diferents. Quan parlem de "l'API Fetch de JavaScript" realment ens referim a una interfície de programació específica en JavaScript per dur a terme sol·licituds web. És a dir, a una part específica del llenguatge JavaScript. En canvi, quan parlem de "treball amb API", "connexió amb una API" o "consultes a una API" ens referim a interactuar amb interfícies de programació genèriques proporcionades per servidors web que ens permeten obtenir o enviar dades.
>
> Dit d'una altra manera més simple encara: API pot significar "una part concreta d'un llenguatge de programació" (com l'API Fetch) o bé "un accés controlat a una base de dades que resideix en un servidor web" (com l'[API dels pokémons](https://pokeapi.co) o l'[API de Star Wars](https://swapi.dev)). En aquest tema ens referirem a l'API amb aquest segon sentit.
> <br><br>

A mode de repàs, en el nivell B1 vam comentar que el navegador web pot fer peticions en segon pla per rebre o enviar dades sobre qualsevol temàtica: llistes de clients, registres meteorològics, resultats de partits de futbol o, en general, qualsevol tipus d'informació que el servidor tingui emmagatzemada. Quan diem que aquestes peticions es porten a terme en segon pla, el que volem dir és que no es produeixen com a conseqüència que l'usuari hagi actualitzat la pàgina en el seu navegador, sinó que es poden iniciar des del codi JavaScript de l'aplicació web en el moment que el programador hagi decidit fer-ho. Per exemple, quan hagin passat trenta segons des de la càrrega inicial de la pàgina o quan l'usuari premi un botó o passi el ratolí per sobre d'una imatge. En aquest cas, el navegador web no actualitza la pàgina, sinó que fa una petició HTTP al servidor per obtenir o enviar dades.

En l'escenari més habitual, el servidor al qual es fa aquesta petició HTTP és un servidor web que conté una base de dades amb la informació que es vol obtenir o enviar. Així mateix, aquest servidor té instal·lada una API de tipus REST que permet fer peticions HTTP per obtenir o enviar dades. En el cas que la petició HTTP sigui de tipus `GET`, l'API retornarà les dades sol·licitades, habitualment en format JSON. En el cas que la petició HTTP sigui de tipus `POST`, `PUT`, `DELETE` o `PATCH`, l'API processarà les dades enviades i portarà a terme la modificació corresponent de la base de dades.

### Principals característiques d'una API REST

Una API REST (o RESTful) és una API que segueix l'estil arquitectònic REST (Representational State Transfer) per al disseny de serveis web. Aquest estil el va definir Roy Fielding a la seva tesi doctoral de 2000 i es basa en els principis clau següents:

1. **Protocol client servidor** – El client que sol·licita els recursos i el servidor que els proporciona són entitats separades i independents. El client no sap com s'emmagatzemen les dades en el servidor i el servidor no sap com es mostren les dades en el client. Aquesta separació permet que el client i el servidor puguin evolucionar de forma independent.

2. **Sense estat** – Cada sol·licitud d'un client conté tota la informació necessària perquè el servidor la pugui entendre. Això significa que el servidor no necessita emmagatzemar informació sobre l'estat de la sessió del client. Per exemple, si el client envia una petició HTTP per obtenir les dades d'un pokémon, el servidor no necessita saber si aquest mateix client ja ha fet altres peticions anteriorment.

3. **Registrable** – Les respostes a les peticions HTTP es defineixen com a registrables o no registrables. Si són registrables, el client les pot emmagatzemar en cau per usar-les posteriorment. Això li permet reutilitzar la informació desada en comptes de tornar-la a sol·licitar al servidor.

4. **Interfície uniforme** – Les peticions HTTP han de seguir un conjunt de regles predefinides. Per exemple, les peticions de tipus `GET` han de retornar les dades sol·licitades, mentre que les peticions de tipus `POST` han de crear noves dades. Altres tipus de peticions també tenen la seva pròpia semàntica. Aquesta interfície uniforme també inclou principis com identificació de recursos, missatges autodescriptius i la inclusió d'enllaços per facilitar la navegació entre recursos.

5. **Sistema en capes** – L'arquitectura permet l'existència de capes intermèdies (com balancejadors de càrrega i cau), que milloren l'escalabilitat i la seguretat, sense afectar la interacció entre client i servidor. En realitat, un client no pot saber si està connectat directament al servidor o a un intermediari. Això permet que el servidor pugui escalar afegint més servidors intermedis.

Totes aquestes característiques fan de les API REST una opció molt popular per al desenvolupament d'aplicacions web modernes, per la simplicitat, escalabilitat i eficiència en la transferència de dades.

A més, tal com ja vàrem explicar en el nivell B1, avui dia el format més habitual per intercanviar dades entre diferents aplicacions i màquines és el format JSON. Aquest format JSON (acrònim de _JavaScript Object Notation_) consisteix en un format lleuger per desar i intercanviar informació. La sintaxi és molt bàsica: claudàtors `[ ]` per marcar arrays, claus `{ }` per delimitar objectes i parelles de clau i valor `"key": "value"` per definir totes les propietats dels objectes. Tot i que el nom ens pugui portar a l'error de creure que és un format exclusiu de JavaScript, la realitat és que JSON és un format estàndard per a l'intercanvi d'informació entre màquines que utilitzen llenguatges de programació molt diferents. 

La sinergia entre les API REST i l'ús del format JSON per a l'intercanvi de dades resulta molt efectiva: les API REST faciliten una arquitectura clara i escalable, mentre que JSON és un format ideal per a la transferència de dades. A les properes unitats explorarem com interactuar amb aquest tipus d'API utilitzant el mecanisme Fetch de JavaScript.

### Maneig de respostes i errors de l'API

El maneig eficaç d'errors i respostes d'una API és crucial per al desenvolupament robust d'aplicacions web. En interactuar amb qualsevol API REST, és essencial entendre i processar adequadament les respostes, incloent els codis d'estat HTTP i els missatges d'error.

Els codis d'estat HTTP són fonamentals per comprendre la resposta d'una API. Alguns dels més comuns inclouen els següents:

- `200 OK` – La sol·licitud va ser reeixida.
- `400 Bad Request` – La sol·licitud no es va poder processar a causa d'un error del client.
- `401 Unauthorized` – Falta autenticació per a la sol·licitud.
- `404 Not Found` – El recurs sol·licitat no es va trobar.
- `500 Internal Server Error` – Error genèric del servidor.
  
A la documentació de la MDN podeu consultar els [HTTP response status codes](https://developer.mozilla.org/en-US/docs/Web/HTTP/Status) amb detall, però de manera més genèrica és important recordar que qualsevol resposta amb un codi d'estat que comença per 2 indica que la sol·licitud va ser reeixida, mentre que qualsevol resposta amb un codi d'estat que comença per 4 o 5 indica que la sol·licitud va fallar: en el primer cas (4xx) a causa d'un error del client i en el segon cas (5xx) a causa d'un error del servidor. 

A més dels codis d'estat HTTP, les respostes d'una API també poden incloure missatges d'error detallats. Per exemple, si una sol·licitud de tipus `POST` no inclou un paràmetre requerit, l'API pot retornar un missatge d'error que indiqui que el paràmetre és obligatori. Aquests missatges d'error poden ser molt útils per identificar i solucionar problemes.

Així doncs, per a un adequat maneig de les respostes que ens pot donar una API i els possibles errors trobats hem de seguir una sèrie de pautes:

- **Validació de respostes** – Hem de verificar sempre el codi d'estat HTTP de la resposta per determinar si la sol·licitud ha estat reeixida o si hi ha haver algun error.

- **Maneig d'errors específics** – Diferents tipus d'errors exigeixen implementar diferents solucions. Per exemple, si rebem un error `401 Unauthorized` podem considerar redirigir l'usuari a una pàgina d'inici de sessió.

- **Missatges d'error descriptius** – Moltes API proporcionen missatges d'error detallats. Els hem d'analitzar per corregir els defectes en el nostre codi. També és bona idea registrar aquests missatges en algun sistema de logs per facilitar-ne la depuració posterior.

- **Reintents en cas d'error** – Si es produeixen errors temporals per part del servidor (com un `503 Service Unavailable`), no podem corregir res en el nostre codi, però sí que podem implementar una política de reintent automàtic de la petició transcorregut un temps prudencial.

- **Resposta a l'usuari** – No hem de traslladar a l'usuari final els codis o els missatges d'error de manera directa, sinó que convé informar sobre aquests errors de manera amigable i comprensible.

### Seguretat en les API REST

La seguretat és un aspecte primordial a l'hora de treballar amb les API REST. En particular, el procés d'autenticació i autorització de peticions resulta imprescindible per garantir que els usuaris només podan accedir a la informació que els correspon:

- L'autenticació s'encarrega de verificar la identitat de l'usuari.
- L'autorització determina els permisos de l'usuari. És a dir, a quins recursos pot accedir i quines operacions pot fer.
 
Si la nostra universitat té una API que permet a tots els docents accedir a la informació dels alumnes, és important que només puguin accedir a aquesta API els docents de la nostra universitat (procés d'autenticació) i també que cada docent només pugui accedir a les dades dels alumnes que li corresponen (procés d'autorització). Per tant, un usuari pot estar autenticat en un sistema, però només tenir permisos per accedir a certs recursos. 

En el cas d'una API REST, l'autenticació i autorització es poden portar a terme mitjançant l'ús de tokens de seguretat. Aquests tokens s'inclouen en les peticions HTTP per identificar l'usuari i determinar-ne els permisos. A l'última unitat d'aquest tema veurem com treballar amb tokens JWT (JSON Web Tokens) en JavaScript.

> <br>💫&nbsp; **Ampliació**

Tant JWT com OAuth ofereixen sistemes robustos per protegir les API REST i assegurar que només els usuaris autoritzats puguin accedir als recursos específics als quals tenen accés.

### Extensió Thunder Client per a Visual Studio Code

Tot i que treballarem els diferents tipus de peticions a una API des de JavaScript, en aquest tema us resultarà especialment útil disposar de l'extensió [Thunder Client](https://marketplace.visualstudio.com/items?itemName=rangav.vscode-thunder-client) instal·lada en el vostre Visual Studio Code. Aquesta extensió, igual que altres aplicacions com [Postman](https://www.postman.com), permeten fer peticions HTTP a una API de manera directa sense necessitat d'escriure cap codi. En el cas de Thunder Client, s'integra directament en l'entorn de desenvolupament de Visual Studio Code, i proporciona una interfície gràfica d'usuari intuïtiva per provar i depurar peticions a API.

Thunder Client, Postman i eines similars ens permeten als desenvolupadors construir, editar i enviar peticions HTTP de qualsevol tipus (incloent `GET`, `POST`, `PUT`, `DELETE` i `PATCH`) d'una manera senzilla i visualitzant les respostes de manera clara i organitzada. També suporten l'edició de capçaleres, paràmetres i cos de la petició, fet que ens facilita poder portar a terme una àmplia gamma de proves amb les API abans de començar a escriure codi.

A més, una de les característiques més interessants de Thunder Client és la capacitat per desar peticions i organitzar-les en coleccions, aspecte que ens permet millorar la nostra eficiència en treballar amb múltiples endpoints d'API.

> <br>💫&nbsp; **Ampliació**
> 
> Thunder Client, a més, permet exportar les peticions a codi JavaScript i a molts altres llenguatges de programació. Aquesta funcionalitat us pot resultar de gran utilitat per començar a traballar amb una API des del vostre codi.
